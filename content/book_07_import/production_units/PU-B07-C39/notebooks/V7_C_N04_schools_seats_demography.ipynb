## Decision contract
Translate demographic scenarios into classroom and staffing review. Projections do not automatically open/close schools, acquire land, hire staff, or assign learners.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7604);n=36
df=pd.DataFrame({'district':[f'E{i:02d}' for i in range(n)],'enrolment':rng.integers(2500,42000,n),'seats':rng.integers(3000,39000,n),'classrooms':rng.integers(80,1100,n),'annual_growth':rng.uniform(-.01,.065,n),'data_age':rng.integers(0,4,n)});df.head()
```

## Evidence and quality contract
Reconcile school census, population projections, enrolment definitions, private/nonformal provision, migration, grade repetition, attendance, classroom condition, and geographic accessibility.

```python
for label,mult in [('low',.65),('reference',1),('high',1.35)]: df[label]=df.enrolment*(1+df.annual_growth*mult)**5
df['seat_gap_ref']=(df.reference-df.seats).clip(lower=0);df['classroom_gap_ref']=np.ceil(df.seat_gap_ref/40);print(df[['seat_gap_ref','classroom_gap_ref']].describe().round(1))
```

## Uncertainty, sensitivity, and abstention
Use low/reference/high demographic scenarios. If the sign of the gap changes across scenarios, mark the decision as sensitive and stage investment.

```python
df['gap_low']=df.low-df.seats;df['gap_high']=df.high-df.seats;df['sensitive']=(df.gap_low*df.gap_high<0);df['status']=np.where(df.data_age>2,'ABSTAIN—REFRESH SCHOOL DATA',np.where(df.seat_gap_ref>0,np.where(df.sensitive,'STAGED CAPACITY REVIEW','CAPACITY REVIEW'),'MONITOR'));result=df[df.status!='MONITOR'][['district','seat_gap_ref','classroom_gap_ref','sensitive','data_age','status']].sort_values('seat_gap_ref',ascending=False);print(result.head(15).round(1).to_string(index=False))
```

## Decision product and authority boundary
The output is a review queue with reason codes, evidence age, uncertainty, accountable owner, and status. It never transfers legal, clinical, professional, procurement, enforcement, or budget authority to the model.
