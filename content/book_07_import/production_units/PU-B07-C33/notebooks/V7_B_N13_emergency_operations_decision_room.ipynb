## Decision contract
Create a time-stamped common operating picture for emergency coordination, verification, resource gaps, and accountable escalation. Analytics do not declare an emergency, issue clinical orders, or replace incident command.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7213); n=36; df=pd.DataFrame({'incident':[f'I{i:03d}' for i in range(n)],'severity':rng.integers(1,6,n),'verified':rng.choice([True,False],n,p=[.7,.3]),'latency_min':rng.integers(5,360,n),'beds_gap':rng.integers(-10,55,n),'supply_days':rng.uniform(.5,12,n),'owner':rng.choice(['EOC','LAB','LOGISTICS','REGION'],n)}); df.head()
```

## Evidence and quality contract
Define incident, location, event time, verification status, severity, affected population, facility status, beds, staff, commodities, transport, laboratory evidence, reporting latency, source authority, action owner, and incident log.


```python
df['evidence_ok']=df.verified&(df.latency_min<=180); df['pressure']=df.beds_gap.clip(lower=0)+10*(df.supply_days<3)+8*(df.severity>=4); print(df[['severity','verified','latency_min','pressure']].describe().round(2))
```

## Uncertainty, sensitivity, and abstention
Early reports are incomplete, duplicated, and rapidly revised. Separate observed, reported, inferred, and forecast values; compare scenarios and abstain from precise claims when verification is weak.


```python
df['status']=np.where(~df.evidence_ok,'VERIFY URGENTLY',np.where((df.severity>=4)|(df.pressure>=20),'INCIDENT COMMAND REVIEW','MONITOR')); queue=df[df.status!='MONITOR']; print(queue.head(20).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
