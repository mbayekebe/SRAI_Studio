## Decision contract
Support housing supply, infrastructure, affordability, and programme scenarios. Owners: housing, planning, finance, and statistical authorities. Outputs do not determine eviction, eligibility, rent, taxation, or land acquisition automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7501);areas=pd.DataFrame({'area':[f'A{i:02d}' for i in range(36)],'households':rng.integers(1800,18000,36),'housing_units':rng.integers(1600,16500,36),'annual_growth':rng.uniform(.015,.06,36),'median_income':rng.integers(180000,720000,36),'median_rent':rng.integers(22000,115000,36),'vacancy_rate':rng.uniform(.02,.16,36),'informal_share':rng.uniform(.05,.62,36)});areas.head().round(2)
```

## Evidence contract
Households, dwellings, occupancy, tenure, rent, income, quality, informality, homelessness, geography, and reference periods are distinct concepts. Administrative records alone may omit vulnerable groups.

```python
assert areas.area.is_unique;areas['unit_gap']=(areas.households-areas.housing_units).clip(lower=0);areas['rent_burden']=12*areas.median_rent/areas.median_income;print(areas[['unit_gap','rent_burden','vacancy_rate']].describe().round(2))
```

## Housing need is multidimensional
A unit gap, high rent burden, low vacancy, poor quality, overcrowding, insecure tenure, and service deficits require different responses.

```python
areas['burden_gap']=(areas.rent_burden-.30).clip(lower=0);areas['scarcity']=(.06-areas.vacancy_rate).clip(lower=0)/.06;areas['need_index']=.4*(areas.unit_gap/areas.households)+.35*areas.burden_gap.clip(upper=1)+.15*areas.scarcity+.10*areas.informal_share;print(areas.nlargest(8,'need_index')[['area','unit_gap','rent_burden','vacancy_rate','informal_share','need_index']].round(2).to_string(index=False))
```

## Projection scenarios
Project household growth under explicit assumptions. These are planning scenarios, not forecasts of individual behaviour.

```python
sc=[]
for label,m in [('lower',.75),('reference',1),('upper',1.25)]:
 hh=areas.households*(1+areas.annual_growth*m)**5;gap=(hh-areas.housing_units).clip(lower=0).sum();sc.append({'scenario':label,'five_year_households':round(hh.sum()),'aggregate_unit_gap':round(gap)})
print(pd.DataFrame(sc).to_string(index=False))
```

## Policy options are not interchangeable
New construction, serviced land, rental support, upgrading, vacancy measures, and transport improvements affect different dimensions and groups.

```python
options=pd.DataFrame({'option':['new supply','settlement upgrading','rental support','transport access'],'primary_target':['unit gap','services/security','rent burden','location constraint'],'main_risk':['cost/delay','displacement','price capture','induced demand'],'required_authority':['housing/planning','municipal/community','housing/finance','transport/planning']});print(options.to_string(index=False))
```

## Equity and prohibited-use controls
Areas with stale or weak population evidence require verification. Never publish household locations or turn “informal share” into an enforcement score.

```python
areas['data_quality']=rng.uniform(.7,1,36);areas['disposition']=np.where(areas.data_quality<.8,'ABSTAIN—VERIFY DATA',np.where(areas.need_index>=areas.need_index.quantile(.8),'POLICY OPTIONS REVIEW','MONITOR'));product={'review_areas':areas[areas.disposition!='MONITOR'].area.tolist(),'prohibited_uses':['eviction targeting','automatic eligibility denial','rent setting','household identification'],'status':'SCENARIO—AUTHORITY AND COMMUNITY REVIEW'};print(product)
```
