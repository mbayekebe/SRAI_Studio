## Decision contract
Evaluate whether an education programme changed learning, participation, or equity and whether implementation explains results. Scores are evidence within a design; they do not rank human worth or justify automatic continuation or termination.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7314); schools=30; rows=[]
for s in range(schools):
 treated=s%2; school_effect=rng.normal(0,3)
 for i in range(18): rows.append((s,treated,rng.normal(50,9),school_effect))
df=pd.DataFrame(rows,columns=['school','treated','baseline','school_effect']); df['followup']=df.baseline+1.5+2.2*df.treated+df.school_effect+rng.normal(0,5,len(df)); df.head()
```

## Evidence and quality contract
Specify intervention, theory of change, assignment, comparison, baseline equivalence, test scale, participation, implementation fidelity, teacher and school clustering, spillovers, attrition, accommodations, costs, harms, and decision rule.


```python
df['gain']=df.followup-df.baseline; school=df.groupby(['school','treated']).gain.mean().reset_index(); effect=school.groupby('treated').gain.mean().diff().iloc[-1]; se=np.sqrt(sum(school.groupby('treated').gain.var()/school.groupby('treated').size())); print({'cluster_level_effect':round(effect,3),'approx_se':round(se,3)})
```

## Uncertainty, sensitivity, and abstention
Test noise, teaching to the test, clustering, attrition, and concurrent reforms can distort impact. Use design-consistent inference, multiple outcomes cautiously, subgroup pre-specification, and sensitivity checks.


```python
school['status']=np.where(abs(school.gain)>10,'IMPLEMENTATION/MEASUREMENT REVIEW','INCLUDE'); print(school[school.status!='INCLUDE'].round(2).to_string(index=False)); decision='EVALUATION BOARD REVIEW' if np.isfinite(effect) else 'ABSTAIN'
```
