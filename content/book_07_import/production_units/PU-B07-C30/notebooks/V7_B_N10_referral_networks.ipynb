## Decision contract
Identify aggregate referral delays and network bottlenecks for service redesign. Outputs do not change clinical priority, direct individual care, penalize staff, or expose patient trajectories.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7910);n=260;df=pd.DataFrame({'route':rng.choice(['Clinic-Hospital','Hospital-Centre','Community-Clinic'],n),'urgent':rng.choice([0,1],n,p=[.75,.25]),'travel_min':rng.gamma(3,25,n),'wait_min':rng.gamma(2,35,n),'completed':rng.choice([0,1],n,p=[.14,.86]),'timestamps_complete':rng.choice([0,1],n,p=[.12,.88])});df.head()
```

## Evidence and quality contract
Define referral reason, sending/receiving facility, timestamps, transport, acceptance, arrival, completion, counter-referral, capacity, urgency, outcome, missing events, and privacy.

```python
df['total_delay']=df.travel_min+df.wait_min;summary=df.groupby('route').agg(median_delay=('total_delay','median'),completion=('completed','mean'),timestamp_quality=('timestamps_complete','mean'),n=('route','size')).reset_index();print(summary.round(2).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Missing completion may mean loss, care elsewhere, or recording failure. Compare thresholds and abstain where timestamp completeness is inadequate.

```python
summary['status']=np.where(summary.timestamp_quality<.8,'ABSTAIN—IMPROVE REFERRAL DATA',np.where((summary.median_delay>120)|(summary.completion<.8),'NETWORK BOTTLENECK REVIEW','MONITOR'));result=summary;print(result.round(2).to_string(index=False));print({'delay90':int((df.total_delay>90).sum()),'delay120':int((df.total_delay>120).sum())})
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
