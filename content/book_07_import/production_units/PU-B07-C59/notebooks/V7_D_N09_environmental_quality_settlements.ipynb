## Decision contract
Support area-level heat, air-quality, and flood-exposure planning. Outputs do not diagnose residents, determine insurance/eligibility, stigmatize settlements, or authorize displacement.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7809);n=50;df=pd.DataFrame({'area':[f'Q{i:02d}' for i in range(n)],'population':rng.integers(1000,70000,n),'pm25':rng.uniform(8,85,n),'heat_days':rng.integers(2,90,n),'flood_index':rng.uniform(0,1,n),'sensor_coverage':rng.uniform(.35,1,n)});df.head()
```

## Evidence and quality contract
Record sensor/model, pollutant/hazard, calibration, time resolution, missingness, spatial support, population, indoor/outdoor distinction, vulnerability, uncertainty, and intervention context.

```python
df['pm_gap']=(df.pm25-15).clip(lower=0)/70;df['heat_gap']=df.heat_days/90;df['exposure_index']=.4*df.pm_gap+.3*df.heat_gap+.3*df.flood_index;df['quality']=df.sensor_coverage>=.65;weighted=(df.exposure_index*df.population).sum()/df.population.sum();print('population-weighted exposure index',round(weighted,3))
```

## Uncertainty, sensitivity, and abstention
Sensors and modeled surfaces have error; exposure is not dose or disease. Compare thresholds and require field/calibration review where coverage is weak.

```python
df['status']=np.where(~df.quality,'ABSTAIN—IMPROVE MONITORING',np.where(df.exposure_index>=df.exposure_index.quantile(.75),'ENVIRONMENTAL OPTIONS REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['area','population','pm25','heat_days','flood_index','sensor_coverage','exposure_index','status']];print(result.head(18).round(2).to_string(index=False));print({'threshold70':int((df.exposure_index>=df.exposure_index.quantile(.7)).sum()),'threshold80':int((df.exposure_index>=df.exposure_index.quantile(.8)).sum())})
```

## Decision product and authority boundary
The output is a documented review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It never replaces legal, professional, clinical, procurement, enforcement, budget, safeguarding, or ethical authority.
