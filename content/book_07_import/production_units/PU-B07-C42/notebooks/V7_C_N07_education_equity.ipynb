## Decision contract
Reveal disparities in access and completion for policy review. Outputs do not label individuals, determine admission, allocate scholarships automatically, or stigmatize communities.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7707);groups=['Urban boys','Urban girls','Rural boys','Rural girls','Displaced','Disability'];n=np.array([850,900,720,760,120,85]);p=np.array([.82,.86,.61,.55,.43,.38]);y=rng.binomial(n,p);df=pd.DataFrame({'group':groups,'n':n,'completed':y});df
```

## Evidence and quality contract
Define outcome, denominator, age/grade, geography, sex, disability, wealth, language, displacement, survey design, missingness, sample size, disclosure, and intersectional analysis.

```python
df['rate']=df.completed/df.n;df['se']=np.sqrt(df.rate*(1-df.rate)/df.n);df['lower']=df.rate-1.96*df.se;df['upper']=df.rate+1.96*df.se;best=df.rate.max();df['absolute_gap']=best-df.rate;df['relative_ratio']=df.rate/best;print(df.round(3).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Group differences have sampling and measurement uncertainty. Suppress unstable cells, compare absolute and relative gaps, and avoid causal claims from descriptive data.

```python
df['stable']=df.n>=100;df['status']=np.where(~df.stable,'ABSTAIN—MORE/POOLED EVIDENCE',np.where(df.upper<.65,'EQUITY BARRIER REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['group','n','rate','lower','upper','absolute_gap','relative_ratio','status']];print(result.round(3).to_string(index=False))
```
