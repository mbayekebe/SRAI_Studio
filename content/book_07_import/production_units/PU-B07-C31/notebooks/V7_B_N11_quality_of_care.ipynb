## Decision contract
Combine process, outcome, safety, and experience indicators for clinical-governance review. The system does not diagnose, discipline clinicians, or compare facilities without case-mix and data-quality review.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7911);n=35;df=pd.DataFrame({'facility':[f'Q{i:02d}' for i in range(n)],'eligible':rng.integers(18,650,n),'process_rate':rng.uniform(.45,.98,n),'outcome_rate':rng.uniform(.55,.98,n),'experience':rng.uniform(.4,.95,n),'case_mix':rng.uniform(.7,1.4,n),'coverage':rng.uniform(.55,1,n)});df.head()
```

## Evidence and quality contract
Define eligible population, guideline process, outcome window, adverse event, experience measure, case mix, transfers, deaths after discharge, reporting coverage, and audit method.

```python
df['adjusted_outcome']=(df.outcome_rate/df.case_mix).clip(upper=1);df['quality_index']=.4*df.process_rate+.4*df.adjusted_outcome+.2*df.experience;df['stable']=(df.eligible>=50)&(df.coverage>=.75);print(df.quality_index.describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Small volumes and case mix distort comparisons. Use intervals, risk adjustment, minimum denominators, and abstention; never interpret ranks as proof of causality.

```python
df['status']=np.where(~df.stable,'ABSTAIN—MORE/COMPLETE EVIDENCE',np.where(df.quality_index<.65,'CLINICAL GOVERNANCE REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['facility','eligible','process_rate','outcome_rate','case_mix','coverage','quality_index','status']];print(result.head(16).round(3).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
