## Decision contract
Reconcile school and household survey evidence to estimate participation and exclusion. The notebook supports official-statistics review; it does not label individual children, replace frames, or authorize targeting from one source alone.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7312); n=480; df=pd.DataFrame({'region':rng.choice(list('WXYZ'),n),'weight':rng.uniform(.5,2.8,n),'enrolled':rng.binomial(1,.79,n),'attended':rng.binomial(1,.73,n),'frame_match':rng.binomial(1,.9,n),'responded':rng.binomial(1,.94,n)}); df.head()
```

## Evidence and quality contract
Define school, learner, household, age, grade, attendance, enrolment, disability, location, frame coverage, sample design, weights, nonresponse, seasonal timing, displaced and mobile populations, linkage rules, and confidentiality.


```python
d=df[df.responded==1]; out=d.groupby('region').apply(lambda g:pd.Series({'enrol_rate':np.average(g.enrolled,weights=g.weight),'attend_rate':np.average(g.attended,weights=g.weight),'frame_match':np.average(g.frame_match,weights=g.weight),'n_eff':g.weight.sum()**2/g.weight.pow(2).sum()}),include_groups=False); print(out.round(3))
```

## Uncertainty, sensitivity, and abstention
EMIS omits those outside registered schools, while household surveys face recall and sample limits. Triangulate sources, report intervals and coverage gaps, and abstain on unsupported small domains.


```python
out['status']=np.where((out.n_eff<70)|(out.frame_match<.85),'ABSTAIN - FRAME/SAMPLE REVIEW','OFFICIAL RECONCILIATION REVIEW'); print(out[['enrol_rate','attend_rate','frame_match','n_eff','status']].round(3))
```
