## Decision contract
Support verification and preparedness for climate-sensitive disease patterns. Outputs do not diagnose patients, prove causality, or trigger public-health restrictions automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7808);weeks=pd.date_range('2025-01-05',periods=80,freq='W');rain=np.maximum(0,60+45*np.sin(np.arange(80)*2*np.pi/52)+rng.normal(0,12,80));cases=np.maximum(0,15+.18*np.roll(rain,3)+rng.normal(0,6,80)).astype(int);df=pd.DataFrame({'week':weeks,'rain':rain,'cases':cases,'completeness':rng.uniform(.7,1,80)});df.head()
```

## Evidence and quality contract
Align cases, tested population, syndrome/diagnosis, location, onset/report dates, rainfall, temperature, vector evidence, population, interventions, reporting changes, and lags.

```python
for lag in [2,3,4,5]:
 df[f'rain_lag{lag}']=df.rain.shift(lag)
corr={lag:df.cases.corr(df[f'rain_lag{lag}']) for lag in [2,3,4,5]}
best=max(corr,key=corr.get);df['baseline']=df.cases.shift(1).rolling(12,min_periods=8).median();df['signal']=df.cases>df.baseline*1.5;print('lag correlations',corr,'best',best)
```

## Uncertainty, sensitivity, and abstention
Lag choice and surveillance completeness change signals. Compare plausible lags and abstain on weak reporting; correlation is not causal attribution.

```python
df['status']=np.where(df.completeness<.8,'ABSTAIN—VERIFY SURVEILLANCE',np.where(df.signal,'EPIDEMIOLOGIC AND FIELD VERIFICATION','MONITOR'));result=df[df.status!='MONITOR'][['week','rain','cases','completeness','status']];print(result.tail(15).round(2).to_string(index=False))
```
