## Decision contract
Integrate housing, land-use, informal-settlement services, disaster shelter, basic services, infrastructure, mobility, tenure, environment, building safety, homelessness support, frames, urban finance, and evaluation into a governed habitat decision service. Outputs support inclusive planning; they never automate enforcement, displacement, eligibility denial, permit approval, or safety certification.


```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7415); areas=[f'Z{i:02d}' for i in range(20)]; df=pd.DataFrame({'area':areas,'housing_burden':rng.uniform(.15,.62,20),'service_gap':rng.uniform(0,.65,20),'hazard_exposure':rng.uniform(0,1,20),'infra_condition':rng.uniform(.3,1,20),'mobility_gap':rng.uniform(0,.55,20),'frame_quality':rng.uniform(.62,1,20),'field_age_months':rng.integers(2,42,20),'implementation_capacity':rng.uniform(.3,1,20),'displacement_risk':rng.uniform(0,.7,20)}); df.head()
```

## Evidence and quality contract
Define people, households, dwellings, buildings, parcels, addresses, services, infrastructure, mobility, hazards, environment, tenure claims, programme areas, finance, source dates, spatial resolution, consent, participation, field verification, professional authority, appeals, versions, incidents, and retirement.


```python
df['evidence_ok']=(df.frame_quality>=.8)&(df.field_age_months<=24); df['planning_need']=.25*df.housing_burden+.25*df.service_gap+.20*df.hazard_exposure+.15*(1-df.infra_condition)+.15*df.mobility_gap; df['rights_gate']=df.displacement_risk<.45; print(df[['planning_need','frame_quality','field_age_months','displacement_risk']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Imagery, sensors, administrative records, addresses, informal supply, mobile populations, and place-based comparisons are incomplete. Show spatial and temporal uncertainty, validate in the field and with communities, audit displacement, and abstain from punitive or unsupported inference.


```python
df['decision_status']=np.where(~df.evidence_ok,'ABSTAIN - FIELD VERIFICATION',np.where(~df.rights_gate,'RIGHTS / PARTICIPATION REVIEW',np.where((df.planning_need>=.48)&(df.implementation_capacity>=.5),'INCLUSIVE PLANNING REVIEW',np.where(df.planning_need>=.48,'CAPACITY / FINANCE SCENARIO','MONITOR')))); brief=df[df.decision_status!='MONITOR'].sort_values('planning_need',ascending=False); model_card={'purpose':'area-level inclusive planning review','prohibited':['enforcement targeting','displacement','eligibility denial','safety certification'],'field_rule':'community and professional verification','owner':'Habitat intelligence board'}; audit={'areas':len(df),'queued':len(brief),'rights_reviews':int((df.decision_status=='RIGHTS / PARTICIPATION REVIEW').sum())}; print(brief.round(3).to_string(index=False)); print(model_card,audit)
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
