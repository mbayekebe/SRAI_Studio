## Decision contract
Prioritize permit and inspection workload using transparent risk factors. Outputs do not certify safety, approve permits, close buildings, impose penalties, or select contractors.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7910);n=75;df=pd.DataFrame({'permit':[f'P{i:03d}' for i in range(n)],'occupancy':rng.integers(10,2500,n),'storeys':rng.integers(1,18,n),'hazard':rng.uniform(0,1,n),'documents_complete':rng.choice([0,1],n,p=[.16,.84]),'prior_issues':rng.integers(0,5,n)});df.head()
```

## Evidence and quality contract
Define permit type, design/use, occupancy, height, materials, site hazards, code version, professional submissions, inspections, violations, corrections, completion, and appeal.

```python
df['risk']=.3*(df.storeys/18)+.25*df.hazard+.25*(df.prior_issues/4)+.2*np.log1p(df.occupancy)/np.log(2501);df['quality']=df.documents_complete==1;print(df.risk.describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Administrative features are incomplete proxies. Compare thresholds, abstain on missing critical documents, and require qualified engineering/building-control inspection.

```python
df['status']=np.where(~df.quality,'ABSTAIN—COMPLETE TECHNICAL FILE',np.where(df.risk>=df.risk.quantile(.75),'QUALIFIED INSPECTION REVIEW','STANDARD PROCESS'));result=df[df.status!='STANDARD PROCESS'][['permit','occupancy','storeys','hazard','prior_issues','risk','status']].sort_values('risk',ascending=False);print(result.head(18).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
