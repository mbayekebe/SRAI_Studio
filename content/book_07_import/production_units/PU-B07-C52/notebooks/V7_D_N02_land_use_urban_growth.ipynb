## Decision contract
Monitor land-cover and urban-growth patterns for planning verification. Classification never establishes legal tenure, ownership, violation, displacement eligibility, or enforcement priority.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7602);n=900
df=pd.DataFrame({'zone':rng.choice(['Metro','Periphery','Corridor'],n),'cover_t0':rng.choice(['built','vegetation','bare','water'],n,p=[.28,.42,.25,.05]),'confidence':rng.uniform(.45,1,n)});change=rng.random(n)<.16;df['cover_t1']=df.cover_t0;df.loc[change,'cover_t1']=rng.choice(['built','vegetation','bare'],change.sum());df['pixel_ha']=2.5;df.head()
```

## Evidence and quality contract
Hold sensor/product, dates/seasons, resolution, legend, processing, boundaries, confidence, and validation design constant. Apparent change may be phenology, cloud, registration, or classification error.

```python
df['urban_gain']=(df.cover_t0!='built')&(df.cover_t1=='built');summary=df.groupby('zone').agg(pixels=('zone','size'),urban_gain_ha=('urban_gain',lambda x:2.5*x.sum()),mean_confidence=('confidence','mean')).reset_index();print(summary.round(2).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Compare confidence cutoffs of 0.6 and 0.75. Low-confidence change is abstained and routed to imagery/field review.

```python
g=df[df.urban_gain].copy();g['status']=np.where(g.confidence<.65,'ABSTAIN—VERIFY CLASSIFICATION','PLANNING CHANGE REVIEW');result=g.groupby(['zone','status']).agg(area_ha=('pixel_ha','sum'),pixels=('pixel_ha','size')).reset_index();print(result.to_string(index=False));print({'gain_conf60_ha':float(g.loc[g.confidence>=.6,'pixel_ha'].sum()),'gain_conf75_ha':float(g.loc[g.confidence>=.75,'pixel_ha'].sum())})
```
