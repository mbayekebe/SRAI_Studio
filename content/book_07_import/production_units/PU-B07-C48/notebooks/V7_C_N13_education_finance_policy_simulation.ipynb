## Decision contract
Compare transparent education-finance allocation scenarios for adequacy, equity, and feasibility. The model informs budget deliberation; it does not appropriate funds, close schools, or replace political and legal authority.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7313); n=40; df=pd.DataFrame({'district':[f'D{i:02d}' for i in range(n)],'learners':rng.integers(1800,24000,n),'need_index':rng.uniform(.7,1.5,n),'remote':rng.binomial(1,.35,n),'execution':rng.uniform(.65,.98,n)}); unit_cost=420; budget=3_100_000; df.head()
```

## Evidence and quality contract
Define learners, schools, grades, needs, unit costs, wage and non-wage components, remoteness, poverty, disability and language supports, capital needs, ceilings, transfers, execution rates, price basis, fiscal year, and formula governance.


```python
df['adequacy_need']=df.learners*unit_cost*df.need_index; df['equity_need']=df.adequacy_need*(1+.12*df.remote); scale=min(1,budget/df.equity_need.sum()); df['scenario_allocation']=df.equity_need*scale; print({'need':round(df.equity_need.sum()),'budget':budget,'scale':round(scale,3)})
```

## Uncertainty, sensitivity, and abstention
Unit costs, enrolment, inflation, and implementation capacity vary. Show baseline and alternative formulas, sensitivity ranges, distributional effects, and unfunded needs instead of a single optimal allocation.


```python
df['implementation_flag']=df.execution<.75; df['unfunded']=df.equity_need-df.scenario_allocation; print(df.nlargest(12,'unfunded')[['district','learners','need_index','remote','scenario_allocation','unfunded','implementation_flag']].round(1).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
