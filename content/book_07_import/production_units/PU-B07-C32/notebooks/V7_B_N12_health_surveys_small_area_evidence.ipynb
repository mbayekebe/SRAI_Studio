## Decision contract
Produce design-aware health estimates and identify domains needing more evidence. Survey and model-assisted estimates support planning review; they do not diagnose individuals or override official statistical methods.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7212); n=420; df=pd.DataFrame({'domain':rng.choice(list('ABCDE'),n),'weight':rng.uniform(.4,3,n),'need':rng.binomial(1,.28,n),'responded':rng.binomial(1,.91,n)}); df.head()
```

## Evidence and quality contract
Define target population, frame, strata, clusters, weights, nonresponse, questionnaire, indicator numerator and denominator, domain, effective sample size, variance method, auxiliary data, linkage authority, and disclosure controls.


```python
d=df[df.responded==1].copy(); est=d.groupby('domain').apply(lambda g: np.average(g.need,weights=g.weight),include_groups=False).rename('weighted_rate'); raw=d.groupby('domain').need.mean().rename('raw_rate'); out=pd.concat([est,raw],axis=1); out['n']=d.groupby('domain').size(); out['n_eff']=d.groupby('domain').weight.apply(lambda w:w.sum()**2/(w.pow(2).sum())); print(out.round(3))
```

## Uncertainty, sensitivity, and abstention
Small samples and model borrowing can create false precision. Report design-based intervals, effective sample size, diagnostics, sensitivity to model assumptions, and abstention for unsupported domains.


```python
out['se']=np.sqrt(out.weighted_rate*(1-out.weighted_rate)/out.n_eff); out['status']=np.where(out.n_eff<45,'ABSTAIN - INSUFFICIENT DOMAIN EVIDENCE','OFFICIAL SURVEY REVIEW'); print(out[['weighted_rate','se','n_eff','status']].round(3))
```
