## Decision contract
Audit education-management information for planning and follow-up. The notebook does not rank teachers, punish schools, exclude learners, or replace census verification.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7706);n=90;df=pd.DataFrame({'school':[f'S{i:03d}' for i in range(n)],'enrolment':rng.integers(80,2600,n),'teachers':rng.integers(3,90,n),'classrooms':rng.integers(2,55,n),'boys':rng.integers(30,1300,n),'girls':rng.integers(30,1300,n),'report_age':rng.integers(0,25,n)});df.loc[:4,'boys']=df.loc[:4,'enrolment'];df.head()
```

## Evidence and quality contract
Define school registry, ownership, level, location, operational status, enrolment by grade/sex/age, teachers, classrooms, facilities, reporting date, duplicates, and revision history.

```python
df['sex_total_gap']=(df.boys+df.girls-df.enrolment).abs()/df.enrolment;df['pupil_teacher']=df.enrolment/df.teachers;df['pupil_classroom']=df.enrolment/df.classrooms;df['quality']=(df.report_age<=14)&(df.sex_total_gap<=.05);print(df[['sex_total_gap','pupil_teacher','pupil_classroom']].describe().round(1))
```

## Uncertainty, sensitivity, and abstention
Outliers may reflect genuine conditions or errors. Compare tolerance bands and abstain on missing/late reports instead of imputing silently.

```python
df['status']=np.where(df.report_age>14,'ABSTAIN—LATE REPORT',np.where(df.sex_total_gap>.05,'VERIFY ENROLMENT TOTALS',np.where((df.pupil_teacher>70)|(df.pupil_classroom>80),'CAPACITY DATA REVIEW','MONITOR')));result=df[df.status!='MONITOR'][['school','enrolment','sex_total_gap','pupil_teacher','pupil_classroom','report_age','status']];print(result.head(18).round(2).to_string(index=False));print({'gap5pct':int((df.sex_total_gap>.05).sum()),'gap10pct':int((df.sex_total_gap>.1).sum())})
```
