## Decision contract
Identify examination patterns requiring protected procedural review. No statistical signal accuses a learner, teacher, school, or centre or determines sanctions or scores.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7911);n=70;df=pd.DataFrame({'centre':[f'X{i:03d}' for i in range(n)],'candidates':rng.integers(18,420,n),'mean_score':rng.normal(55,9,n),'answer_similarity':rng.uniform(.05,.45,n),'incident_reports':rng.poisson(.4,n),'data_complete':rng.choice([0,1],n,p=[.08,.92])});df.loc[:2,'answer_similarity']=[.86,.81,.77];df.head()
```

## Evidence and quality contract
Preserve exam form, subject, centre, candidate counts, item responses, seating/time, accommodations, administration incidents, marking, moderation, appeals, and chain of custody.

```python
z=(df.mean_score-df.mean_score.mean())/df.mean_score.std();df['multi_signal']=(z.abs()>2).astype(int)+(df.answer_similarity>.7).astype(int)+(df.incident_reports>=2).astype(int);print(df.multi_signal.value_counts().sort_index().to_string())
```

## Uncertainty, sensitivity, and abstention
Rare patterns occur by chance and centres differ. Use multiple evidence, control false discoveries, protect identities, and require due process.

```python
df['status']=np.where(df.data_complete==0,'ABSTAIN—COMPLETE EXAM RECORD',np.where(df.multi_signal>=2,'CONFIDENTIAL PROCEDURAL REVIEW','NO STATISTICAL ESCALATION'));result=df[df.status!='NO STATISTICAL ESCALATION'][['centre','candidates','mean_score','answer_similarity','incident_reports','multi_signal','status']];print(result.round(2).to_string(index=False));print({'one_signal':int((df.multi_signal>=1).sum()),'two_signals':int((df.multi_signal>=2).sum())})
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
