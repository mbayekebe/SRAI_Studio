## Decision contract
Support aggregate service-capacity and pathway coordination for people experiencing homelessness. No output ranks deservingness, denies shelter, predicts criminality, or exposes locations or identities.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7911);services=['Emergency shelter','Transitional housing','Permanent support','Outreach'];df=pd.DataFrame({'service':services,'requests':[920,430,310,1200],'capacity':[610,280,145,520],'median_wait_days':[1,18,95,4],'accessible_share':[.62,.55,.48,.7],'data_coverage':[.82,.77,.68,.58]});df
```

## Evidence and quality contract
Use consented, purpose-limited records; define housing situation, service request, vulnerability needs, household composition, accessibility, capacity, wait, referral, outcome, repeat episodes, and privacy.

```python
df['capacity_gap']=(df.requests-df.capacity).clip(lower=0);df['served_ratio']=df.capacity/df.requests;df['quality']=df.data_coverage>=.7;print(df.round(2).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Service records exclude people who do not seek or cannot reach services. Report coverage and capacity scenarios; abstain from population prevalence or individual eligibility claims.

```python
df['status']=np.where(~df.quality,'ABSTAIN—IMPROVE SERVICE COVERAGE',np.where((df.served_ratio<.7)|(df.accessible_share<.6),'CAPACITY AND ACCESSIBILITY REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['service','requests','capacity','capacity_gap','median_wait_days','accessible_share','data_coverage','status']];print(result.round(2).to_string(index=False))
```
