## Decision contract
Support transparent textbook and input allocation under limited stock. The model does not bypass curriculum approval, procurement, financial controls, or local receipt verification.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7605);n=45
df=pd.DataFrame({'school':[f'S{i:02d}' for i in range(n)],'enrolment':rng.integers(120,2200,n),'usable_books':rng.integers(30,1800,n),'loss_rate':rng.uniform(.01,.16,n),'lead_weeks':rng.integers(2,18,n),'remoteness':rng.uniform(0,1,n),'report_age':rng.integers(0,20,n)});df.head()
```

## Evidence and quality contract
Match title, curriculum, language, grade, edition, condition, unit, school, enrolment date, stock date, pipeline, losses, and delivery confirmation.

```python
df['need']=(df.enrolment-df.usable_books).clip(lower=0);df['quality']=df.report_age<=10;df['priority']=df.need*(1+.25*df.remoteness+.2*(df.lead_weeks/18));available=18000;valid=df[df.quality].copy();valid['allocation']=np.floor(available*valid.priority/valid.priority.sum()).astype(int);valid['allocation']=valid[['allocation','need']].min(axis=1);print(valid[['need','allocation']].sum().to_string())
```

## Uncertainty, sensitivity, and abstention
Compare proportional-to-need and equity-weighted allocations. Stale inventories abstain. Residual stock requires a documented second allocation round.

```python
base=np.floor(available*valid.need/valid.need.sum()).astype(int);valid['base_allocation']=np.minimum(base,valid.need);delta=(valid.allocation-valid.base_allocation).abs().sum();valid['status']=np.where(valid.need>0,'ALLOCATION AND DELIVERY REVIEW','MONITOR');stale=df[~df.quality].assign(status='ABSTAIN—VERIFY INVENTORY');result=pd.concat([valid[valid.status!='MONITOR'][['school','need','allocation','lead_weeks','report_age','status']],stale.assign(allocation=0)[['school','need','allocation','lead_weeks','report_age','status']]]);print(result.head(15).to_string(index=False));print('allocation_sensitivity_units',int(delta))
```
