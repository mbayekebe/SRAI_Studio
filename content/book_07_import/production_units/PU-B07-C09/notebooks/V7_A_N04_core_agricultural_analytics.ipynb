## Decision contract
Compute transparent production, yield, area, structure, and risk indicators for official review. Indicators do not become forecasts, causal claims, subsidies, or budget allocations automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7904);n=500;df=pd.DataFrame({'region':rng.choice(['N','C','S','E'],n),'area':rng.gamma(2,2,n),'yield_t_ha':rng.normal(3.2,.9,n).clip(.2,8),'weight':rng.uniform(.5,4,n)});df['production']=df.area*df.yield_t_ha;df.head()
```

## Evidence and quality contract
Define farm/holding, parcel, crop, area concept, production, losses, livestock units, prices, population, weights, domains, reference periods, and confidentiality.

```python
def stats(g):
 w=g.weight.to_numpy();y=g.production.to_numpy();a=g.area.to_numpy();return pd.Series({'production_est':(w*y).sum(),'area_est':(w*a).sum(),'ratio_yield':(w*y).sum()/(w*a).sum(),'n_eff':w.sum()**2/(w*w).sum()})
res=df.groupby('region').apply(stats,include_groups=False).reset_index();res['unweighted_yield']=df.groupby('region').apply(lambda g:g.production.sum()/g.area.sum(),include_groups=False).to_numpy();print(res.round(2).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Weighted estimates require sampling variance and small-domain controls. Compare weighted/unweighted results and abstain on low effective sample size.

```python
res['difference']=res.ratio_yield-res.unweighted_yield;res['status']=np.where(res.n_eff<80,'ABSTAIN—SMALL EFFECTIVE SAMPLE',np.where(res.difference.abs()>.15,'WEIGHTING REVIEW','INDICATOR REVIEW'));result=res;print(result.round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
