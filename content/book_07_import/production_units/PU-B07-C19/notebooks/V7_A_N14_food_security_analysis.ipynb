## Decision contract
Triangulate availability, access, utilization, and stability indicators for programme review. The output does not classify households, declare crisis phases, or authorize targeting or stock release.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7814);n=44;df=pd.DataFrame({'area':[f'FS{i:02d}' for i in range(n)],'availability':rng.uniform(.3,1,n),'access':rng.uniform(.2,1,n),'utilization':rng.uniform(.35,1,n),'stability':rng.uniform(.15,1,n),'coverage':rng.uniform(.55,1,n)});df.head()
```

## Evidence and quality contract
Separate production, trade, stocks, prices, income, consumption, nutrition, coping, seasonality, population, assistance, geography, survey design, and source timeliness.

```python
dims=['availability','access','utilization','stability'];df['gap_equal']=1-df[dims].mean(axis=1);df['gap_access_weighted']=1-(.2*df.availability+.4*df.access+.2*df.utilization+.2*df.stability);df['quality']=df.coverage>=.75;print(df[['gap_equal','gap_access_weighted','coverage']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Composite weights and thresholds are normative. Report dimensions separately, test weights, suppress unstable estimates, and require authorized multi-source interpretation.

```python
df['sensitive']=(df.gap_equal-df.gap_access_weighted).abs()>.07;df['status']=np.where(~df.quality,'ABSTAIN—IMPROVE COVERAGE',np.where(df.gap_access_weighted>=df.gap_access_weighted.quantile(.75),'MULTI-SECTOR FOOD-SECURITY REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['area']+dims+['coverage','sensitive','status']];print(result.head(16).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a documented review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It never replaces legal, professional, clinical, procurement, enforcement, budget, safeguarding, or ethical authority.
