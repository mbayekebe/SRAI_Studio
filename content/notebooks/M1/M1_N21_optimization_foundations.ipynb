## Objective functions

An optimization problem takes the form

\[
\min_{\mathbf x} f(\mathbf x)
\quad \text{subject to}\quad
g_i(\mathbf x)\le 0.
\]


```python
Q=np.array([[4.,1.],[1.,2.]])
c=np.array([-8.,-3.])
f=lambda x: quadratic_value(x,Q,c)
g=lambda x: quadratic_gradient(x,Q,c)
assert is_convex_quadratic(Q)
x_star=np.linalg.solve(Q,-c)
{"solution":x_star,"objective":f(x_star),"gradient":g(x_star)}
```

## Convexity

For a quadratic objective, positive semidefinite Hessian implies convexity.

```python
eigenvalues=np.linalg.eigvalsh(Q)
eigenvalues
```

## Lagrangian and KKT intuition

\[
\mathcal L(x,\lambda)=f(x)+\lambda^\top g(x).
\]

KKT conditions combine stationarity, feasibility, dual feasibility, and complementarity.


```python
A=np.array([[1.,1.],[-1.,0.],[0.,-1.]])
b=np.array([1.5,0.,0.])
x=np.array([1.,0.5])
lam=np.array([1.,0.,0.])
res=kkt_residual(g(x),A,x,b,lam)
{"lagrangian":lagrangian(f(x),A@x-b,lam),"KKT_residuals":res}
```

## Decision Intelligence case

Allocate a fixed budget across health, education, and infrastructure while minimizing modeled risk.

```python
risk=np.array([[2.,.3,.1],[.3,1.5,.2],[.1,.2,1.]])
score=np.array([1.2,1.0,.8])
grid=[]
for h in np.linspace(0,1,101):
    for e in np.linspace(0,1-h,101):
        i=1-h-e
        x=np.array([h,e,i])
        value=.5*x@risk@x-score@x
        grid.append((value,*x))
best=min(grid,key=lambda z:z[0])
pd.Series(best,index=["objective","health","education","infrastructure"])
```
