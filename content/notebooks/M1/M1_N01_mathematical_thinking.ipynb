## Worked example — Electricity demand

$\
D(T)=\beta_0+\beta_1T
$

This intentionally simple model illustrates abstraction. It does not claim that real electricity demand is fully linear.


```python
from dataclasses import dataclass
import numpy as np

@dataclass(frozen=True)
class LinearDemandModel:
    baseline: float
    temperature_sensitivity: float

    def predict(self, temperature_c):
        t = np.asarray(temperature_c, dtype=float)
        return self.baseline + self.temperature_sensitivity * t

model = LinearDemandModel(120.0, 2.5)
temperatures = np.array([20.0, 25.0, 30.0])
predicted_demand = model.predict(temperatures)
predicted_demand
```

## Numerical verification


```python
expected = np.array([170.0, 182.5, 195.0])
assert np.allclose(predicted_demand, expected, atol=1e-12)
print("Verification passed.")
```

## Sensitivity


$\frac{dD}{dT}=\beta_1$


Each additional degree Celsius increases modeled demand by 2.5 units.


```python
delta = model.predict([31.0]) - model.predict([30.0])
assert np.allclose(delta, [2.5])
delta
```

## Modeling limitations

The model omits humidity, seasonality, weekdays, economic activity, prices, outages, and nonlinear thresholds. Reproducible code cannot compensate for a poorly specified model.
