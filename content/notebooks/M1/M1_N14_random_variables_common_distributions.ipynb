## 2. Random variables

A random variable is a measurable mapping

\[
X:\Omega\to\mathbb R.
\]

Discrete random variables assign probabilities to countable values. Continuous random variables use probability densities.


## 3. PMF, PDF, and CDF

For a discrete variable,

\[
p_X(x)=P(X=x).
\]

For a continuous variable,

\[
P(a\le X\le b)=\int_a^b f_X(x)\,dx.
\]

For both,

\[
F_X(x)=P(X\le x).
\]


## 4. Bernoulli distribution

```python
p = 0.3
bernoulli_table = pd.DataFrame({
    "x": [0, 1],
    "P(X=x)": [bernoulli_pmf(0, p), bernoulli_pmf(1, p)],
})
bernoulli_table
```

A Bernoulli variable represents one success/failure trial:

\[
P(X=1)=p,\qquad P(X=0)=1-p.
\]


## 5. Binomial distribution

```python
n, p = 10, 0.4
k_values = np.arange(n + 1)
binomial_probs = np.array([binomial_pmf(int(k), n, p) for k in k_values])

assert np.isclose(binomial_probs.sum(), 1.0)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(k_values, binomial_probs)
ax.set_xlabel("Number of successes")
ax.set_ylabel("Probability")
ax.set_title("Binomial Distribution: n=10, p=0.4")
plt.show()
```

If \(X\sim\operatorname{Binomial}(n,p)\),

\[
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}.
\]


## 6. Poisson distribution

```python
rate = 3.0
k_values = np.arange(0, 15)
poisson_probs = np.array([poisson_pmf(int(k), rate) for k in k_values])

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(k_values, poisson_probs)
ax.set_xlabel("Event count")
ax.set_ylabel("Probability")
ax.set_title("Poisson Distribution: rate=3")
plt.show()
```

The Poisson distribution models counts over an interval:

\[
P(X=k)=e^{-\lambda}\frac{\lambda^k}{k!}.
\]


## 7. Uniform distribution

```python
x = np.linspace(-1, 2, 500)
density = uniform_pdf(x, 0.0, 1.0)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, density)
ax.set_xlabel("x")
ax.set_ylabel("Density")
ax.set_title("Uniform Distribution on [0,1]")
plt.show()
```

## 8. Normal distribution

```python
x = np.linspace(-5, 5, 1000)
curves = {
    "N(0,1)": normal_pdf(x, 0, 1),
    "N(0,2²)": normal_pdf(x, 0, 2),
    "N(2,1)": normal_pdf(x, 2, 1),
}

fig, ax = plt.subplots(figsize=(7, 4))
for label, y in curves.items():
    ax.plot(x, y, label=label)
ax.set_xlabel("x")
ax.set_ylabel("Density")
ax.set_title("Normal Distributions")
ax.legend()
plt.show()
```

The normal density is

\[
f(x)
=
\frac{1}{\sigma\sqrt{2\pi}}
\exp\left[
-\frac{(x-\mu)^2}{2\sigma^2}
\right].
\]


## 9. Exponential distribution

```python
x = np.linspace(0, 8, 500)

fig, ax = plt.subplots(figsize=(7, 4))
for rate in [0.5, 1.0, 2.0]:
    ax.plot(x, exponential_pdf(x, rate), label=f"rate={rate}")
ax.set_xlabel("Waiting time")
ax.set_ylabel("Density")
ax.set_title("Exponential Distributions")
ax.legend()
plt.show()
```

The exponential distribution models nonnegative waiting times:

\[
f(x)=\lambda e^{-\lambda x},
\qquad x\ge0.
\]


## 10. Empirical versus theoretical distributions

```python
samples = sample_distribution(
    "normal",
    size=20_000,
    seed=42,
    mean=1.0,
    std=2.0,
)

moments = sample_moments(samples)
moments
```

```python
x = np.linspace(-6, 8, 500)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(samples, bins=50, density=True, alpha=0.4, label="Simulation")
ax.plot(x, normal_pdf(x, 1.0, 2.0), label="Theoretical density")
ax.set_xlabel("x")
ax.set_ylabel("Density")
ax.set_title("Empirical and Theoretical Normal Distribution")
ax.legend()
plt.show()
```

## 11. Empirical CDF

```python
points = np.linspace(-4, 6, 300)
ecdf = empirical_cdf(samples, points)
theoretical_cdf = stats.norm.cdf(points, loc=1.0, scale=2.0)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(points, ecdf, label="Empirical CDF")
ax.plot(points, theoretical_cdf, label="Theoretical CDF")
ax.set_xlabel("x")
ax.set_ylabel("Cumulative probability")
ax.set_title("Empirical and Theoretical CDF")
ax.legend()
plt.show()
```

## 12. Distribution choice matters

Different distributions encode different supports and assumptions:

| Distribution | Typical use |
|---|---|
| Bernoulli | One binary event |
| Binomial | Number of successes in fixed trials |
| Poisson | Counts over an interval |
| Uniform | Equal density on a bounded interval |
| Normal | Symmetric continuous variation |
| Exponential | Positive waiting times |


## 13. Statistics interpretation

Random-variable distributions define probability models for observations. They support likelihoods, estimators, confidence intervals, hypothesis tests, and predictive models.

## 15. Decision Intelligence case — Emergency-demand distributions

Suppose daily emergency demand is represented by a Poisson count model, while service time follows an exponential distribution.


```python
daily_cases = sample_distribution(
    "poisson",
    size=10_000,
    seed=7,
    rate=8.0,
)
service_times = sample_distribution(
    "exponential",
    size=10_000,
    seed=8,
    rate=1/2.5,
)

summary = {
    "mean_daily_cases": daily_cases.mean(),
    "probability_cases_exceed_12": np.mean(daily_cases > 12),
    "mean_service_time": service_times.mean(),
    "probability_service_exceeds_5": np.mean(service_times > 5),
}
summary
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(daily_cases, bins=np.arange(daily_cases.min(), daily_cases.max()+2)-0.5, density=True)
ax.set_xlabel("Daily emergency cases")
ax.set_ylabel("Relative frequency")
ax.set_title("Simulated Emergency Demand")
plt.show()
```

### Interpretation

A distribution provides a compact uncertainty model. It should be checked against data, revised when conditions change, and stress-tested for tail events before operational use.


## 16. Engineering notes

- Distribution support must match the variable.
- Count data are not automatically Poisson.
- Normal models may assign probability to impossible values.
- Tail behavior matters in risk-sensitive decisions.
- Parameter estimates have uncertainty.
- Random-number generation should use explicit seeds in reproducible studies.


## 17. Common errors

- Confusing a density value with a probability.
- Using continuous formulas for discrete variables.
- Ignoring support restrictions.
- Selecting a distribution only because it is familiar.
- Reporting mean behavior without tail probabilities.
- Treating simulated data as observed evidence.
