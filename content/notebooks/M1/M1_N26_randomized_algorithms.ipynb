## Monte Carlo estimation

\[
\pi \approx 4\frac{\#\{(x,y):x^2+y^2\le1\}}{N}.
\]


```python
rows=[]
for n in [1000,10000,100000]:
    estimate,se=monte_carlo_pi(n,seed=42)
    rows.append({"samples":n,"estimate":estimate,"standard_error":se})
pd.DataFrame(rows)
```

```python
estimates=[monte_carlo_pi(5000,seed=s)[0] for s in range(200)]
fig,ax=plt.subplots(figsize=(7,4))
ax.hist(estimates,bins=25,density=True)
ax.axvline(np.pi,linestyle="--")
ax.set_title("Monte Carlo Estimates of π")
plt.show()
```

## Random projection

For a random matrix \(R\),

\[
Y=XR
\]

can preserve pairwise distances approximately in a lower-dimensional space.


```python
rng=np.random.default_rng(1)
X=rng.normal(size=(200,100))
Y,R=random_project(X,20,seed=2)
distortion=distance_distortion(X,Y)
distortion
```

## CountSketch

```python
sketched,S=count_sketch(X,50,seed=3)
{"original_shape":X.shape,"sketched_shape":sketched.shape}
```

## Decision Intelligence case

Use random projection to compress a large indicator matrix before clustering or visualization.

```python
indicators=rng.normal(size=(300,80))
compressed,_=random_project(indicators,10,seed=4)
{"variance_original":indicators.var(),"variance_compressed":compressed.var()}
```

## Engineering notes

Randomized methods trade exactness for speed, memory savings, and probabilistic guarantees.

## Key insight

Randomization can make otherwise expensive computations feasible while controlling approximation error.
