```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_math.scientific import (
    chunked_mean,moving_average,pairwise_squared_distances,
    sparse_matvec,standardize_columns,to_csr,
)
from scipy import sparse
import time
```

## Vectorization

```python
rng=np.random.default_rng(42)
X=rng.normal(size=(100000,20))
start=time.perf_counter()
loop=np.array([row.mean() for row in X])
loop_time=time.perf_counter()-start
start=time.perf_counter()
vectorized=X.mean(axis=1)
vectorized_time=time.perf_counter()-start
{"loop_seconds":loop_time,"vectorized_seconds":vectorized_time,"speedup":loop_time/vectorized_time}
```

## Broadcasting and standardization

```python
Z,mean,std=standardize_columns(X[:1000])
{"mean_after":np.abs(Z.mean(axis=0)).max(),"std_after_error":np.abs(Z.std(axis=0)-1).max()}
```

## Pairwise distances

```python
D=pairwise_squared_distances(X[:100])
{"shape":D.shape,"symmetry_error":np.linalg.norm(D-D.T)}
```

## Sparse matrices

```python
dense=(rng.random((1000,1000))<0.01)*rng.normal(size=(1000,1000))
csr=to_csr(dense)
{"dense_MB":dense.nbytes/1e6,"sparse_MB":(csr.data.nbytes+csr.indices.nbytes+csr.indptr.nbytes)/1e6}
```

## Sparse matrix-vector multiplication

```python
v=rng.normal(size=1000)
result=sparse_matvec(csr,v)
np.linalg.norm(result)
```

## Chunked processing

```python
chunk_mean=chunked_mean(X,chunk_size=10000)
assert np.allclose(chunk_mean,X.mean(axis=0))
chunk_mean[:5]
```

## Moving averages

```python
series=np.sin(np.linspace(0,10,300))+rng.normal(scale=.2,size=300)
smooth=moving_average(series,15)
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(series,label="raw")
ax.plot(np.arange(14,300),smooth,label="moving average")
ax.legend(); ax.set_title("Scientific Computing Workflow")
plt.show()
```

## Decision Intelligence case

Process a large monitoring matrix efficiently without loading unnecessary intermediate structures.

```python
monitoring=rng.normal(size=(200000,30))
summary=chunked_mean(monitoring,20000)
summary[:10]
```
