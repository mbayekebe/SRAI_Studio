## Shannon entropy

\[
H(X)=-\sum_i p_i\log_2 p_i.
\]


```python
distributions={"certain":[1,0],"biased":[.9,.1],"moderate":[.7,.3],"uniform":[.5,.5]}
pd.Series({k:entropy(v) for k,v in distributions.items()},name="entropy_bits")
```

```python
p=np.linspace(0,1,500)
h=np.array([binary_entropy(v) for v in p])
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(p,h); ax.set_xlabel("Bernoulli probability p"); ax.set_ylabel("Entropy (bits)"); ax.set_title("Binary Entropy")
plt.show()
```

## Self-information

\[
I(x)=-\log_2 p(x).
\]


```python
probabilities=np.array([.5,.25,.125,.01])
pd.DataFrame({"probability":probabilities,"self_information_bits":-np.log2(probabilities)})
```

## Joint and conditional entropy

\[
H(X,Y)=-\sum_{x,y}p(x,y)\log_2p(x,y),
\qquad
H(X\mid Y)=H(X,Y)-H(Y).
\]


```python
independent=np.array([[.25,.25],[.25,.25]])
dependent=np.array([[.5,0],[0,.5]])
{"joint_entropy_independent":joint_entropy(independent),"conditional_entropy_independent":conditional_entropy(independent),"conditional_entropy_dependent":conditional_entropy(dependent)}
```

## Mutual information

\[
I(X;Y)=H(X)+H(Y)-H(X,Y).
\]


```python
{"independent":mutual_information(independent),"perfectly_dependent":mutual_information(dependent)}
```

## Cross-entropy and KL divergence

\[
H(p,q)=-\sum_i p_i\log_2q_i,
\qquad
D_{KL}(p\|q)=\sum_i p_i\log_2\frac{p_i}{q_i}.
\]

\[
H(p,q)=H(p)+D_{KL}(p\|q).
\]


```python
p_true=np.array([.7,.3]); q_good=np.array([.65,.35]); q_bad=np.array([.3,.7])
comparison=pd.DataFrame({"model":["Good model","Bad model"],"cross_entropy":[cross_entropy(p_true,q_good),cross_entropy(p_true,q_bad)],"KL_divergence":[kl_divergence(p_true,q_good),kl_divergence(p_true,q_bad)]})
assert np.isclose(cross_entropy(p_true,q_good),entropy(p_true)+kl_divergence(p_true,q_good))
comparison
```

## Information gain

Information gain is the entropy reduction produced by a split and is widely used in decision trees.


```python
parent=[.5,.5]
pd.Series({"perfect_split":information_gain(parent,[(.5,[1,0]),(.5,[0,1])]),"weak_split":information_gain(parent,[(.5,[.6,.4]),(.5,[.4,.6])])})
```

## Mutual information from simulated data

```python
rng=np.random.default_rng(42)
x=rng.integers(0,2,20000); noise=rng.binomial(1,.15,20000); y=np.bitwise_xor(x,noise)
joint=np.zeros((2,2))
for xi,yi in zip(x,y): joint[xi,yi]+=1
joint/=joint.sum()
{"joint_distribution":joint,"mutual_information_bits":mutual_information(joint)}
```

## AI interpretation

Information measures support cross-entropy loss, decision trees, feature selection, representation learning, variational inference, language modeling, compression, and calibration.


## Decision Intelligence case — Information value of a warning signal

```python
warning_joint=np.array([[.63,.07],[.08,.22]])
row,col=marginal_probabilities(warning_joint)
{"disruption_entropy":entropy(row),"warning_entropy":entropy(col),"mutual_information":mutual_information(warning_joint),"remaining_uncertainty":conditional_entropy(warning_joint,condition_on="columns")}
```

Mutual information measures average uncertainty reduction from the warning signal. Operational value also depends on timeliness, false alarms, missed events, costs, and governance.


## Engineering notes

- Zero probabilities require careful handling.
- Cross-entropy can be infinite when a model assigns zero probability to an event that occurs.
- KL divergence is directional.
- Mutual-information estimation from limited continuous data is difficult.
- Information gain can favor high-cardinality variables.
