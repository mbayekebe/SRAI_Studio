## 2. Limit intuition

\[
\lim_{x\to a} f(x)=L
\]

means that \(f(x)\) can be made arbitrarily close to \(L\) by taking \(x\) sufficiently close to \(a\), without requiring \(x=a\).


## 3. Polynomial example

For

\[
f(x)=x^2+2x+1,
\]

continuity implies

\[
\lim_{x\to 2}f(x)=f(2)=9.
\]


```python
f = lambda x: x**2 + 2*x + 1
xs = approach_sequence(2.0, side="both")
table = pd.DataFrame({"x": xs, "f(x)": [f(x) for x in xs]})
table
```

```python
estimate = numerical_limit(f, 2.0)
assert np.isclose(estimate, 9.0, rtol=1e-5)
estimate
```

## 4. Removable discontinuity

Consider

\[
g(x)=\frac{x^2-1}{x-1},\qquad x\neq 1.
\]

Since

\[
g(x)=x+1
\]

for \(x\neq 1\),

\[
\lim_{x\to1}g(x)=2.
\]


```python
g = lambda x: (x**2 - 1)/(x - 1)
g_limit = numerical_limit(g, 1.0)
assert np.isclose(g_limit, 2.0, rtol=1e-5)
g_limit
```

## 5. Filling the hole

```python
g_extended = removable_extension(g, 1.0, 2.0)

assert g_extended(1.0) == 2.0
assert is_continuous_at(g_extended, 1.0)

g_extended(1.0)
```

```python
x_plot = np.linspace(-1, 3, 400)
y_plot = np.array([g_extended(x) for x in x_plot])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_plot, y_plot)
ax.scatter([1.0], [2.0])
ax.set_xlabel("x")
ax.set_ylabel("g(x)")
ax.set_title("Removable Discontinuity After Extension")
plt.show()
```

## 6. One-sided limits and jump discontinuity

Define

\[
h(x)=
\begin{cases}
-1,&x<0,\\
1,&x\ge0.
\end{cases}
\]

Then

\[
\lim_{x\to0^-}h(x)=-1,
\qquad
\lim_{x\to0^+}h(x)=1.
\]

Because the one-sided limits differ, the two-sided limit does not exist.


```python
h = lambda x: -1.0 if x < 0 else 1.0
left_limit, right_limit = one_sided_limits(h, 0.0)

assert np.isclose(left_limit, -1.0)
assert np.isclose(right_limit, 1.0)

left_limit, right_limit
```

```python
assert not is_continuous_at(h, 0.0)
print("Jump discontinuity detected.")
```

## 7. Infinite behavior

For

\[
q(x)=\frac{1}{x^2},
\]

the values grow without bound as \(x\to0\). This is not a finite limit.


```python
q = lambda x: 1/(x**2)
scales = 10.0 ** (-np.arange(1, 7))
infinite_table = pd.DataFrame({
    "x": scales,
    "q(x)": [q(x) for x in scales],
})
infinite_table
```

## 8. Continuity at a point

A function is continuous at \(a\) when:

1. \(f(a)\) is defined;
2. \(\lim_{x\to a}f(x)\) exists;
3. \(\lim_{x\to a}f(x)=f(a)\).


```python
continuous_cases = {
    "x^2 at 1": is_continuous_at(lambda x: x**2, 1.0),
    "sin(x) at 0": is_continuous_at(np.sin, 0.0),
    "jump at 0": is_continuous_at(h, 0.0),
}
continuous_cases
```

## 9. Epsilon–delta reasoning

The formal statement

\[
\lim_{x\to a}f(x)=L
\]

means that for every \(\varepsilon>0\), there exists \(\delta>0\) such that

\[
0<|x-a|<\delta
\Rightarrow
|f(x)-L|<\varepsilon.
\]


```python
linear = lambda x: 3*x + 2

check = epsilon_delta_check(
    linear,
    point=1.0,
    limit_value=5.0,
    epsilon=0.03,
    delta=0.009,
)
assert check
check
```

## 10. Visualizing an epsilon–delta neighborhood

```python
a = 1.0
L = 5.0
epsilon = 0.3
delta = epsilon / 3.0

x_local = np.linspace(a - 0.2, a + 0.2, 400)
y_local = linear(x_local)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_local, y_local)
ax.axvline(a - delta, linestyle="--")
ax.axvline(a + delta, linestyle="--")
ax.axhline(L - epsilon, linestyle="--")
ax.axhline(L + epsilon, linestyle="--")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.set_title("Epsilon–Delta Neighborhood for a Linear Function")
plt.show()
```

## 11. Continuity and composition

If \(f\) is continuous at \(a\), and \(g\) is continuous at \(f(a)\), then \(g\circ f\) is continuous at \(a\).


```python
f1 = lambda x: x**2 + 1
g1 = np.sqrt
composition = lambda x: g1(f1(x))

assert is_continuous_at(composition, 2.0)
composition(2.0)
```

## 12. Statistics interpretation

Continuity supports smooth likelihoods, stable estimators, and approximation arguments. Discontinuities may arise in threshold rules, censoring, truncation, and indicator functions.

## 14. Decision Intelligence case — Threshold policy rules

A discontinuous policy rule can change sharply when an indicator crosses a threshold.


```python
def emergency_level(risk_score):
    return 0 if risk_score < 0.70 else 1

risk_scores = np.linspace(0.60, 0.80, 200)
levels = np.array([emergency_level(x) for x in risk_scores])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(risk_scores, levels)
ax.axvline(0.70, linestyle="--")
ax.set_xlabel("Risk score")
ax.set_ylabel("Emergency status")
ax.set_title("Discontinuous Threshold Policy")
plt.show()
```

### Interpretation

Thresholds can be operationally necessary, but tiny measurement changes near the boundary may trigger large decision changes. Decision systems should document uncertainty, hysteresis, review rules, and appeal mechanisms.


## 15. Smooth policy alternative

```python
def smooth_emergency_score(risk_score, center=0.70, steepness=40.0):
    return 1.0 / (1.0 + np.exp(-steepness * (risk_score - center)))

smooth_levels = smooth_emergency_score(risk_scores)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(risk_scores, smooth_levels)
ax.axvline(0.70, linestyle="--")
ax.set_xlabel("Risk score")
ax.set_ylabel("Continuous emergency score")
ax.set_title("Smooth Approximation to a Threshold Rule")
plt.show()
```

A smooth score supports gradual escalation, but the final policy still requires accountable decision rules and safeguards.


## 16. Engineering notes

- Numerical limits depend on scale and floating-point precision.
- Subtractive cancellation can corrupt values near removable discontinuities.
- Sampling does not prove a limit.
- Piecewise functions require explicit boundary handling.
- Smooth approximations may improve optimization but change policy semantics.


## 17. Common errors

- Substituting the point when the function is undefined there.
- Assuming equal function values imply continuity.
- Ignoring one-sided limits.
- Confusing divergence with a large finite value.
- Treating numerical evidence as a formal proof.
- Smoothing a rule without evaluating governance consequences.
