```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_math.case_studies import (
    normalized_scores,portfolio_risk,rank_options,scenario_loss,weighted_score,
)
from srai_math.information import entropy
from srai_math.optimization import project_simplex,projected_gradient_descent
```

## Case 1 — Multi-criteria programme ranking

```python
labels=["Programme_A","Programme_B","Programme_C","Programme_D"]
X=np.array([
    [80,60,70,55],
    [75,85,65,60],
    [90,55,80,50],
    [70,75,60,90],
],dtype=float)
Z=normalized_scores(X)
weights=np.array([.35,.25,.25,.15])
scores=weighted_score(Z,weights)
rank_options(scores,labels)
```

## Case 2 — Scenario-weighted risk

```python
probabilities=np.array([.2,.5,.3])
losses=np.array([120,40,70])
{"expected_loss":scenario_loss(probabilities,losses),"scenario_entropy_bits":entropy(probabilities)}
```

## Case 3 — Portfolio allocation

```python
returns=np.array([1.2,1.0,.9])
cov=np.array([[1,.3,.1],[.3,.8,.2],[.1,.2,.6]])
objective=lambda w: float(.5*w@cov@w-.6*returns@w)
gradient=lambda w: cov@w-.6*returns
w,h,_=projected_gradient_descent(
    objective,gradient,[1/3]*3,lambda x: project_simplex(x,1),
    learning_rate=.2,max_iter=500
)
{"allocation":w,"risk":portfolio_risk(w,cov),"objective":h[-1]}
```

## Case 4 — Uncertainty-aware monitoring

```python
rng=np.random.default_rng(42)
observations=rng.normal(loc=[10,20,15],scale=[2,5,3],size=(1000,3))
means=observations.mean(axis=0)
ses=observations.std(axis=0,ddof=1)/np.sqrt(len(observations))
pd.DataFrame({"mean":means,"standard_error":ses},index=["Indicator_1","Indicator_2","Indicator_3"])
```

## Case 5 — Sensitivity of ranking to weights

```python
weight_sets=[
    np.array([.4,.2,.2,.2]),
    np.array([.2,.4,.2,.2]),
    np.array([.25,.25,.25,.25]),
]
rows=[]
for i,wgt in enumerate(weight_sets,1):
    s=weighted_score(Z,wgt)
    rows.append({"scenario":i,"top_option":rank_options(s,labels)[0][0]})
pd.DataFrame(rows)
```
