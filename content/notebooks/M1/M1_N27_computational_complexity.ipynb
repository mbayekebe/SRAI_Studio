## Empirical scaling

```python
def matmul_work(n):
    rng=np.random.default_rng(1)
    A=rng.normal(size=(n,n))
    B=rng.normal(size=(n,n))
    A@B

sizes=[50,80,120,180]
runtime=empirical_runtime(matmul_work,sizes,repeats=2)
runtime
```

```python
slope,_=estimate_loglog_slope([n for n,t in runtime],[t for n,t in runtime])
{"empirical_loglog_slope":slope}
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.loglog([n for n,t in runtime],[t for n,t in runtime],marker="o")
ax.set_xlabel("Matrix size n"); ax.set_ylabel("Runtime")
ax.set_title("Empirical Matrix Multiplication Scaling")
plt.show()
```
