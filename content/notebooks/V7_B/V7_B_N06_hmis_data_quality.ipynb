## Decision contract
Audit aggregate health-management information before performance or planning use. The notebook does not rank staff, sanction facilities, alter care, or publish identifiable records.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7706);n=70;df=pd.DataFrame({'facility':[f'F{i:03d}' for i in range(n)],'expected_reports':12,'received_reports':rng.integers(6,13,n),'outpatient':rng.integers(400,9000,n),'population':rng.integers(3000,75000,n),'late_days':rng.integers(0,30,n)});df.loc[:3,'outpatient']=df.loc[:3,'population']*3;df.head()
```

## Evidence and quality contract
Track indicator definition, numerator, denominator, reporting universe, facility status, duplicates, completeness, timeliness, internal consistency, revisions, and private/community coverage.

```python
df['completeness']=df.received_reports/df.expected_reports;df['utilization']=df.outpatient/df.population;df['quality_score']=.6*df.completeness+.4*(1-df.late_days/30).clip(lower=0);df['plausible']=df.utilization<=2.5;print(df[['completeness','utilization','quality_score']].describe().round(2))
```

## Uncertainty, sensitivity, and abstention
Quality rules are diagnostic. Compare strict and moderate tolerances; missing reports produce abstention rather than zero service.

```python
df['status']=np.where(df.completeness<.8,'ABSTAIN—MISSING REPORTS',np.where(~df.plausible,'VERIFY NUMERATOR/DENOMINATOR',np.where(df.quality_score<.8,'DATA QUALITY REVIEW','MONITOR')));result=df[df.status!='MONITOR'][['facility','completeness','utilization','late_days','quality_score','status']];print(result.head(18).round(2).to_string(index=False));print({'moderate_flags':int((df.quality_score<.8).sum()),'strict_flags':int((df.quality_score<.9).sum())})
```
