## Decision contract
Support aggregate antimicrobial-resistance surveillance and laboratory verification. Outputs do not prescribe treatment, identify patients, close facilities, or trigger public alerts automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7707);n=48;df=pd.DataFrame({'site':[f'Lab{i:02d}' for i in range(n)],'tested':rng.integers(8,420,n),'resistant':rng.integers(0,90,n),'coverage':rng.uniform(.5,1,n),'breakpoint_current':rng.choice([0,1],n,p=[.12,.88])});df['resistant']=np.minimum(df.resistant,df.tested);df.head()
```

## Evidence and quality contract
Standardize organism, specimen, antimicrobial, method, breakpoint version, susceptible/intermediate/resistant coding, deduplication, patient episode, site coverage, and testing practices.

```python
df['rate']=df.resistant/df.tested;df['se']=np.sqrt(df.rate*(1-df.rate)/df.tested);df['lower']=(df.rate-1.96*df.se).clip(lower=0);df['upper']=(df.rate+1.96*df.se).clip(upper=1);df['quality']=(df.tested>=30)&(df.coverage>=.7)&(df.breakpoint_current==1);print(df[['rate','lower','upper']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Small denominators and changing laboratory coverage distort rates. Use minimum counts, intervals, and abstention; trends require stable definitions and testing populations.

```python
df['status']=np.where(~df.quality,'ABSTAIN—LAB/COVERAGE REVIEW',np.where(df.lower>.25,'AMR STEWARDSHIP REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['site','tested','resistant','rate','lower','upper','coverage','status']].sort_values('rate',ascending=False);print(result.head(15).round(3).to_string(index=False))
```
