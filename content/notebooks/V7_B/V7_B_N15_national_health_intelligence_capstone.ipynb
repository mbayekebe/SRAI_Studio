## Decision contract
Integrate surveillance, capacity, access, commodities, HMIS quality, AMR, climate, nutrition, referrals, quality, surveys, emergency operations, and evaluation into a governed health decision service. Outputs support authorized public-health and health-system review; they do not diagnose, prescribe, triage individual patients, or replace incident command.


```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7215); facilities=[f'F{i:02d}' for i in range(24)]; df=pd.DataFrame({'facility':facilities,'syndrome_rate':rng.uniform(.02,.18,24),'baseline_rate':rng.uniform(.03,.11,24),'bed_pressure':rng.uniform(.35,1.25,24),'stock_days':rng.uniform(1,25,24),'reporting':rng.uniform(.55,1,24),'case_mix':rng.uniform(.7,1.4,24),'access_gap':rng.uniform(0,.55,24),'response_capacity':rng.uniform(.3,1,24)}); df.head()
```

## Evidence and quality contract
Define population, case and service concepts, reporting coverage, laboratory confirmation, facility capacity, commodities, referrals, survey design, quality measures, outcomes, privacy, case mix, intervention capacity, clinical/public-health authority, model version, monitoring, incidents, and retirement.


```python
df['signal_ratio']=df.syndrome_rate/df.baseline_rate.clip(lower=.01); df['evidence_ok']=(df.reporting>=.8)&df.case_mix.between(.75,1.35); df['system_pressure']=.35*(df.signal_ratio/2).clip(0,1)+.25*(df.bed_pressure/1.2).clip(0,1)+.20*(1-df.stock_days/25).clip(0,1)+.20*df.access_gap; print(df[['signal_ratio','system_pressure','reporting']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Delayed reporting, missing care, case mix, selection, small numbers, diagnostic error, forecast uncertainty, and unequal access can distort signals. Calibrate alerts, show intervals, protect small groups, test subgroups, and abstain when evidence or response capacity is inadequate.


```python
df['decision_status']=np.where(~df.evidence_ok,'ABSTAIN - VERIFY DATA/CASE MIX',np.where((df.system_pressure>=.58)&(df.response_capacity>=.5),'PUBLIC HEALTH / SYSTEM REVIEW',np.where(df.system_pressure>=.58,'CAPACITY ESCALATION','MONITOR'))); brief=df[df.decision_status!='MONITOR'].sort_values('system_pressure',ascending=False); model_card={'purpose':'aggregate health-system signal review','prohibited':['diagnosis','individual triage','automatic discipline'],'privacy':'aggregate and suppress small cells','owner':'Health intelligence board'}; audit={'facilities':len(df),'review_queue':len(brief),'abstentions':int((df.decision_status.str.startswith('ABSTAIN')).sum())}; print(brief.round(3).to_string(index=False)); print(model_card,audit)
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
