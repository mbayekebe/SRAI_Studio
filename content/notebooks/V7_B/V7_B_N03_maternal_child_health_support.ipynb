## Decision contract
Identify service and referral-system gaps at aggregate area level. No output diagnoses a person, changes treatment, denies care, or publishes identifiable maternal/child records.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7603);n=40
df=pd.DataFrame({'area':[f'H{i:02d}' for i in range(n)],'births':rng.integers(250,2600,n),'anc4':rng.uniform(.35,.95,n),'facility_delivery':rng.uniform(.4,.98,n),'referral_minutes':rng.integers(12,190,n),'staff_fill':rng.uniform(.45,1,n),'data_age':rng.integers(0,15,n)});df.head()
```

## Evidence and quality contract
Use aggregate, disclosure-controlled indicators with complete denominators. Audit missing facilities, private/community care, referral definitions, timeliness, and denominator revisions.

```python
df['quality']=df.data_age<=7;df['gap']=.35*(1-df.anc4)+.30*(1-df.facility_delivery)+.20*(df.referral_minutes/180).clip(upper=1)+.15*(1-df.staff_fill);print(df.gap.describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Thresholds are planning choices. Compare the top 15% and 25%; stale reports cause abstention. Associations do not identify clinical cause.

```python
q=df.gap.quantile(.75);df['status']=np.where(~df.quality,'ABSTAIN—VERIFY REPORT',np.where(df.gap>=q,'SERVICE AND REFERRAL REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['area','births','gap','referral_minutes','data_age','status']].sort_values('gap',ascending=False);print(result.round(2).to_string(index=False));print({'top15':int((df.gap>=df.gap.quantile(.85)).sum()),'top25':int((df.gap>=q).sum())})
```

## Decision product and authority boundary
The output is a review queue with reason codes, evidence age, uncertainty, accountable owner, and status. It never transfers legal, clinical, professional, procurement, enforcement, or budget authority to the model.
