## Decision contract
Estimate whether a health programme plausibly changed outcomes and for whom, while separating impact from implementation. The analysis supports evaluation governance; it does not prove causality from association or withdraw services automatically.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7214); n=500; df=pd.DataFrame({'treated':rng.binomial(1,.5,n),'baseline':rng.normal(50,10,n),'fidelity':rng.uniform(.45,1,n)}); df['followup']=df.baseline+2+3*df.treated*df.fidelity+rng.normal(0,6,n); df['observed']=rng.binomial(1,.92-.05*df.treated,n); df.head()
```

## Evidence and quality contract
Specify theory of change, eligibility, assignment, comparison, baseline, outcome window, implementation fidelity, spillovers, attrition, confounders, harms, costs, subgroup hypotheses, protocol, registration, and decision owner.


```python
obs=df[df['observed']==1].copy(); obs['change']=obs.followup-obs.baseline; effect=obs.groupby('treated').change.mean().diff().iloc[-1]; attrition=1-df.groupby('treated')['observed'].mean(); print('change_difference',round(effect,3)); print('attrition',attrition.round(3).to_dict())
```

## Uncertainty, sensitivity, and abstention
Selection, time trends, concurrent policies, and missing outcomes can bias estimates. Use credible designs, intervals, robustness checks, negative controls where appropriate, and abstention when identification fails.


```python
high=obs[obs.fidelity>=.75]; sensitivity=high.groupby('treated').change.mean().diff().iloc[-1]; status='EVALUATION REVIEW' if max(attrition)<.12 else 'ABSTAIN - ATTRITION REVIEW'; print({'primary':round(effect,3),'high_fidelity':round(sensitivity,3),'status':status})
```
