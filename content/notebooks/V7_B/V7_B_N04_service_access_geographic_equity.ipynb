## Decision contract
Support geographic service planning using population-weighted access evidence. Straight-line proximity is a screening measure, not proof of travel time, eligibility, capacity, or effective access.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7604);m=55;f=9
communities=pd.DataFrame({'community':[f'C{i:02d}' for i in range(m)],'x':rng.uniform(0,100,m),'y':rng.uniform(0,100,m),'population':rng.integers(500,9000,m),'vulnerability':rng.uniform(0,1,m)})
fac=pd.DataFrame({'facility':[f'F{i}' for i in range(f)],'x':rng.uniform(0,100,f),'y':rng.uniform(0,100,f),'capacity':rng.integers(3000,25000,f)});communities.head()
```

## Evidence and quality contract
Use authoritative facility status, service package, hours, capacity, road/season barriers, population surfaces, boundaries, and update dates. Geographic errors can systematically disadvantage remote areas.

```python
xy=communities[['x','y']].to_numpy();ff=fac[['x','y']].to_numpy();d=np.sqrt(((xy[:,None,:]-ff[None,:,:])**2).sum(2));communities['nearest_km']=d.min(1);communities['nearest_facility']=fac.facility.iloc[d.argmin(1)].to_numpy();print(communities.nearest_km.describe().round(1))
```

## Uncertainty, sensitivity, and abstention
Compare 15 km and 25 km screening thresholds. High vulnerability and uncertain travel conditions receive verification, not exclusion.

```python
communities['priority']=communities.nearest_km*(1+.5*communities.vulnerability);communities['status']=np.where(communities.nearest_km>25,'ACCESS OPTIONS REVIEW',np.where(communities.nearest_km>15,'VERIFY TRAVEL BARRIERS','MONITOR'));result=communities[communities.status!='MONITOR'][['community','population','nearest_km','vulnerability','priority','status']].sort_values('priority',ascending=False);print(result.head(15).round(2).to_string(index=False));print({'pop_over_15km':int(communities.loc[communities.nearest_km>15,'population'].sum()),'pop_over_25km':int(communities.loc[communities.nearest_km>25,'population'].sum())})
```
