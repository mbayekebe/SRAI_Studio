# V5 Notebook 05 — Model Serving and API Contracts

## Learning objectives

- Wrap a model behind a stable request/response contract.
- Track health, latency, requests, and errors.
- Prepare for FastAPI-based serving.

```python
from srai_math.utils import environment_info,set_seed
from pathlib import Path
import json
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import ModelService,PredictionRequest,response_to_dict
```

```python
weights=np.array([.4,-.2,.8])
service=ModelService(
    predict_function=lambda x:[1/(1+np.exp(-(x@weights)))],
    model_version="1.0.0",
)
responses=[]
for i,features in enumerate([[1,2,1],[0,1,2],[2,0,1]],1):
    response=service.predict(PredictionRequest(f"req-{i}",features))
    responses.append(response_to_dict(response))
{"responses":responses,
 "health":service.health(),
 "metrics":service.metrics()}
```

## API contract

A production endpoint should validate inputs, identify model versions, return traceable responses, and expose health and metrics.

## Key insight

Serving is not merely calling `predict`; it is maintaining a reliable, observable, versioned interface.
