# V5 Notebook 12 — Prefect Workflows

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import PrefectFlow,PrefectTask,deployment_spec
```

```python
flow=PrefectFlow("risk_model_flow",[PrefectTask("extract",lambda p,d:pd.DataFrame({"x":[1,2,3]})),PrefectTask("transform",lambda p,d:d["extract"].assign(x2=d["extract"]["x"]*2),dependencies=["extract"]),PrefectTask("train",lambda p,d:{"mean_feature":float(d["transform"]["x2"].mean())},retries=2,dependencies=["transform"])])
{"order":flow.execution_order(),"results":flow.run({"environment":"development"}),"deployment":deployment_spec("risk_model_flow","0 6 * * *","ml-pool")}
```

## Key insight

Prefect emphasizes Python-native workflows, runtime state, retries, and flexible deployments.
