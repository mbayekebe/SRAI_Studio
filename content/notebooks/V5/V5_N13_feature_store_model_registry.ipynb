# V5 Notebook 13 — Feature Store and Model Registry

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import FeatureStore,FeatureView,ModelRegistry
```

```python
features=pd.DataFrame({"district_id":[1,2,3],"rainfall_anomaly":[-.3,.1,-.6],"ndvi":[.55,.72,.40],"price_index":[125,105,140]})
store=FeatureStore().register_view(FeatureView("district_risk_features","district_id",["rainfall_anomaly","ndvi","price_index"],"1"))
store.write_offline("district_risk_features",features).materialize_online("district_risk_features")
registry=ModelRegistry(); registry.register("crop_risk_model","1.0.0",{"auc":.84},"development"); registry.register("crop_risk_model","1.1.0",{"auc":.88},"staging"); registry.transition("crop_risk_model","1.1.0","production","ML Governance Board")
{"online_features":store.get_online_features("district_risk_features",2),"production_model":registry.latest("crop_risk_model","production")}
```

## Key insight

Feature stores standardize model inputs, while model registries govern version promotion and lineage.
