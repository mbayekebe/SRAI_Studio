```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import drift_status,governance_report,performance_change,population_stability_index,subgroup_metric
```

```python
rng=np.random.default_rng(14); reference=rng.normal(0,1,1000); current=rng.normal(.5,1.2,1000)
psi=population_stability_index(reference,current,10)
frame=pd.DataFrame({"region":["A","A","B","B","C","C"],"y_true":[1,0,1,0,1,0],"y_pred":[1,0,0,0,1,1]})
performance=performance_change(.88,.82,True)
report=governance_report("crop_risk_model","1.1.0","AI Office",{"psi":psi,"status":drift_status(psi)},performance,False)
{"drift":report["drift"],"subgroups":subgroup_metric(frame,"region","y_true","y_pred"),"report":report}
```
