# V5 Notebook 03 — Experiment Tracking

## Learning objectives

- Log parameters, metrics, tags, and artifacts.
- Compare competing runs.

```python
from srai_math.utils import environment_info,set_seed
from pathlib import Path
import json
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import ExperimentRun,compare_runs
```

```python
run_dir=Path("artifacts")/"v5_experiments"
paths=[]
for name,alpha,score in [
    ("baseline",0.0,.78),
    ("ridge",.1,.84),
    ("tuned_ridge",.05,.87),
]:
    run=ExperimentRun(name,run_dir)
    run.log_params(alpha=alpha)
    run.log_metrics(validation_score=score)
    run.set_tags(stage="development",owner="SRAI")
    paths.append(run.finish())

pd.DataFrame(compare_runs(paths,"validation_score"))
```

## Key insight

Experiment tracking converts trial-and-error into an auditable model-selection process.
