```python
from srai_math.utils import environment_info,set_seed
from pathlib import Path
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_mlops import Task,Workflow,retry
```

```python
tasks=[Task("extract",lambda c,d:pd.DataFrame({"x":[1,2,3]})),Task("transform",lambda c,d:d["extract"].assign(x2=d["extract"]["x"]*2),["extract"]),Task("train",lambda c,d:{"coefficient":float(d["transform"]["x2"].mean())},["transform"])]
workflow=Workflow(tasks); results=workflow.run()
{"order":workflow.execution_order(),"training_result":results["train"]}
```
