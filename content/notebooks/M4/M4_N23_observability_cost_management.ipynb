```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import aggregate_traces,budget_alert,estimate_token_cost,latency_summary,request_trace
```

```python
rng=np.random.default_rng(23)
traces=[]
for i in range(100):
    prompt_tokens=int(rng.integers(300,1200))
    completion_tokens=int(rng.integers(80,500))
    latency=float(rng.lognormal(np.log(900),.35))
    traces.append(request_trace(str(i),"general-model",prompt_tokens,completion_tokens,latency))
summary=aggregate_traces(traces)
cost=sum(estimate_token_cost(t["prompt_tokens"],t["completion_tokens"],2.5,10) for t in traces)
{"summary":summary,
 "latency":latency_summary([t["latency_ms"] for t in traces]),
 "estimated_cost":cost,
 "budget_status":budget_alert(cost,2.0,.8)}
```

## Key insight

Observability connects model behavior to reliability, cost, quality, and incident response.
