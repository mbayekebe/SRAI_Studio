# M4 Notebook 13 — Planning and Memory

## Learning objectives

- Decompose goals into steps.
- Maintain short-term and long-term memory.
- Retrieve prior information to support action.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_genai import LongTermMemory,ShortTermMemory,decompose_goal,summarize_memory,update_plan
```

```python
plan=decompose_goal(
    "Build an enterprise RAG assistant",
    ["ingest documents","create embeddings","configure retrieval","evaluate answers","deploy"],
)
update_plan(plan,0,"completed","Documents loaded")
pd.DataFrame(plan)
```

```python
short=ShortTermMemory(3)
for event in ["query received","retrieval completed","answer drafted","answer verified"]:
    short.add(event)
long=LongTermMemory()
long.add("SEPE dashboards monitor public-enterprise performance",{"project":"SEPE"})
long.add("Agricultural census measures farms, crops, land, and livestock",{"project":"Census"})
{"short_term":short.read(),
 "summary":summarize_memory(short.read()),
 "retrieved":long.search("farm census",1)}
```

## Key insight

Planning structures future work, while memory preserves useful context across steps and sessions.
