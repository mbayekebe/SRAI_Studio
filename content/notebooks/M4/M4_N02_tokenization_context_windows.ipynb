## Learning objectives

- Compare tokenization strategies.
- Measure token counts.
- Understand context-window constraints.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import WordTokenizer,token_statistics,whitespace_tokens
```

```python
samples=[
    "National Agricultural Intelligence Platform",
    "SEPE enterprise monitoring dashboard",
    "Les modèles de langage traitent des jetons.",
]
tokenizer=WordTokenizer().fit(samples)
pd.DataFrame({
    "text":samples,
    "whitespace_tokens":[len(whitespace_tokens(t)) for t in samples],
    "word_punctuation_tokens":[len(tokenizer.encode(t,False)) for t in samples],
})
```

## Context windows

Longer context enables more source material but increases memory, latency, and attention cost.

## Key insight

Tokenization determines model inputs, cost, context usage, and how text fragments are represented.
