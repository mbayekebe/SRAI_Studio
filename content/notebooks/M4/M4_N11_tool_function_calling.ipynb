# M4 Notebook 11 — Tool and Function Calling

## Learning objectives

- Define tool schemas.
- Validate arguments.
- Execute tool calls safely and return structured results.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_genai import ToolRegistry,ToolSpec,parse_tool_call,tool_result_message,validate_arguments
```

```python
spec=ToolSpec(
    name="calculate_budget_gap",
    description="Calculate the difference between required and committed budgets.",
    parameters={
        "type":"object",
        "properties":{
            "required":{"type":"number"},
            "committed":{"type":"number"},
        },
        "required":["required","committed"],
    },
)
registry=ToolRegistry().register(
    spec,
    lambda required,committed:{"gap":required-committed},
)
payload={"name":"calculate_budget_gap","arguments":{"required":16.3,"committed":8.6}}
name,args=parse_tool_call(payload)
{"validation":validate_arguments(spec,args),
 "result_message":tool_result_message(name,registry.execute(name,args))}
```

## Key insight

Tool calling turns language-model output into typed, executable actions governed by schemas and validation.
