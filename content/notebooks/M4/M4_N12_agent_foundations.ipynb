## Learning objectives

- Understand agent state, actions, observations, and stopping conditions.
- Build a small ReAct-style loop.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_genai import AgentState,RuleBasedAgent,run_agent
```

```python
rules=[
    (lambda s:len(s.observations)==0,lambda s:{"type":"search","query":s.goal}),
    (lambda s:len(s.observations)==1,lambda s:{"type":"summarize"}),
    (lambda s:True,lambda s:{"type":"stop","reason":"Goal addressed"}),
]
agent=RuleBasedAgent(rules)
def environment(action,state):
    if action["type"]=="search":
        return {"result":"Found relevant project documents","done":False}
    if action["type"]=="summarize":
        return {"result":"Summary prepared","done":False}
    return {"done":True}
state,trace=run_agent(agent,AgentState("summarize SEPE project status"),environment,10)
pd.DataFrame(trace)
```

## Key insight

An agent repeatedly observes, reasons, acts, and updates state until a goal or stopping rule is reached.
