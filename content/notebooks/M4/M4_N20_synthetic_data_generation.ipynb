```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import duplicate_rate,generate_instruction_examples,lexical_diversity,privacy_filter,train_test_overlap
```

```python
rows=generate_instruction_examples(
    ["retrieval-augmented generation","model monitoring","public-sector AI governance"],
    n_per_topic=4,
)
outputs=[row["output"] for row in rows]
{"records":len(rows),
 "duplicate_rate":duplicate_rate(outputs),
 "lexical_diversity":lexical_diversity(outputs)}
```

```python
train=outputs[:8]
test=outputs[8:]+[outputs[0]]
clean,rejected=privacy_filter(
    ["safe synthetic example","contact user@example.com"],
    ["@"],
)
{"train_test_overlap":train_test_overlap(train,test),
 "clean":clean,
 "rejected":rejected}
```

## Key insight

Synthetic data can expand coverage and reduce collection costs, but quality, leakage, bias, and privacy controls remain essential.
