```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import dpo_loss,kl_penalty,pairwise_preference_loss,reward_model_accuracy
```

```python
chosen=np.array([2.5,1.8,3.1,2.2])
rejected=np.array([1.2,1.6,2.0,1.9])
{"pairwise_loss":pairwise_preference_loss(chosen,rejected,.5),
 "reward_accuracy":reward_model_accuracy(chosen,rejected)}
```

```python
policy_chosen=np.log([.7,.6,.8])
policy_rejected=np.log([.2,.3,.1])
reference_chosen=np.log([.6,.55,.7])
reference_rejected=np.log([.25,.35,.15])
{"DPO_loss":dpo_loss(policy_chosen,policy_rejected,
                     reference_chosen,reference_rejected,.1),
 "KL_proxy":kl_penalty(policy_chosen,reference_chosen)}
```

## Key insight

Preference optimization aligns model behavior to comparative judgments while controlling departure from a reference policy.
