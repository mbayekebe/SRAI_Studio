# M4 Notebook 09 — Retrieval Reranking

## Learning objectives

- Combine semantic and lexical signals.
- Rerank initial retrieval candidates.
- Use reciprocal-rank fusion.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import rerank,reciprocal_rank_fusion_results
```

```python
query="farm census livestock"
candidates=[
    {"chunk_id":0,"text":"The agricultural census covers farms, crops, and livestock.","score":.62},
    {"chunk_id":1,"text":"A census is conducted every ten years.","score":.67},
    {"chunk_id":2,"text":"Livestock counts are collected from agricultural holdings.","score":.58},
]
pd.DataFrame(rerank(query,candidates))
```

## Reciprocal-rank fusion

```python
ranking_a=[candidates[1],candidates[0],candidates[2]]
ranking_b=[candidates[0],candidates[2],candidates[1]]
pd.DataFrame(reciprocal_rank_fusion_results([ranking_a,ranking_b]))
```

## Key insight

Reranking improves retrieval quality by applying a stronger relevance model to a smaller candidate set.
