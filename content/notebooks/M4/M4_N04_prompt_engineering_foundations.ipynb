```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import build_prompt,few_shot_prompt,prompt_quality_check
```

```python
prompt=build_prompt(
    role="Senior public-sector data analyst",
    task="Summarize the quarterly performance report.",
    context="The audience is the SEPE steering committee.",
    constraints=["Use plain English","Highlight risks","Do not invent figures"],
    output_format="Five bullets followed by three recommendations.",
)
print(prompt)
print(prompt_quality_check(prompt))
```

## Few-shot prompting

```python
print(few_shot_prompt(
    "Classify each issue as data, process, technology, or governance.",
    [("Missing values","data"),("Approval delay","process")],
    "No named owner for the model"
))
```

## Key insight

Prompt quality improves when instructions are explicit, contextualized, testable, and paired with a clear output contract.
