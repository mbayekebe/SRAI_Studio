## Learning objectives

- Understand resources, tools, prompts, and client–server exchange.
- Build a small MCP-style server manifest.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_genai import MCPPrompt,MCPResource,MCPServerManifest,MCPTool,mcp_request,mcp_response
```

```python
manifest=(
    MCPServerManifest("sepe-context-server","1.0")
    .add_resource(MCPResource(
        "sharepoint://sepe/reports",
        "SEPE Reports",
        "application/pdf",
        "Approved enterprise-monitoring reports",
    ))
    .add_tool(MCPTool(
        "query_enterprise_metrics",
        "Query approved enterprise indicators",
        {
            "type":"object",
            "properties":{"enterprise":{"type":"string"}},
            "required":["enterprise"],
        },
    ))
    .add_prompt(MCPPrompt(
        "executive_brief",
        "Prepare an evidence-based executive brief",
        ["enterprise","period"],
    ))
)
manifest.to_dict()
```

## Protocol exchange

```python
request=mcp_request("tools/list",request_id=7)
response=mcp_response(manifest.to_dict()["tools"],request_id=7)
{"request":request,"response":response}
```

## Key insight

MCP standardizes how AI applications discover and use external tools, resources, and reusable prompts.
