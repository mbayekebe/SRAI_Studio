## Learning objectives

- Build a basic RAG pipeline.
- Retrieve relevant chunks.
- Construct grounded prompts and evaluate retrieval.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import SimpleRAG,grounded_prompt,retrieval_recall_at_k
```

```python
documents=[
    "The agricultural census collects structural information about farms, land, crops, and livestock.",
    "The SEPE dashboard monitors financial and operational performance of public enterprises.",
    "Food-security early warning combines rainfall, production, prices, and household vulnerability.",
]
rag=SimpleRAG(chunk_size=30,overlap=5).fit(documents)
context,hits=rag.build_context("What information does the agricultural census collect?",k=2)
print(context)
print(grounded_prompt("What information does the agricultural census collect?",context))
```

## Retrieval evaluation

```python
retrieved=[h["source_id"] for h in hits]
retrieval_recall_at_k(retrieved,[0],k=2)
```

## Key insight

RAG improves factual grounding by retrieving external evidence before generation.
