```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import aggregate_evaluation,exact_match,groundedness_score,pairwise_preference_rate,token_f1
```

```python
rows=[
    {
        "exact_match":exact_match("The census collects farm data.","The census collects farm data."),
        "token_f1":token_f1("farm census livestock","census farm livestock"),
        "groundedness":groundedness_score(
            "The census collects farm and livestock data.",
            ["The agricultural census collects farm and livestock data."]
        ),
    },
    {
        "exact_match":exact_match("Budget improved.","Revenue improved."),
        "token_f1":token_f1("budget revenue improved","revenue improved"),
        "groundedness":groundedness_score(
            "Revenue improved.",
            ["Revenue increased during the quarter."]
        ),
    },
]
{"aggregate":aggregate_evaluation(rows),
 "pairwise":pairwise_preference_rate([4,3,5],[3,4,2])}
```
