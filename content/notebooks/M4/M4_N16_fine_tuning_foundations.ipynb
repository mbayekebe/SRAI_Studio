## Learning objectives

- Distinguish pretraining, feature extraction, and full fine-tuning.
- Use learning-rate schedules and assess forgetting.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import catastrophic_forgetting_score,learning_rate_schedule,train_linear_head
```

```python
rng=np.random.default_rng(16)
X=rng.normal(size=(500,12))
y=(X[:,0]-X[:,1]+.5*X[:,2]>0).astype(int)
W,b,history=train_linear_head(X,y,learning_rate=.15,epochs=800,l2=.001)
pred=np.argmax(X@W+b,axis=1)
{"training_accuracy":float(np.mean(pred==y)),
 "final_loss":history[-1],
 "forgetting_example":catastrophic_forgetting_score(.91,.84)}
```

```python
steps=np.arange(200)
schedule=[learning_rate_schedule(s,20,200,.001) for s in steps]
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(steps,schedule)
ax.set_title("Warmup and Linear Decay")
plt.show()
```

## Key insight

Fine-tuning adapts pretrained representations, but learning rate, data quality, and forgetting controls determine whether adaptation succeeds.
