# M4 Notebook 08 — Document Ingestion and Chunking

## Learning objectives

- Clean and normalize documents.
- Split documents into retrieval units.
- Compare chunk size and overlap.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import chunk_documents,chunk_statistics,ingest_texts
```

```python
texts=[
    "Agricultural Census\nThe census measures farm structure, crops, livestock, labour, and land use. "*8,
    "Public Enterprise Monitoring\nThe dashboard tracks budgets, revenue, expenditure, debt, and service delivery. "*8,
]
documents=ingest_texts(texts,["census_note","sepe_note"])
rows=[]
for size,overlap in [(30,0),(30,10),(60,10)]:
    chunks=chunk_documents(documents,size,overlap)
    stats=chunk_statistics(chunks)
    stats.update({"chunk_size":size,"overlap":overlap})
    rows.append(stats)
pd.DataFrame(rows)
```

## Chunking trade-offs

Small chunks improve precision but may lose context; large chunks preserve context but introduce irrelevant text.

## Key insight

Chunking is a retrieval design decision, not merely a text-splitting operation.
