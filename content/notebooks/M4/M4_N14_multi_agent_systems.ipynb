## Learning objectives

- Delegate tasks among specialized agents.
- Exchange messages.
- Aggregate reports and resolve disagreement.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_genai import Message,MessageBus,aggregate_reports,consensus_vote,delegate_tasks
```

```python
tasks=[
    "review data quality",
    "review architecture",
    "review governance",
    "review security",
]
agents=["DataAgent","ArchitectureAgent","GovernanceAgent"]
assignments=delegate_tasks(tasks,agents)
assignments
```

```python
bus=MessageBus()
bus.send(Message("DataAgent","GovernanceAgent","Missing data lineage","risk"))
bus.send(Message("ArchitectureAgent","GovernanceAgent","No rollback plan","risk"))
reports={
    "DataAgent":"Data quality acceptable with two lineage gaps.",
    "ArchitectureAgent":"Architecture viable; rollback plan needed.",
    "GovernanceAgent":"Conditional approval recommended.",
}
{"inbox":[m.content for m in bus.inbox("GovernanceAgent")],
 "aggregate":aggregate_reports(reports),
 "vote":consensus_vote(["approve_with_conditions","approve_with_conditions","reject"])}
```

## Key insight

Multi-agent systems coordinate specialization, but require communication protocols, conflict resolution, and shared governance.
