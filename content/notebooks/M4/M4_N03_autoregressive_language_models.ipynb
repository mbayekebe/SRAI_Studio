## Learning objectives

- Train an n-gram language model.
- Compute sequence likelihood and perplexity.
- Explore temperature-based decoding.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_genai import NGramLanguageModel,WordTokenizer,perplexity,temperature_sampling
```

```python
corpus=[
    "data supports decisions",
    "data supports planning",
    "planning supports action",
    "decisions support action",
]
tok=WordTokenizer().fit(corpus)
seqs=[tok.encode(t,False) for t in corpus]
model=NGramLanguageModel(len(tok.itos_),n=2,alpha=.5).fit(seqs)
generated=model.generate([tok.stoi_["data"]],6,seed=3)
{"generated":tok.decode(generated),
 "perplexity":perplexity(model.sequence_log_probability(seqs[0]),len(seqs[0]))}
```

## Decoding

Greedy, sampling, temperature, top-k, and nucleus decoding produce different diversity–reliability trade-offs.

## Key insight

Autoregressive models generate by repeatedly sampling the next token conditioned on prior tokens.
