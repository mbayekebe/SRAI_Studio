# V7 Track A Notebook 09 — Crop-Risk Early Warning

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_applied import alert_level,fuse_risks,prioritize_interventions,production_loss_estimate
```

```python
districts=pd.DataFrame({
    "district":["D1","D2","D3","D4","D5"],
    "drought":[.8,.2,.6,.4,.9],
    "price":[.6,.3,.7,.2,.8],
    "vegetation":[.7,.1,.5,.3,.9],
    "pest":[.2,.1,.8,.2,.6],
    "expected_production":[1000,1200,900,1500,800],
})
districts["risk_score"]=fuse_risks(
    districts.drought,districts.price,districts.vegetation,districts.pest
)
districts["alert_level"]=[alert_level(v) for v in districts.risk_score]
districts["estimated_loss"]=production_loss_estimate(
    districts.expected_production,districts.risk_score
)
prioritize_interventions(districts,capacity=3)
```

## Key insight

Early warning fuses independent risk signals into prioritized, explainable intervention decisions.
