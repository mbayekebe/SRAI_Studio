```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_applied import alert_table,dashboard_readiness,kpi_summary,yield_per_hectare
```

```python
df=pd.DataFrame({
    "region":["Oromia","Amhara","Somali"],
    "crop":["maize","wheat","sorghum"],
    "area":[1200,900,500],
    "production":[3600,1800,400],
    "risk_score":[.25,.45,.82],
})
df=yield_per_hectare(df)
kpis=kpi_summary(df)
alerts=alert_table(df)
{"kpis":kpis,
 "readiness":dashboard_readiness(kpis,.91,8),
 "alerts":alerts}
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.bar(df["region"],df["yield_per_hectare"])
ax.set_ylabel("Yield per hectare")
ax.set_title("Regional Yield Comparison")
plt.show()
```
