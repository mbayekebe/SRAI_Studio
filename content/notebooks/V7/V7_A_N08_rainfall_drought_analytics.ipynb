```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_applied import cumulative_rainfall,drought_class,onset_date,rainfall_anomaly_percent,standardized_precipitation_index
```

```python
rainfall=np.array([5,8,12,25,40,60,75,65,45,25,10,5],float)
climatology=np.array([8,10,15,30,50,70,80,70,50,30,15,8],float)
spi=standardized_precipitation_index(rainfall)
pd.DataFrame({
    "month":np.arange(1,13),
    "rainfall":rainfall,
    "anomaly_percent":rainfall_anomaly_percent(rainfall,climatology),
    "SPI_proxy":spi,
    "class":[drought_class(v) for v in spi],
})
```

```python
{"season_onset_index":onset_date(rainfall,threshold=40,window=3),
 "cumulative_rainfall":cumulative_rainfall(rainfall)}
```

## Key insight

Drought monitoring depends on anomalies relative to local climatology, not rainfall totals alone.
