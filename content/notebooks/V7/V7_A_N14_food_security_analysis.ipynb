```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_applied import food_consumption_score,food_security_class,household_food_expenditure_share,prevalence_by_group,reduced_coping_strategy_index
```

```python
h=pd.DataFrame({"region":["A","A","B","B"],"staples":[7,5,4,7],"pulses":[3,2,1,4],"vegetables":[4,2,1,5],"fruit":[2,1,0,3],"meat":[2,0,0,2],"milk":[1,0,0,3],"sugar":[3,2,1,4],"oil":[4,3,2,5],"food_exp":[70,80,90,50],"total_exp":[100,100,110,100]})
h["FCS"]=[food_consumption_score(*row) for row in h[["staples","pulses","vegetables","fruit","meat","milk","sugar","oil"]].to_numpy()]
h["rCSI"]=reduced_coping_strategy_index([2,4,5,1],[1,2,3,0],[1,3,4,1],[0,2,3,0],[1,3,4,1]); h["food_share"]=household_food_expenditure_share(h.food_exp,h.total_exp)
h["class"]=[food_security_class(f,r,s) for f,r,s in zip(h.FCS,h.rCSI,h.food_share)]
{"households":h,"prevalence":prevalence_by_group(h,"region","class")}
```
