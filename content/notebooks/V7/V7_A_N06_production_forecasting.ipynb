```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_applied import ensemble_forecast,forecast_interval,mae,moving_average_forecast,seasonal_naive,trend_forecast
```

```python
rng=np.random.default_rng(6)
years=np.arange(2005,2025)
production=100+3*(years-years[0])+8*np.sin(2*np.pi*np.arange(len(years))/4)+rng.normal(0,4,len(years))
train=production[:-4]; actual=production[-4:]
f1=trend_forecast(train,4)
f2=moving_average_forecast(train,4,3)
forecast=ensemble_forecast([f1,f2],[.7,.3])
lo,hi=forecast_interval(forecast,np.std(train-trend_forecast(train,0 if False else len(train))[:len(train)]))
{"MAE":mae(actual,forecast),"forecast":forecast,"lower":lo,"upper":hi}
```

```python
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(years,production,label="actual")
ax.plot(years[-4:],forecast,label="forecast")
ax.fill_between(years[-4:],lo,hi,alpha=.25)
ax.legend(); ax.set_title("Agricultural Production Forecast")
plt.show()
```

## Key insight

Operational crop forecasting should combine transparent baselines, uncertainty intervals, and backtesting.
