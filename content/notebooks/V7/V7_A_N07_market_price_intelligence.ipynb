```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_applied import food_basket_index,market_integration_correlation,market_spread,price_index,rolling_volatility,shock_flags
```

```python
months=pd.date_range("2024-01-01",periods=24,freq="MS")
producer=np.linspace(100,140,24)+np.sin(np.arange(24))*4
retail=producer+25+np.random.default_rng(7).normal(0,2,24)
frame=pd.DataFrame({
    "month":months,"producer":producer,"retail":retail,
    "spread":market_spread(producer,retail),
    "price_index":price_index(retail,3),
    "volatility":rolling_volatility(retail,4),
    "shock":shock_flags(retail),
})
{"integration":market_integration_correlation(producer,retail),"latest":frame.tail()}
```
