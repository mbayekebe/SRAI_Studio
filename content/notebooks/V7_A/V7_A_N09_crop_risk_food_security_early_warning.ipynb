## Decision contract
Detect converging crop, rainfall, market, and nutrition signals for authorized verification and preparedness. The system does not declare famine, target households, or release stocks automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7609);n=48
df=pd.DataFrame({'district':[f'D{i:02d}' for i in range(n)],'rainfall_pct':rng.normal(92,18,n),'ndvi_z':rng.normal(-.3,.9,n),'price_z':rng.normal(.5,1.1,n),'nutrition_z':rng.normal(.2,.8,n),'report_age':rng.integers(0,18,n)})
df.loc[:5,['rainfall_pct','ndvi_z','price_z','nutrition_z']]=[55,-2.0,2.4,1.8];df.head()
```

## Evidence and quality contract
Indicators have different meanings, lags, coverage, denominators, and revisions. Preserve source, reference period, geography, unit, quality flag, and release status.

```python
df['quality']=df.report_age<=10
df['signal_count']=(df.rainfall_pct<70).astype(int)+(df.ndvi_z<-1.5).astype(int)+(df.price_z>2).astype(int)+(df.nutrition_z>1.5).astype(int)
df['severity']=df.signal_count+(.5*(df.price_z>2.5))
print(df.signal_count.value_counts().sort_index().to_string())
```

## Uncertainty, sensitivity, and abstention
Compare a two-signal and three-signal escalation rule. Stale evidence causes abstention. Concordant indicators strengthen a case for verification, not certainty about cause or household condition.

```python
df['status']=np.where(~df.quality,'ABSTAIN—UPDATE EVIDENCE',np.where(df.signal_count>=3,'URGENT MULTI-AGENCY VERIFICATION',np.where(df.signal_count>=2,'VERIFY','MONITOR')))
result=df[df.status!='MONITOR'][['district','signal_count','severity','report_age','status']].sort_values('severity',ascending=False)
print(result.head(12).to_string(index=False));print({'alerts_at_2':int((df.signal_count>=2).sum()),'alerts_at_3':int((df.signal_count>=3).sum())})
```
