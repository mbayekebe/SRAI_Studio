## Decision contract
Integrate agricultural census, surveys, production, markets, climate, livestock, food-security, platform, and executive-review evidence into a controlled national decision service. The capstone produces an auditable decision brief; it does not replace official estimates, declarations, budgets, procurement, or ministerial authority.


```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7115); regions=[f'R{i:02d}' for i in range(18)]; df=pd.DataFrame({'region':regions,'production_index':rng.normal(100,12,18),'market_pressure':rng.uniform(.7,1.45,18),'drought_risk':rng.uniform(0,1,18),'food_insecurity':rng.uniform(.08,.48,18),'data_quality':rng.uniform(.72,1,18),'effective_n':rng.integers(25,320,18),'response_capacity':rng.uniform(.35,1,18)}); df.head()
```

## Evidence and quality contract
Register each source, concept, unit, reference period, geography, weight, revision, quality gate, model version, interval, scenario, affected group, owner, intervention capacity, approval, incident, and retirement condition. Preserve a model card, audit trail, executive brief, and reproducibility record.


```python
df['evidence_ok']=(df.data_quality>=.82)&(df.effective_n>=50); df['risk_signal']=.30*(1-df.production_index/120).clip(0,1)+.25*((df.market_pressure-1)/.45).clip(0,1)+.25*df.drought_risk+.20*(df.food_insecurity/.5).clip(0,1); print(df[['risk_signal','data_quality','effective_n']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Source disagreement, survey error, forecast error, spatial uncertainty, missing markets, and threshold costs must remain visible. Compare baselines and scenarios, test sensitivity, document error trade-offs, and abstain when action cannot be justified.


```python
df['decision_status']=np.where(~df.evidence_ok,'ABSTAIN - EVIDENCE REVIEW',np.where((df.risk_signal>=.55)&(df.response_capacity>=.5),'AUTHORITY VERIFICATION QUEUE',np.where(df.risk_signal>=.55,'CAPACITY ESCALATION','MONITOR'))); brief=df[df.decision_status!='MONITOR'].sort_values('risk_signal',ascending=False); model_card={'purpose':'regional agricultural risk review','baseline':'transparent weighted composite','limitations':['synthetic data','no causal claim','no autonomous action'],'owner':'National agricultural intelligence board'}; audit={'records':len(df),'queued':len(brief),'abstained':int((df.decision_status.str.startswith('ABSTAIN')).sum())}; print(brief.round(3).to_string(index=False)); print(model_card); print(audit)
```
