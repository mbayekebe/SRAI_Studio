## Decision contract
Assemble a decision-room view of agricultural indicators, uncertainty, data quality, and accountable follow-up. The dashboard opens questions and assigns review; it does not declare emergencies, set prices, allocate budgets, or replace official releases.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7105); df=pd.DataFrame({'indicator':['production','yield','market_pressure','drought','food_security'],'estimate':[112,2.4,1.18,.37,.22],'lower':[101,2.1,1.05,.28,.16],'upper':[124,2.7,1.31,.48,.30],'fresh_days':[18,18,5,3,12],'quality':[.94,.91,.89,.86,.83],'owner':['NSO','NSO','MARKET','EWS','FSU']}); df
```

## Evidence and quality contract
Preserve indicator definitions, denominators, geographic and temporal grain, source date, revision status, interval, scenario, quality flag, decision owner, response capacity, escalation state, and release version.


```python
df['relative_width']=(df.upper-df.lower)/df.estimate.abs(); df['usable']=(df.fresh_days<=30)&(df.quality>=.85)&(df.relative_width<=.45); print(df[['indicator','estimate','lower','upper','quality','usable']].round(3).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Traffic-light summaries can hide magnitude, precision, and disagreement. Display intervals and freshness, compare scenarios, and abstain when evidence cannot support the requested decision horizon.


```python
df['status']=np.where(~df.usable,'ABSTAIN - VERIFY',np.where((df.indicator.isin(['market_pressure','drought']))&(df.estimate>.3),'OWNER REVIEW','MONITOR')); queue=df[df.status!='MONITOR']; print(queue[['indicator','owner','status','fresh_days','relative_width']].round(3).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
