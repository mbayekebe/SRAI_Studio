## Decision contract
Support livestock population, feed, water, disease-surveillance, and mobility planning at aggregate level. The output does not order culling, movement restrictions, treatment, compensation, or enforcement.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7713);n=38;df=pd.DataFrame({'district':[f'L{i:02d}' for i in range(n)],'cattle':rng.integers(3000,90000,n),'shoats':rng.integers(5000,180000,n),'feed_tonnes':rng.integers(5000,120000,n),'water_index':rng.uniform(.2,1,n),'coverage':rng.uniform(.55,1,n),'report_age':rng.integers(0,20,n)});df.head()
```

## Evidence and quality contract
Separate species, production system, ownership/management, resident/mobile herds, reference date, births, deaths, sales, slaughter, feed/water access, vaccination, and coverage.

```python
df['tlu']=df.cattle+.1*df.shoats;df['feed_need']=df.tlu*1.8;df['feed_gap']=(df.feed_need-df.feed_tonnes).clip(lower=0);df['quality']=(df.coverage>=.75)&(df.report_age<=12);print(df[['tlu','feed_gap','water_index']].describe().round(1))
```

## Uncertainty, sensitivity, and abstention
Herds are mobile and counts are seasonal. Compare normal and drought feed requirements; stale or low-coverage reporting causes abstention.

```python
df['drought_gap']=(df.feed_need*1.25-df.feed_tonnes).clip(lower=0);df['priority']=df.drought_gap*(1.2-df.water_index);df['status']=np.where(~df.quality,'ABSTAIN—VERIFY COVERAGE',np.where(df.priority>=df.priority.quantile(.75),'LIVESTOCK SUPPORT REVIEW','MONITOR'));result=df[df.status!='MONITOR'].sort_values('priority',ascending=False)[['district','tlu','feed_gap','drought_gap','water_index','coverage','priority','status']];print(result.head(15).round(1).to_string(index=False))
```
