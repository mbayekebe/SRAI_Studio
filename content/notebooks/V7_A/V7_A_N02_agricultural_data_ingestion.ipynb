## Decision contract
Build a replayable intake and quarantine workflow for survey, administrative, market, weather, and geospatial files. Ingestion does not certify truth or authorize publication.

```python
import pandas as pd,hashlib,io
raw=[('survey_2026.csv','survey','2026Q1',1200,'v2'),('market_2026.csv','market','2026Q1',840,'v1'),('market_2026_copy.csv','market','2026Q1',840,'v1'),('weather.csv','weather','2026Q1',0,'v3'),('admin.csv','admin','2025Q4',650,'v1')]
df=pd.DataFrame(raw,columns=['file','source','period','rows','schema']);df['checksum']=[hashlib.sha256(f'{r.source}|{r.period}|{r.rows}|{r.schema}'.encode()).hexdigest() for _,r in df.iterrows()];df
```

## Evidence and quality contract
Every delivery needs source, owner, reference period, geography, schema/version, units, confidentiality class, checksum, receipt time, correction lineage, and disposition.

```python
df['duplicate']=df.duplicated('checksum',keep='first');df['late']=df.period<'2026Q1';df['empty']=df.rows==0;df['quality']=~(df.duplicate|df.empty);print(df[['file','duplicate','late','empty','quality']].to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Schema drift, duplicates, late files, and partial coverage must be visible. Failed records go to quarantine; they are never silently discarded or coerced.

```python
df['status']=df.apply(lambda r:'QUARANTINE—DUPLICATE' if r.duplicate else ('QUARANTINE—EMPTY DELIVERY' if r.empty else ('ACCEPT WITH LATE FLAG' if r.late else 'ACCEPT FOR VALIDATION')),axis=1);result=df[['file','source','period','schema','checksum','status']];print(result.to_string(index=False))
```
