## Learning objectives

- Understand bagging and feature subsampling.
- Compare single trees with forests.
- Study variance reduction.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    DecisionTreeClassifier,RandomForestClassifier,accuracy_score,train_test_split,
)
```

```python
rng=np.random.default_rng(7)
X=rng.normal(size=(500,8))
y=(X[:,0]*X[:,1]+.5*X[:,2]-X[:,3]>0).astype(int)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=7)
tree=DecisionTreeClassifier(max_depth=8,seed=7).fit(Xtr,ytr)
forest=RandomForestClassifier(n_estimators=80,max_depth=7,seed=7).fit(Xtr,ytr)
{"tree_test_accuracy":accuracy_score(yte,tree.predict(Xte)),
 "forest_test_accuracy":accuracy_score(yte,forest.predict(Xte))}
```

## Ensemble-size stability

```python
sizes=[1,5,10,20,40,80]
scores=[]
for n in sizes:
    m=RandomForestClassifier(n_estimators=n,max_depth=7,seed=7).fit(Xtr,ytr)
    scores.append(accuracy_score(yte,m.predict(Xte)))
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(sizes,scores,marker="o")
ax.set_xlabel("Number of trees"); ax.set_ylabel("Test accuracy")
ax.set_title("Random Forest Ensemble Size")
plt.show()
```

## Why forests work

Bootstrap resampling reduces variance; random feature subsets decorrelate trees.

## Decision Intelligence case

Forests can improve robustness, but explanations require feature importance or local interpretation tools.

## Key insight

Random forests trade some interpretability for substantially improved stability and predictive performance.
