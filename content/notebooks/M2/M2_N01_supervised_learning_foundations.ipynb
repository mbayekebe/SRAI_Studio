```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import LinearRegression,mean_squared_error,train_test_split
```

## Supervised learning

Given examples \((x_i,y_i)\), learn a function

\[
\hat f:\mathcal X\to\mathcal Y
\]

that predicts unseen targets.


```python
rng=np.random.default_rng(42)
X=np.linspace(-3,3,120)[:,None]
y=2+1.5*X[:,0]+rng.normal(0,1,120)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.25,seed=42)
model=LinearRegression().fit(X_train,y_train)
{"train_MSE":mean_squared_error(y_train,model.predict(X_train)),
 "test_MSE":mean_squared_error(y_test,model.predict(X_test))}
```

## Generalization and data leakage

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.scatter(X_train[:,0],y_train,label="train")
ax.scatter(X_test[:,0],y_test,label="test")
grid=np.linspace(-3,3,200)[:,None]
ax.plot(grid[:,0],model.predict(grid),label="model")
ax.legend(); ax.set_title("Train/Test Generalization")
plt.show()
```

## Bias–variance intuition

Simple models may underfit; overly flexible models may fit noise.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression as SKLinearRegression
degrees=[1,3,10]
rows=[]
for d in degrees:
    m=make_pipeline(PolynomialFeatures(d),SKLinearRegression()).fit(X_train,y_train)
    rows.append({"degree":d,
                 "train_MSE":mean_squared_error(y_train,m.predict(X_train)),
                 "test_MSE":mean_squared_error(y_test,m.predict(X_test))})
pd.DataFrame(rows)
```
