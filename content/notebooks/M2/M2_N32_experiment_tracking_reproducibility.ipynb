```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import ExperimentTracker,LinearRegression,RidgeRegression,compare_runs,dataset_signature,mean_squared_error,reproducibility_manifest
from pathlib import Path
import platform
```

```python
rng=np.random.default_rng(32)
X=rng.normal(size=(300,8)); y=1+X@np.array([2,-1,.5,0,0,0,0,0])+rng.normal(size=300)
run_dir=Path('artifacts')/'m2_r7_experiments'; paths=[]
for name,model in [('linear',LinearRegression()),('ridge',RidgeRegression(alpha=1.0))]:
 model.fit(X,y); tr=ExperimentTracker(name,run_dir); tr.log_params(model=name); tr.log_metrics(negative_mse=-mean_squared_error(y,model.predict(X))); paths.append(tr.finish())
[Path(p).name for p in paths]
```

```python
compare_runs(paths[0],paths[1],'negative_mse')
```

```python
reproducibility_manifest(32,platform.python_version(),{'numpy':np.__version__},dataset_signature(X,y))
```
