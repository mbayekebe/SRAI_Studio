```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    LinearDiscriminantAnalysis,QuadraticDiscriminantAnalysis,
    accuracy_score,train_test_split,
)
```

```python
rng=np.random.default_rng(12)
X0=rng.multivariate_normal([-1,0],[[1,.7],[.7,1]],250)
X1=rng.multivariate_normal([1,0],[[1,-.4],[-.4,.5]],250)
X=np.r_[X0,X1]; y=np.r_[np.zeros(250,int),np.ones(250,int)]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=12)
models={
    "LDA":LinearDiscriminantAnalysis().fit(Xtr,ytr),
    "QDA":QuadraticDiscriminantAnalysis().fit(Xtr,ytr),
}
pd.Series({name:accuracy_score(yte,m.predict(Xte)) for name,m in models.items()})
```

## Decision boundaries

```python
xx,yy=np.meshgrid(np.linspace(-4,4,200),np.linspace(-4,4,200))
grid=np.c_[xx.ravel(),yy.ravel()]
fig,ax=plt.subplots(figsize=(7,5))
ax.contour(xx,yy,models["QDA"].predict(grid).reshape(xx.shape),levels=[.5])
ax.scatter(Xte[:,0],Xte[:,1],c=yte,s=12)
ax.set_title("Quadratic Discriminant Boundary")
plt.show()
```

## LDA versus QDA

LDA assumes shared covariance; QDA allows class-specific covariance.

## Decision Intelligence case

Discriminant analysis is useful when class distributions are approximately Gaussian and interpretability matters.

## Key insight

The covariance assumptions determine whether the decision boundary is linear or quadratic.
