```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import drift_alert,ks_drift_test,label_shift,population_stability_index,rolling_metric
```

```python
rng=np.random.default_rng(33)
reference=rng.normal(0,1,5000); current=rng.normal(.35,1.15,5000)
psi=population_stability_index(reference,current,10); ks,p=ks_drift_test(reference,current)
{'PSI':psi,'KS':ks,'p_value':p,'alert':drift_alert(psi,.1,.25)}
```

```python
label_shift(rng.binomial(1,.15,2000),rng.binomial(1,.28,2000))
```

```python
daily=np.r_[rng.normal(.90,.01,60),rng.normal(.80,.02,40)]; rolling=rolling_metric(daily,10)
fig,ax=plt.subplots(figsize=(8,4)); ax.plot(np.arange(9,len(daily)),rolling); ax.axhline(.85,linestyle='--'); ax.set_title('Performance Drift'); plt.show()
```
