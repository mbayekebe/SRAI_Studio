## Learning objectives

- Apply PCA and random projection.
- Quantify explained variance and reconstruction error.
- Use reduced representations in downstream models.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import PCA,RandomProjection
```

```python
rng=np.random.default_rng(14)
latent=rng.normal(size=(500,2))
W=rng.normal(size=(2,10))
X=latent@W+rng.normal(scale=.2,size=(500,10))
pca=PCA(2).fit(X)
Z=pca.transform(X)
{"explained_variance_ratio":pca.explained_variance_ratio_,
 "reconstruction_MSE":np.mean((X-pca.inverse_transform(Z))**2)}
```

```python
fig,ax=plt.subplots(figsize=(7,5))
ax.scatter(Z[:,0],Z[:,1],s=10)
ax.set_title("PCA Representation")
plt.show()
```

## Random projection

```python
rp=RandomProjection(3,seed=14)
R=rp.fit_transform(X)
{"original_shape":X.shape,"projected_shape":R.shape}
```

## Leakage warning

Fit dimensionality-reduction transformations only on training data, then apply them to validation and test data.

## Decision Intelligence case

Compress large administrative indicator sets while retaining transparent reporting of the original indicators.

Dimensionality reduction can improve computation and visualization, but it may obscure substantive meaning.
