```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    LinearRegression,OneHotEncoder,PolynomialFeatures,interaction_terms,
    log_transform,mean_squared_error,median_impute,missing_indicator,
    train_test_split,
)
```

## Polynomial features

```python
rng=np.random.default_rng(16)
x=np.linspace(-3,3,300)[:,None]
y=1+2*x[:,0]-0.8*x[:,0]**2+rng.normal(scale=.8,size=300)
Xtr,Xte,ytr,yte=train_test_split(x,y,test_size=.25,seed=16)
rows=[]
for degree in [1,2,5]:
    pf=PolynomialFeatures(degree)
    A=pf.fit_transform(Xtr); B=pf.transform(Xte)
    m=LinearRegression().fit(A,ytr)
    rows.append({"degree":degree,"test_MSE":mean_squared_error(yte,m.predict(B))})
pd.DataFrame(rows)
```

## Categorical encoding

```python
categories=np.array([["Urban"],["Rural"],["Peri-urban"],["Urban"]],object)
encoder=OneHotEncoder()
encoder.fit_transform(categories)
```

## Missing-value indicators

```python
X=np.array([[1,np.nan],[2,5],[np.nan,7],[4,9]],float)
indicator=missing_indicator(X)
imputed,medians=median_impute(X)
{"imputed":imputed,"missing_indicator":indicator,"medians":medians}
```

## Leakage warning

Fit encoders and imputers on training data only; reuse learned parameters on validation and test data.

## Decision Intelligence case

Feature engineering should reflect domain mechanisms, not merely generate large numbers of variables.
