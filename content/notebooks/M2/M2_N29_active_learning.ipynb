## Learning objectives

- Select informative cases for labeling.
- Compare entropy, margin, and least-confidence sampling.
- Simulate an active-learning loop.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    GaussianNB,active_learning_round,accuracy_score,entropy_uncertainty,
    query_top_k,
)
```

```python
rng=np.random.default_rng(29)
X=np.r_[rng.normal([-1,-1],1,(300,2)),
        rng.normal([1,1],1,(300,2))]
y=np.r_[np.zeros(300,int),np.ones(300,int)]
initial=np.r_[np.arange(10),np.arange(300,310)]
labeled=initial.copy()
history=[]
for round_no in range(8):
    model,chosen=active_learning_round(
        lambda:GaussianNB(),X,y,labeled,query_size=20,strategy="entropy"
    )
    labeled=np.unique(np.r_[labeled,chosen])
    history.append((len(labeled),accuracy_score(y,model.predict(X))))
pd.DataFrame(history,columns=["labeled_examples","accuracy"])
```

```python
frame=pd.DataFrame(history,columns=["labeled_examples","accuracy"])
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(frame.labeled_examples,frame.accuracy,marker="o")
ax.set_xlabel("Labeled examples"); ax.set_ylabel("Accuracy")
ax.set_title("Active-Learning Progress")
plt.show()
```

## Query strategies

```python
example=np.array([[.5,.5],[.8,.2],[.55,.45],[.99,.01]])
for strategy in ["entropy","margin","least_confidence"]:
    print(strategy,query_top_k(example,2,strategy)[0])
```

## Decision Intelligence case

Active learning directs scarce expert attention toward ambiguous or high-value records.

## Key insight

Active learning optimizes which labels to acquire, not only how to learn from labels already available.
