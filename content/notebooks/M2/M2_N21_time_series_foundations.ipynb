```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    autocorrelation,difference,lag_matrix,rolling_mean,seasonal_naive_forecast,
    time_series_split,
)
```

```python
rng=np.random.default_rng(21)
t=np.arange(240)
series=20+0.05*t+3*np.sin(2*np.pi*t/12)+rng.normal(scale=1,size=len(t))
fig,ax=plt.subplots(figsize=(9,4))
ax.plot(series)
ax.set_title("Trend and Seasonality")
plt.show()
```

## Differencing and autocorrelation

```python
diff=difference(series,12)
acf=autocorrelation(diff,24)
pd.Series(acf,index=np.arange(len(acf))).head(13)
```

```python
fig,ax=plt.subplots(figsize=(8,4))
ax.bar(np.arange(len(acf)),acf)
ax.set_xlabel("Lag"); ax.set_ylabel("Autocorrelation")
ax.set_title("Autocorrelation after Seasonal Differencing")
plt.show()
```
