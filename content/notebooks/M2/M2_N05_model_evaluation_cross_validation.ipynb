## Learning objectives

- Use k-fold cross-validation.
- Select metrics that match the task.
- Build learning curves and avoid test-set misuse.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    LinearRegression,cross_val_score,learning_curve,mean_absolute_error,
    mean_squared_error,root_mean_squared_error,
)
```

## Cross-validation

```python
rng=np.random.default_rng(21)
X=rng.normal(size=(300,5))
y=1+X@np.array([2,-1,.5,0,1.5])+rng.normal(scale=1,size=300)
scores=cross_val_score(lambda:LinearRegression(),X,y,root_mean_squared_error,n_splits=5,seed=21)
{"fold_scores":scores,"mean":scores.mean(),"std":scores.std(ddof=1)}
```

## Metric choice

```python
pred=y+rng.normal(scale=1,size=len(y))
pd.Series({"MSE":mean_squared_error(y,pred),
           "RMSE":root_mean_squared_error(y,pred),
           "MAE":mean_absolute_error(y,pred)})
```

## Learning curves

```python
curve=learning_curve(lambda:LinearRegression(),X,y,
                         root_mean_squared_error,[40,80,120,180,240,300],seed=21)
frame=pd.DataFrame(curve,columns=["train_size","train_RMSE","validation_RMSE"])
frame
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(frame.train_size,frame.train_RMSE,marker="o",label="train")
ax.plot(frame.train_size,frame.validation_RMSE,marker="o",label="validation")
ax.set_xlabel("Training examples"); ax.set_ylabel("RMSE")
ax.set_title("Learning Curve"); ax.legend()
plt.show()
```

## Evaluation protocol

Use training data for fitting, validation data for tuning, and the test set once for final assessment.

## Decision Intelligence case

Report uncertainty across folds rather than a single favorable score.

## Key insight

Evaluation design is part of the model; weak evaluation can make a poor model look strong.
