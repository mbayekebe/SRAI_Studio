```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    DecisionTreeClassifier,DecisionTreeRegressor,accuracy_score,gini_impurity,
    mean_squared_error,train_test_split,
)
```

## Classification impurity

\[
G=1-\sum_k p_k^2.
\]


```python
for labels in ([0,0,0,0],[0,0,1,1],[0,1,2,2]):
    print(labels,gini_impurity(labels))
```

```python
rng=np.random.default_rng(42)
X=rng.normal(size=(300,2))
y=((X[:,0]**2+X[:,1])>0.5).astype(int)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=42)
rows=[]
for depth in [1,2,3,5,10]:
    m=DecisionTreeClassifier(max_depth=depth).fit(Xtr,ytr)
    rows.append({"depth":depth,"train_accuracy":accuracy_score(ytr,m.predict(Xtr)),
                 "test_accuracy":accuracy_score(yte,m.predict(Xte))})
pd.DataFrame(rows)
```

## Regression trees

```python
xr=np.linspace(-3,3,200)[:,None]
yr=np.sin(xr[:,0])+rng.normal(scale=.15,size=200)
m=DecisionTreeRegressor(max_depth=4).fit(xr,yr)
fig,ax=plt.subplots(figsize=(7,4))
ax.scatter(xr[:,0],yr,s=10)
ax.plot(xr[:,0],m.predict(xr))
ax.set_title("Piecewise-Constant Regression Tree")
plt.show()
```
