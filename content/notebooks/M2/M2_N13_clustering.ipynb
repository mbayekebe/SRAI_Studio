## Learning objectives

- Apply K-means and hierarchical clustering.
- Evaluate clusters with silhouette score.
- Understand cluster instability.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    AgglomerativeClustering,KMeans,silhouette_score,
)
```

```python
rng=np.random.default_rng(13)
X=np.r_[rng.normal([-2,-2],.5,(120,2)),
        rng.normal([2,0],.6,(120,2)),
        rng.normal([0,3],.4,(120,2))]
rows=[]
models={}
for k in [2,3,4,5]:
    m=KMeans(k,seed=13).fit(X)
    models[k]=m
    rows.append({"k":k,"inertia":m.inertia_,"silhouette":silhouette_score(X,m.labels_)})
pd.DataFrame(rows)
```

```python
m=models[3]
fig,ax=plt.subplots(figsize=(7,5))
ax.scatter(X[:,0],X[:,1],c=m.labels_,s=15)
ax.scatter(m.cluster_centers_[:,0],m.cluster_centers_[:,1],marker="x",s=120)
ax.set_title("K-Means Clusters")
plt.show()
```

## Hierarchical clustering

```python
small=X[:40]
h=AgglomerativeClustering(3).fit(small)
np.bincount(h.labels_)
```
