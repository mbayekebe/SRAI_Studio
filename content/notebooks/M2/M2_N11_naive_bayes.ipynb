```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    GaussianNB,MultinomialNB,accuracy_score,confusion_matrix,train_test_split,
)
```

## Naive Bayes rule

\[
P(y\mid x_1,\ldots,x_p)
\propto
P(y)\prod_jP(x_j\mid y).
\]


```python
rng=np.random.default_rng(11)
X=np.r_[rng.normal([-1,-1],.8,(200,2)),rng.normal([1,1],.8,(200,2))]
y=np.r_[np.zeros(200,int),np.ones(200,int)]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=11)
m=GaussianNB().fit(Xtr,ytr)
pred=m.predict(Xte)
{"accuracy":accuracy_score(yte,pred)}
```

```python
pd.DataFrame(confusion_matrix(yte,pred),
             index=["Actual 0","Actual 1"],columns=["Predicted 0","Predicted 1"])
```

## Multinomial Naive Bayes

```python
X=np.array([[4,1,0],[3,2,0],[0,1,4],[0,0,5]],float)
y=np.array([0,0,1,1])
m=MultinomialNB(alpha=1).fit(X,y)
{"training_accuracy":accuracy_score(y,m.predict(X))}
```

Naive Bayes can perform strongly even when its independence assumption is only approximately true.
