## Learning objectives

- Learn from limited labels and abundant unlabeled data.
- Apply pseudo-labeling and self-training.
- Understand confirmation bias.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    GaussianNB,SelfTrainingClassifier,accuracy_score,nearest_centroid_labeling,
)
```

```python
rng=np.random.default_rng(28)
X=np.r_[rng.normal([-1,-1],.7,(200,2)),
        rng.normal([1,1],.7,(200,2))]
true_y=np.r_[np.zeros(200,int),np.ones(200,int)]
observed=np.full(len(true_y),-1)
labeled_idx=np.r_[np.arange(15),np.arange(200,215)]
observed[labeled_idx]=true_y[labeled_idx]
self_train=SelfTrainingClassifier(lambda:GaussianNB(),threshold=.8,max_iter=10).fit(X,observed)
pred=self_train.predict(X)
{"labeled_initially":len(labeled_idx),
 "pseudo_label_rounds":self_train.history_,
 "final_accuracy":accuracy_score(true_y,pred)}
```

## Nearest-centroid pseudo-labeling

```python
centroid_labels=nearest_centroid_labeling(X,observed)
{"accuracy":accuracy_score(true_y,centroid_labels)}
```

## Failure modes

Incorrect early pseudo-labels can reinforce themselves. Confidence does not guarantee correctness.

## Decision Intelligence case

Semi-supervised methods can reduce labeling cost, but high-impact decisions require rigorous review of pseudo-label errors.

## Key insight

Unlabeled data help only when the structure of the input distribution aligns with the target task.
