## Learning objectives

- Diagnose class imbalance.
- Use class weights and resampling.
- Evaluate with precision–recall metrics.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    LogisticRegressionGD,accuracy_score,average_precision,balanced_accuracy,
    class_weights,precision_recall_curve,random_oversample,train_test_split,
)
```

```python
rng=np.random.default_rng(17)
X0=rng.normal(0,1,(950,2))
X1=rng.normal(2,1,(50,2))
X=np.r_[X0,X1]; y=np.r_[np.zeros(950,int),np.ones(50,int)]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=17)
baseline=LogisticRegressionGD(.1,3000).fit(Xtr,ytr)
prob=baseline.predict_proba(Xte)[:,1]
pred=baseline.predict(Xte)
{"accuracy":accuracy_score(yte,pred),
 "balanced_accuracy":balanced_accuracy(yte,pred),
 "average_precision":average_precision(yte,prob)}
```

## Random oversampling

```python
Xo,yo=random_oversample(Xtr,ytr,seed=17)
oversampled=LogisticRegressionGD(.1,3000).fit(Xo,yo)
pred_o=oversampled.predict(Xte)
{"class_weights":class_weights(ytr),
 "oversampled_balanced_accuracy":balanced_accuracy(yte,pred_o)}
```

## Precision–recall curve

```python
precision,recall,thresholds=precision_recall_curve(yte,prob)
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(recall,precision)
ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
ax.set_title("Precision–Recall Curve")
plt.show()
```
