```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    AdaBoostStumpClassifier,GradientBoostingRegressor,accuracy_score,
    mean_squared_error,train_test_split,
)
```

## Gradient boosting

At each stage, fit a weak learner to residuals or negative gradients.


```python
rng=np.random.default_rng(8)
X=np.linspace(-3,3,300)[:,None]
y=np.sin(X[:,0])+0.3*np.cos(3*X[:,0])+rng.normal(scale=.12,size=300)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,seed=8)
rows=[]
for lr in [.02,.05,.1,.2]:
    m=GradientBoostingRegressor(120,lr,2,seed=8).fit(Xtr,ytr)
    rows.append({"learning_rate":lr,"test_MSE":mean_squared_error(yte,m.predict(Xte))})
pd.DataFrame(rows)
```

```python
m=GradientBoostingRegressor(120,.05,2,seed=8).fit(Xtr,ytr)
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(m.train_loss_)
ax.set_xlabel("Boosting stage"); ax.set_ylabel("Training MSE")
ax.set_title("Gradient Boosting Learning Curve")
plt.show()
```

## AdaBoost

```python
Xc=rng.normal(size=(300,2))
yc=(Xc[:,0]+Xc[:,1]>.5).astype(int)
Xtr,Xte,ytr,yte=train_test_split(Xc,yc,test_size=.3,seed=8)
ada=AdaBoostStumpClassifier(30,.8).fit(Xtr,ytr)
{"test_accuracy":accuracy_score(yte,ada.predict(Xte)),
 "estimators_used":len(ada.models_)}
```
