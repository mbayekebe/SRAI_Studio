```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    RidgeRegression,StandardScaler,mean_squared_error,train_test_split,
)
```

## Ridge objective

\[
\min_\beta
\|y-X\beta\|_2^2+\alpha\|\beta\|_2^2.
\]


```python
rng=np.random.default_rng(11)
z=rng.normal(size=(300,1))
X=np.column_stack([z[:,0],z[:,0]+rng.normal(scale=.05,size=300),
                   rng.normal(size=300),100*rng.normal(size=300)])
y=3+2*X[:,0]-2*X[:,1]+.5*X[:,2]+rng.normal(scale=1,size=300)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,seed=11)
scaler=StandardScaler()
Xtr_s=scaler.fit_transform(Xtr); Xte_s=scaler.transform(Xte)
rows=[]
for alpha in [0,.1,1,10,100]:
    m=RidgeRegression(alpha=alpha).fit(Xtr_s,ytr)
    rows.append({"alpha":alpha,"test_MSE":mean_squared_error(yte,m.predict(Xte_s)),
                 "coef_norm":np.linalg.norm(m.coef_)})
pd.DataFrame(rows)
```

## Coefficient paths

```python
alphas=np.logspace(-3,3,50)
coefs=np.array([RidgeRegression(a).fit(Xtr_s,ytr).coef_ for a in alphas])
fig,ax=plt.subplots(figsize=(7,4))
for j in range(coefs.shape[1]):
    ax.semilogx(alphas,coefs[:,j],label=f"β{j+1}")
ax.set_xlabel("alpha"); ax.set_ylabel("coefficient")
ax.set_title("Ridge Coefficient Paths"); ax.legend()
plt.show()
```

## Why scaling matters

Regularization acts on coefficient size, so feature units directly affect the penalty.

## Decision Intelligence case

Regularization stabilizes models built from correlated administrative indicators.

## Key insight

Regularization trades a small amount of bias for lower variance and more stable prediction.
