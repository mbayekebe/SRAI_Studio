## Learning objectives

- Understand instance-based learning.
- Study distance metrics and feature scaling.
- Select the number of neighbors.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    KNNClassifier,StandardScaler,accuracy_score,train_test_split,
)
```

```python
rng=np.random.default_rng(10)
X=rng.normal(size=(400,2))
y=((X[:,0]**2+X[:,1]**2)>1).astype(int)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=10)
rows=[]
for k in [1,3,5,11,25,51]:
    m=KNNClassifier(k).fit(Xtr,ytr)
    rows.append({"k":k,"train_accuracy":accuracy_score(ytr,m.predict(Xtr)),
                 "test_accuracy":accuracy_score(yte,m.predict(Xte))})
pd.DataFrame(rows)
```

## Feature scaling matters

```python
X_bad=X.copy(); X_bad[:,1]*=100
Xtr,Xte,ytr,yte=train_test_split(X_bad,y,test_size=.3,seed=10)
raw=KNNClassifier(7).fit(Xtr,ytr)
scaler=StandardScaler(); Xtr_s=scaler.fit_transform(Xtr); Xte_s=scaler.transform(Xte)
scaled=KNNClassifier(7).fit(Xtr_s,ytr)
{"raw_accuracy":accuracy_score(yte,raw.predict(Xte)),
 "scaled_accuracy":accuracy_score(yte,scaled.predict(Xte_s))}
```

## Curse of dimensionality

As dimensions grow, distances become less informative and data requirements increase.

```python
dimensions=[2,5,10,25,50]
ratios=[]
for d in dimensions:
    Z=rng.normal(size=(500,d))
    q=rng.normal(size=(1,d))
    distances=np.linalg.norm(Z-q,axis=1)
    ratios.append(distances.min()/distances.max())
pd.DataFrame({"dimension":dimensions,"nearest_to_farthest_ratio":ratios})
```

## Decision Intelligence case

Nearest-neighbor methods support similarity-based case retrieval, but scaling and representativeness are critical.

## Key insight

KNN makes almost no structural assumptions, placing the burden on distance quality and local data density.
