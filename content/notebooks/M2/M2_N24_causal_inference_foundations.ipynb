```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    PropensityScoreLogit,difference_in_means,inverse_probability_weighted_ate,
    propensity_score_matching,standardized_mean_difference,
)
```

```python
rng=np.random.default_rng(24)
n=2000
X=rng.normal(size=(n,3))
prop=1/(1+np.exp(-(.8*X[:,0]-.5*X[:,1])))
treat=rng.binomial(1,prop)
true_tau=2.0
outcome=true_tau*treat+X[:,0]+.7*X[:,1]+rng.normal(scale=1,size=n)
naive=difference_in_means(outcome,treat)
ps=PropensityScoreLogit(.1,3000).fit(X,treat).predict_proba(X)
ipw=inverse_probability_weighted_ate(outcome,treat,ps)
matched=propensity_score_matching(outcome,treat,ps)
{"naive_difference":naive,"IPW_ATE":ipw,"matched_ATE":matched,"true_effect":true_tau}
```

## Balance diagnostics

```python
before=standardized_mean_difference(X,treat)
pd.Series(before,index=["X1","X2","X3"],name="SMD_before")
```

## Identification assumptions

Causal interpretation requires consistency, positivity, and no unmeasured confounding under observational designs.

## Decision Intelligence case

A predictive model may identify high-risk units but cannot by itself estimate the impact of an intervention.

## Key insight

Causal inference depends on design and assumptions that are not recoverable from predictive accuracy alone.
