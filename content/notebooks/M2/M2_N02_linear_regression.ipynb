```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    LinearRegression,mean_absolute_error,mean_squared_error,r2_score,
    root_mean_squared_error,train_test_split,
)
```

## Model

\[
y=X\beta+\varepsilon,
\qquad
\hat\beta=(X^\top X)^{-1}X^\top y.
\]


```python
rng=np.random.default_rng(5)
X=rng.normal(size=(250,3))
beta=np.array([2.0,-1.5,0.7])
y=4+X@beta+rng.normal(scale=.8,size=250)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,seed=5)
m=LinearRegression().fit(Xtr,ytr)
{"intercept":m.intercept_,"coefficients":m.coef_}
```

```python
pred=m.predict(Xte)
pd.Series({
    "MAE":mean_absolute_error(yte,pred),
    "RMSE":root_mean_squared_error(yte,pred),
    "R2":r2_score(yte,pred),
})
```

## Residual diagnostics

```python
residuals=yte-pred
fig,ax=plt.subplots(figsize=(7,4))
ax.scatter(pred,residuals)
ax.axhline(0,linestyle="--")
ax.set_xlabel("Predicted"); ax.set_ylabel("Residual")
ax.set_title("Residual Plot")
plt.show()
```

## Coefficient interpretation

A coefficient is an expected change in the target for a one-unit feature change, holding other included features constant.

## Decision Intelligence case

```python
features=pd.DataFrame(Xte,columns=["Staffing","Digitalization","Distance"])
result=pd.DataFrame({"Observed":yte,"Predicted":pred,"Residual":residuals})
result.head()
```

## Key insight

Linear regression is both a prediction model and a transparent baseline for understanding relationships.
