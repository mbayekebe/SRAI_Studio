## Learning objectives

- Understand collaborative filtering.
- Build user-based recommendations.
- Train a matrix-factorization model.
- Evaluate ranking quality.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    MatrixFactorization,UserKNNRecommender,precision_at_k,recall_at_k,
)
```

```python
ratings=np.array([
    [5,4,np.nan,1,np.nan],
    [4,np.nan,4,1,2],
    [np.nan,5,5,np.nan,1],
    [1,1,np.nan,5,4],
    [2,np.nan,1,4,5],
],float)
knn=UserKNNRecommender(2).fit(ratings)
knn.recommend(0,3)
```

## Matrix factorization

```python
mf=MatrixFactorization(n_factors=3,learning_rate=.02,
                           regularization=.02,epochs=150,seed=26).fit(ratings)
recommendations=mf.recommend(0,ratings,3)
recommendations
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(mf.loss_history_)
ax.set_xlabel("Epoch"); ax.set_ylabel("Observed-rating MSE")
ax.set_title("Matrix Factorization Training")
plt.show()
```

## Ranking metrics

```python
relevant=[2,4]
{"precision_at_2":precision_at_k(recommendations,relevant,2),
 "recall_at_2":recall_at_k(recommendations,relevant,2)}
```

## Decision Intelligence case

Recommendation systems should optimize relevance while controlling exposure, popularity bias, fairness, and harmful feedback loops.

## Key insight

Recommenders estimate preferences under severe missingness; ranking quality and governance matter more than reconstructing observed ratings.
