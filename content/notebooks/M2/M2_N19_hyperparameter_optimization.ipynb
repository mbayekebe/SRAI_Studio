## Learning objectives

- Perform grid and random search.
- Use cross-validation correctly.
- Avoid tuning on the test set.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    RidgeRegression,grid_search,kfold_indices,mean_squared_error,
    random_search,select_best,
)
```

```python
rng=np.random.default_rng(19)
X=rng.normal(size=(300,20))
beta=np.r_[np.array([2,-1,1.5]),np.zeros(17)]
y=X@beta+rng.normal(scale=2,size=300)
splits=list(kfold_indices(len(X),5,seed=19))
results=grid_search(
    lambda alpha:RidgeRegression(alpha=alpha),
    {"alpha":[0,.01,.1,1,10,100]},
    X,y,mean_squared_error,splits
)
best=select_best(results,higher_is_better=False)
pd.DataFrame(results).sort_values("mean_score").head()
```

## Random search

```python
random_results=random_search(
    lambda alpha:RidgeRegression(alpha=alpha),
    {"alpha":[0,.001,.01,.1,1,10,100,1000]},
    5,X,y,mean_squared_error,splits,seed=19
)
pd.DataFrame(random_results)
```

## Nested evaluation principle

Hyperparameters must be selected inside the training process; the final test set should remain untouched.

## Decision Intelligence case

Search budgets should reflect operational value, computational cost, and reproducibility.

## Key insight

Hyperparameter optimization is an experiment design problem, not merely a loop over settings.
