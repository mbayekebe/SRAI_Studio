```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import LinearSVM,accuracy_score,train_test_split
```

## Maximum-margin classification

\[
\min_{w,b}
\frac{\lambda}{2}\|w\|^2
+
\frac1n\sum_i\max(0,1-y_i(w^\top x_i+b)).
\]


```python
rng=np.random.default_rng(9)
X=np.r_[rng.normal([-1,-1],.7,size=(150,2)),
        rng.normal([1,1],.7,size=(150,2))]
y=np.r_[np.zeros(150,dtype=int),np.ones(150,dtype=int)]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=9)
rows=[]
for lam in [.001,.01,.1,1]:
    m=LinearSVM(.05,lam,2000).fit(Xtr,ytr)
    rows.append({"lambda":lam,"accuracy":accuracy_score(yte,m.predict(Xte)),
                 "margin_proxy":1/np.linalg.norm(m.coef_)})
pd.DataFrame(rows)
```

```python
m=LinearSVM(.05,.01,2000).fit(Xtr,ytr)
fig,ax=plt.subplots(figsize=(6,6))
ax.scatter(Xte[:,0],Xte[:,1],c=yte)
xx=np.linspace(X[:,0].min(),X[:,0].max(),100)
yy=-(m.coef_[0]*xx+m.intercept_)/m.coef_[1]
ax.plot(xx,yy)
ax.set_title("Linear SVM Decision Boundary")
plt.show()
```

## Kernel intuition

Kernels implicitly map inputs into richer feature spaces, enabling nonlinear boundaries.

## Decision Intelligence case

SVMs can work well in high-dimensional settings but probability calibration requires extra care.

## Key insight

Support vector machines prioritize a wide separating margin rather than merely fitting labels.
