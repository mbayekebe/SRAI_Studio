## Learning objectives

- Detect univariate and multivariate anomalies.
- Compare Z-score and Mahalanobis methods.
- Understand contamination thresholds and false alarms.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_ml import (
    IsolationLikeDetector,MahalanobisDetector,ZScoreDetector,
)
```

```python
rng=np.random.default_rng(15)
normal=rng.multivariate_normal([0,0],[[1,.7],[.7,1]],500)
outliers=np.array([[5,5],[-5,-4],[4,-5],[-4,5]])
X=np.r_[normal,outliers]
z=ZScoreDetector(3).fit(normal)
m=MahalanobisDetector(.99).fit(normal)
scores=pd.DataFrame({
    "z_score":z.score_samples(X),
    "mahalanobis":m.score_samples(X),
})
scores.tail()
```

```python
fig,ax=plt.subplots(figsize=(7,5))
ax.scatter(X[:,0],X[:,1],c=m.predict(X))
ax.set_title("Mahalanobis Anomaly Detection")
plt.show()
```

## Isolation-style scoring

```python
iso=IsolationLikeDetector(200,64,seed=15).fit(normal)
iso_scores=iso.score_samples(X)
pd.Series(iso_scores[-4:],index=["Outlier_1","Outlier_2","Outlier_3","Outlier_4"])
```

## Decision Intelligence case

Use anomaly scores to prioritize review—not to automatically accuse, sanction, or exclude.

## Engineering notes

Thresholds must reflect expected contamination, review capacity, and the costs of false positives and false negatives.

## Key insight

Anomaly detection identifies unusual patterns relative to a reference distribution; unusual does not necessarily mean wrong or harmful.
