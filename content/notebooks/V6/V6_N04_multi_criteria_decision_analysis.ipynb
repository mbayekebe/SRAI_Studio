## Learning objectives

- Normalize criteria.
- Apply weighted scoring and TOPSIS.
- Test sensitivity to changing weights.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import rank_alternatives,sensitivity_weights,topsis,weighted_sum
```

```python
alternatives=["Field surveys","Satellite monitoring","Hybrid system"]
matrix=np.array([
    [70,60,90,50],
    [50,95,80,90],
    [80,90,95,80],
],float)
weights=np.array([.25,.25,.30,.20])
benefit_flags=[True,True,True,False]
weighted_scores=weighted_sum(matrix,weights,benefit_flags)
topsis_scores=topsis(matrix,weights,benefit_flags)
{"weighted_ranking":rank_alternatives(alternatives,weighted_scores),
 "topsis_ranking":rank_alternatives(alternatives,topsis_scores),
 "sensitivity":sensitivity_weights(
     matrix,
     [[.4,.2,.2,.2],[.2,.4,.2,.2],[.2,.2,.4,.2]],
     benefit_flags,
     alternatives,
 )}
```

## Key insight

MCDA makes trade-offs explicit and shows whether recommendations are robust to different stakeholder priorities.
