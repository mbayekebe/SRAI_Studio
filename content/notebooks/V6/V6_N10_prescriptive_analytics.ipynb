## Learning objectives

- Convert predictions and priorities into recommended actions and allocations.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import expected_net_benefit,intervention_plan,recommend_actions,resource_allocation
```

```python
districts=["D1","D2","D3","D4"]
risk_scores=np.array([.85,.60,.35,.20])
actions=recommend_actions(
    risk_scores,
    thresholds=[.25,.50,.75,1.0],
    actions=["Monitor","Prepare","Intervene","Emergency response"],
)
plan=intervention_plan(
    districts,
    priorities=risk_scores,
    budget=1_000_000,
    minimum=[50_000,50_000,25_000,25_000],
)
{"recommended_actions":dict(zip(districts,actions)),
 "allocation_plan":plan,
 "expected_net_benefit":expected_net_benefit(
     [400_000,300_000,180_000],
     [200_000,160_000,100_000],
     [.4,.35,.25],
 )}
```

## Key insight

Prescriptive analytics links evidence to concrete actions, budgets, and intervention priorities.
