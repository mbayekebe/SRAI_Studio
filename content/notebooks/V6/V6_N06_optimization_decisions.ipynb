## Learning objectives

- Formulate objectives, constraints, and feasible decisions.
- Solve small resource-allocation problems.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import maximize_linear,minimize_cost,shadow_value
```

```python
production=maximize_linear(
    objective=[5,4],
    bounds=[(0,8),(0,8)],
    constraints=[
        lambda x:2*x[0]+x[1]<=12,
        lambda x:x[0]+2*x[1]<=14,
    ],
)
cost=minimize_cost(
    costs=[3,5],
    requirements=[10,8],
    contributions=np.array([[2,1],[1,2]],float),
    bounds=[(0,10),(0,10)],
)
{"production_solution":production,
 "minimum_cost_solution":cost,
 "shadow_value_example":shadow_value(102,100,1)}
```

## Key insight

Optimization converts policy objectives and constraints into explicit choices about scarce resources.
