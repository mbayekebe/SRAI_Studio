```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import SimpleDigitalTwin,calibration_error,twin_scenario
```

```python
def transition(state,controls,disturbance):
    return {"production":max(0,state["production"]+.4*controls.get("investment",0)-disturbance.get("shock",0))}
def factory(): return SimpleDigitalTwin({"production":100},transition)
policies={"No investment":lambda t,s:{"investment":0},"Moderate investment":lambda t,s:{"investment":10},"Adaptive":lambda t,s:{"investment":15 if s["production"]<105 else 5}}
results=twin_scenario(factory,8,policies)
{"final":{k:float(v.production.iloc[-1]) for k,v in results.items()},"calibration_error":calibration_error([100,104,108],[100,103,109])}
```
