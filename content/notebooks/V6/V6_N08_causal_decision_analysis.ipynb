# V6 Notebook 08 — Causal Decision Analysis

## Learning objectives

- Distinguish prediction from intervention effects.
- Estimate treatment effects and target policies.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import difference_in_means,heterogeneous_treatment_effect,inverse_probability_weighted_ate,policy_value
```

```python
rng=np.random.default_rng(8)
n=500
region=rng.choice(["A","B"],n)
propensity=np.where(region=="A",.65,.35)
treatment=rng.binomial(1,propensity)
baseline=rng.normal(50,10,n)
effect=np.where(region=="A",8,3)
outcome=baseline+treatment*effect+rng.normal(0,3,n)
frame=pd.DataFrame({
    "region":region,
    "treatment":treatment,
    "outcome":outcome,
    "propensity":propensity,
})
{"naive_effect":difference_in_means(frame.outcome,frame.treatment),
 "IPW_ATE":inverse_probability_weighted_ate(
     frame.outcome,frame.treatment,frame.propensity
 ),
 "heterogeneous_effects":heterogeneous_treatment_effect(
     frame,"outcome","treatment","region"
 ),
 "policy_value":policy_value([8,3],[4,4],[True,True])}
```

## Key insight

Causal analysis asks what will happen because of an intervention, not merely what is likely to happen.
