## Learning objectives

- Define the decision, alternatives, criteria, stakeholders, constraints, and uncertainties.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import DecisionProblem,constraint_check,decision_statement,stakeholder_map
```

```python
problem=DecisionProblem(
    objective="Select the best national crop-monitoring investment",
    alternatives=["Expand field surveys","Invest in satellites","Build a hybrid system"],
    criteria=["cost","coverage","timeliness","accuracy"],
    stakeholders=["Ministry","NSO","Farmers","Development partners"],
    constraints=["Budget ceiling","Implementation deadline"],
    uncertainties=["Satellite availability","Survey response","Climate shocks"],
)
stakeholders=[
    {"name":"Ministry","power":5,"interest":5},
    {"name":"Farmers","power":2,"interest":5},
    {"name":"Development partners","power":4,"interest":4},
]
{"problem":problem.summary(),
 "statement":decision_statement("National monitoring modernization","Minister","2026-12-31"),
 "stakeholder_priority":stakeholder_map(stakeholders)}
```

## Key insight

Good decision analysis begins by structuring the decision before calculating anything.
