# V6 Notebook 05 — Scenario-Based Decision Analysis

## Learning objectives

- Compare alternatives across scenarios.
- Compute regret.
- Identify robust choices.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import minimax_regret,regret_matrix,robust_score,scenario_table
```

```python
alternatives=["Field surveys","Satellite monitoring","Hybrid system"]
scenarios=["Stable funding","Budget pressure","Climate emergency"]
payoffs=np.array([
    [80,55,60],
    [70,75,85],
    [88,72,92],
],float)
{"payoff_table":scenario_table(alternatives,scenarios,payoffs),
 "regret_matrix":regret_matrix(payoffs),
 "minimax_regret":minimax_regret(payoffs,alternatives),
 "robust_scores":dict(zip(alternatives,robust_score(payoffs,[.4,.3,.3],.5)))}
```

## Key insight

Scenario analysis identifies choices that remain acceptable when the future differs from the central forecast.
