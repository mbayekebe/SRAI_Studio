# V6 Notebook 03 — Bayesian Decision Analysis

## Learning objectives

- Update beliefs with evidence.
- Minimize expected loss.
- Calculate the value of perfect information.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import bayes_action,expected_value_perfect_information,posterior,posterior_predictive
```

```python
prior=np.array([.6,.4])
likelihood=np.array([.3,.8])
post=posterior(prior,likelihood)
loss_matrix=np.array([
    [0,20],
    [8,0],
])
decision=bayes_action(loss_matrix,post,["No intervention","Intervene"])
payoff=np.array([
    [10,2],
    [5,12],
])
{"posterior":post,
 "decision":decision,
 "EVPI":expected_value_perfect_information(payoff,post),
 "posterior_expected_loss":posterior_predictive([5,25],post)}
```

## Key insight

Bayesian decisions combine updated beliefs with the consequences of acting under uncertainty.
