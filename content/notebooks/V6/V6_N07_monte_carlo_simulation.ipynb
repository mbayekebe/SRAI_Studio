# V6 Notebook 07 — Monte Carlo Simulation

## Learning objectives

- Propagate uncertainty through a decision model.
- Compare expected value, downside risk, and probability of loss.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_decision import compare_simulated_alternatives,monte_carlo,risk_adjusted_value,simulation_summary
```

```python
def field_survey(rng):
    coverage=rng.normal(.85,.05)
    cost=rng.normal(70,8)
    benefit=120*coverage
    return benefit-cost

def hybrid_system(rng):
    coverage=rng.normal(.93,.03)
    cost=rng.normal(85,10)
    benefit=135*coverage
    return benefit-cost

simulations={
    "Field survey":monte_carlo(field_survey,5000,7),
    "Hybrid system":monte_carlo(hybrid_system,5000,8),
}
{"comparison":compare_simulated_alternatives(simulations),
 "risk_adjusted":{
     name:risk_adjusted_value(values,.5)
     for name,values in simulations.items()
 }}
```

## Key insight

Simulation reveals not only the average outcome, but the distribution of possible outcomes and downside risk.
