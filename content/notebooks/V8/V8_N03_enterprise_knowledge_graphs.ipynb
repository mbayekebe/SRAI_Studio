```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
set_seed(42)
environment_info()
from srai_enterprise import KnowledgeGraph,entity_resolution_score,graph_summary
```

```python
graph=KnowledgeGraph()
graph.add_node("SEPE","Organization",{"country":"Djibouti"})
graph.add_node("EP01","PublicEnterprise",{"sector":"Transport"})
graph.add_node("REPORT_Q1","Report",{"period":"2026-Q1"})
graph.add_node("KPI_REVENUE","Indicator",{"unit":"USD"})
graph.add_edge("SEPE","SUPERVISES","EP01")
graph.add_edge("EP01","SUBMITS","REPORT_Q1")
graph.add_edge("REPORT_Q1","CONTAINS","KPI_REVENUE")
{"summary":graph_summary(graph),
 "path":graph.shortest_path("SEPE","KPI_REVENUE"),
 "entity_resolution":entity_resolution_score("Djibouti Telecom SA","Djibouti Telecom")}
```

## Key insight
