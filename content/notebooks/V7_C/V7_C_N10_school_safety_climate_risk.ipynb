## Decision contract
Support school safety, preparedness, shelter, and continuity planning. Outputs do not close schools, relocate learners, procure works, or certify buildings automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7910);n=55;df=pd.DataFrame({'school':[f'K{i:02d}' for i in range(n)],'learners':rng.integers(120,2600,n),'flood':rng.uniform(0,1,n),'heat':rng.uniform(0,1,n),'condition':rng.uniform(.3,1,n),'preparedness':rng.uniform(.15,1,n),'inspection_age':rng.integers(0,48,n)});df.head()
```

## Evidence and quality contract
Combine hazard probability/intensity, school coordinates, building condition, learners/staff, disability access, routes, shelters, water/energy, communications, season, inspection date, and contingency plans.

```python
df['risk']=np.maximum(df.flood,df.heat)*(1-df.condition)*np.log1p(df.learners)*(1.2-df.preparedness);df['quality']=df.inspection_age<=24;print(df.risk.describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Hazard maps and condition scores have error. Compare scenarios, abstain on stale inspections, and require competent safety authority review.

```python
df['stress_risk']=np.maximum((df.flood*1.2).clip(upper=1),(df.heat*1.2).clip(upper=1))*(1-df.condition)*np.log1p(df.learners)*(1.2-df.preparedness);df['status']=np.where(~df.quality,'ABSTAIN—SAFETY INSPECTION',np.where(df.stress_risk>=df.stress_risk.quantile(.75),'CONTINUITY AND SAFETY REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['school','learners','flood','heat','condition','preparedness','inspection_age','status']];print(result.head(18).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, or ethical authority.
