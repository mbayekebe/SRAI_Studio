## Decision contract
Support skills and labour-market alignment using graduate and vacancy evidence. Outputs do not rank individuals, close programmes, or determine admissions, hiring, or funding automatically.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7808);fields=['ICT','Agriculture','Health','Construction','Business','Education'];df=pd.DataFrame({'field':fields,'graduates':rng.integers(500,9000,6),'employed_related':rng.uniform(.35,.9,6),'vacancies':rng.integers(150,5000,6),'survey_response':rng.uniform(.45,.92,6),'forecast_growth':rng.uniform(-.03,.12,6)});df
```

## Evidence and quality contract
Harmonize qualification, field, occupation/industry classifications, graduation cohort, employment definition, earnings, informality, vacancies, skills, geography, migration, and response bias.

```python
df['related_employed']=df.graduates*df.employed_related;df['demand_ratio']=df.vacancies/df.graduates;df['future_ratio']=df.demand_ratio*(1+df.forecast_growth)**3;df['quality']=df.survey_response>=.6;print(df.round(3).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Vacancies are not total demand and graduate surveys miss people. Compare weighting schemes and disclose coverage, lag, and structural change.

```python
df['status']=np.where(~df.quality,'ABSTAIN—ADDRESS RESPONSE BIAS',np.where((df.future_ratio<.2)|(df.future_ratio>1),'CURRICULUM AND LABOUR-MARKET REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['field','graduates','employed_related','vacancies','survey_response','future_ratio','status']];print(result.round(3).to_string(index=False))
```

## Decision product and authority boundary
The output is a documented review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It never replaces legal, professional, clinical, procurement, enforcement, budget, safeguarding, or ethical authority.
