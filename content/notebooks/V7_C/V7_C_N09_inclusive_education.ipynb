## Decision contract
Support accessible education planning for learners with disabilities and other inclusion needs. Outputs never diagnose, label, segregate, deny admission, or expose identifiable learners.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7809);n=42;df=pd.DataFrame({'district':[f'I{i:02d}' for i in range(n)],'enrolment':rng.integers(2500,42000,n),'registered_need':rng.integers(20,1800,n),'accessible_schools':rng.uniform(.1,.9,n),'support_reach':rng.uniform(.08,.85,n),'data_quality':rng.uniform(.55,1,n)});df.head()
```

## Evidence and quality contract
Use rights-based functional domains, accessibility, accommodation, assistive technology, trained staff, participation, learning, referral, consent, under-identification, denominator, and disclosure controls.

```python
df['registered_rate']=df.registered_need/df.enrolment;df['service_gap']=(1-df.accessible_schools)*.45+(1-df.support_reach)*.55;df['quality']=df.data_quality>=.75;print(df[['registered_rate','service_gap','data_quality']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Administrative counts often reflect access to identification rather than prevalence. Compare registered need and service coverage; abstain from prevalence claims.

```python
df['status']=np.where(~df.quality,'ABSTAIN—IMPROVE INCLUSION DATA',np.where(df.service_gap>=df.service_gap.quantile(.75),'ACCESSIBILITY AND SUPPORT REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['district','enrolment','registered_rate','accessible_schools','support_reach','service_gap','status']];print(result.head(16).round(3).to_string(index=False))
```

## Decision product and authority boundary
The output is a documented review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It never replaces legal, professional, clinical, procurement, enforcement, budget, safeguarding, or ethical authority.
