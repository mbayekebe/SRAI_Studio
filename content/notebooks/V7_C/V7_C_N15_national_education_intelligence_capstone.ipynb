## Decision contract
Integrate attendance, workforce, learning, school capacity, materials, EMIS quality, equity, transitions, inclusion, safety, examinations, surveys, finance, and evaluation into a governed education decision service. Outputs support beneficial system action; they do not punish learners or staff, automate placement, or substitute statistical, pedagogical, financial, or safeguarding authority.


```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7315); districts=[f'D{i:02d}' for i in range(22)]; df=pd.DataFrame({'district':districts,'attendance':rng.uniform(.62,.96,22),'learning_index':rng.normal(50,8,22),'teacher_gap':rng.uniform(0,.38,22),'material_coverage':rng.uniform(.55,1,22),'inclusion_gap':rng.uniform(0,.5,22),'emis_quality':rng.uniform(.68,1,22),'survey_n_eff':rng.integers(35,260,22),'support_capacity':rng.uniform(.3,1,22)}); df.head()
```

## Evidence and quality contract
Define learners, schools, households, teachers, programmes, attendance, assessment scales, inclusion, infrastructure, materials, finance, frames, weights, geography, time, privacy, safeguarding, intervention capacity, owners, appeals, versions, incidents, and retirement conditions.


```python
df['evidence_ok']=(df.emis_quality>=.82)&(df.survey_n_eff>=55); df['support_need']=.25*(1-df.attendance)+.20*((60-df.learning_index)/30).clip(0,1)+.20*df.teacher_gap+.15*(1-df.material_coverage)+.20*df.inclusion_gap; df['support_need']=(df.support_need/df.support_need.max()).clip(0,1); print(df[['support_need','emis_quality','survey_n_eff']].describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Administrative exclusion, under-identification, test error, mobility, clustering, missingness, and resource constraints can make apparent precision harmful. Reconcile sources, show uncertainty, audit groups, model capacity, and abstain from unsupported individual inference.


```python
df['decision_status']=np.where(~df.evidence_ok,'ABSTAIN - RECONCILE EVIDENCE',np.where((df.support_need>=.62)&(df.support_capacity>=.5),'BENEFICIAL SUPPORT REVIEW',np.where(df.support_need>=.62,'CAPACITY / FINANCE SCENARIO','MONITOR'))); brief=df[df.decision_status!='MONITOR'].sort_values('support_need',ascending=False); model_card={'purpose':'district education support planning','prohibited':['punitive ranking','automatic placement','learner labelling'],'appeal':'correction and human review','owner':'Education intelligence board'}; audit={'districts':len(df),'queued':len(brief),'abstained':int((df.decision_status.str.startswith('ABSTAIN')).sum())}; print(brief.round(3).to_string(index=False)); print(model_card,audit)
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
