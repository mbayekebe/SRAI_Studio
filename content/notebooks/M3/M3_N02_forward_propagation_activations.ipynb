## Learning objectives

- Compare sigmoid, tanh, ReLU, and softmax.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import (
    Activation,Dense,Sequential,relu,sigmoid,softmax,tanh,
)
```

## Activation comparison

```python
x=np.linspace(-6,6,400)
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(x,sigmoid(x),label="sigmoid")
ax.plot(x,tanh(x),label="tanh")
ax.plot(x,relu(x),label="ReLU")
ax.legend(); ax.set_title("Activation Functions")
plt.show()
```

## Softmax output

```python
logits=np.array([[2.0,1.0,.5],[.1,.1,.1]])
prob=softmax(logits,axis=1)
prob
```

## Deep forward pass

```python
net=Sequential([
    Dense(3,5,seed=1),Activation("relu"),
    Dense(5,4,seed=2),Activation("tanh"),
    Dense(4,3,seed=3),Activation("linear"),
])
X=np.random.default_rng(2).normal(size=(6,3))
scores=net.forward(X)
softmax(scores,axis=1)
```

## Engineering notes

Activation choice affects gradient flow, representation sparsity, and numerical stability.

## Key insight

Forward propagation maps inputs to outputs through repeated affine transformations and nonlinearities.
