## Learning objectives

- Build a vocabulary.
- Encode text and retrieve embeddings.
- Create document representations.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import Embedding,Vocabulary,average_embedding,nearest_embeddings
```

```python
texts=['public service delivery','digital service platform','agricultural data system','service monitoring dashboard']
v=Vocabulary().fit(texts); enc=np.array([v.encode(t,6) for t in texts])
{'vocabulary':v.itos_,'encoded':enc}
```

```python
e=Embedding(len(v.itos_),8,13); docs=average_embedding(enc,e.weight,0)
pd.DataFrame(np.corrcoef(docs))
```

```python
sid=v.stoi_['service']
[(v.itos_[i],s) for i,s in nearest_embeddings(e.weight,sid,4)]
```

## Key insight

Embeddings map discrete tokens into continuous spaces that support similarity and learning.
