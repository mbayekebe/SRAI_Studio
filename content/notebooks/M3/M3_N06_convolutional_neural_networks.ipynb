## Discrete convolution

\[
Y_{i,j}
=
\sum_{u,v}X_{i+u,j+v}K_{u,v}.
\]


```python
image=np.zeros((12,12))
image[3:9,5:7]=1
kernel=edge_kernels()[1,0]
feature=conv2d_single(image,kernel,padding=1)
fig,ax=plt.subplots(figsize=(5,5))
ax.imshow(feature)
ax.set_title("Vertical Edge Response")
plt.show()
```

## Multi-kernel feature extraction

```python
X=image[None,None,:,:]
features=relu(conv2d_batch(X,edge_kernels(),padding=1))
pooled=max_pool2d(features,2)
{"feature_shape":features.shape,"pooled_shape":pooled.shape}
```
