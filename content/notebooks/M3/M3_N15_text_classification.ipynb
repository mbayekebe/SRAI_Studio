## Learning objectives

- Represent text with bag-of-words and embeddings.
- Train a small classifier.
- Evaluate classification.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import AverageEmbeddingTextClassifier,bag_of_words
from srai_ml import accuracy_score,confusion_matrix,train_test_split
```

```python
texts=np.array(['excellent service','fast response','helpful support','good platform','long delay','bad service','poor response','repeated errors']*20)
y=np.array([1,1,1,1,0,0,0,0]*20)
Xtr,Xte,ytr,yte=train_test_split(texts,y,.25,15)
m=AverageEmbeddingTextClassifier(20,.2,1500,15).fit(Xtr,ytr,8); pred=m.predict(Xte)
{'accuracy':accuracy_score(yte,pred)}
```

```python
pd.DataFrame(confusion_matrix(yte,pred),index=['Actual negative','Actual positive'],columns=['Pred negative','Pred positive'])
```

```python
bow,vocab=bag_of_words(texts)
{'shape':bow.shape,'vocabulary_size':len(vocab)}
```

## Key insight

Text classification quality depends on representation, labeling, language coverage, and deployment-domain match.
