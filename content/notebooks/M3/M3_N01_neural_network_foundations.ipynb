## Multi-layer network

```python
network=Sequential([
    Dense(2,4,seed=1),
    Activation("relu"),
    Dense(4,1,seed=2,scale=.1),
    Activation("sigmoid"),
])
X=np.array([[0,0],[0,1],[1,0],[1,1]],float)
network.forward(X)
```
