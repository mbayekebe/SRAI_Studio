## Learning objectives

- Estimate next-token probabilities.
- Compute perplexity.
- Use temperature and top-k sampling.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import BigramLanguageModel,Vocabulary,perplexity,temperature_scale,top_k_sampling
```

```python
corpus=['data supports decisions','data supports planning','planning supports action','decisions support action']
v=Vocabulary().fit(corpus); seq=[v.encode(t) for t in corpus]; m=BigramLanguageModel(len(v.itos_),.5).fit(seq); gen=m.generate(v.stoi_['data'],8,14)
v.decode(gen)
```

```python
lp=m.sequence_log_probability(seq[0])
{'log_probability':lp,'perplexity':perplexity(lp,max(1,len(seq[0])-1))}
```

```python
logits=np.array([1.,2.,3.,4.])
pd.DataFrame({'T0.5':temperature_scale(logits,.5),'T1':temperature_scale(logits,1),'T2':temperature_scale(logits,2)})
```

```python
p=temperature_scale(logits,1)
[top_k_sampling(p,2,s) for s in range(10)]
```

## Key insight

Language models estimate sequence probabilities; decoding determines how those probabilities become generated text.
