## Learning objectives

- Apply the chain rule through a neural network.
- Compute parameter gradients.
- Verify gradients numerically.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import (
    Activation,Dense,Sequential,binary_cross_entropy,binary_output_gradient,
)
```

## Backpropagation

For composed functions,

\[
\frac{\partial L}{\partial W}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
\frac{\partial z}{\partial W}.
\]


```python
X=np.array([[-1.0],[1.0]])
y=np.array([[0.0],[1.0]])
dense=Dense(1,1,seed=3,scale=.2)
net=Sequential([dense,Activation("sigmoid")])
pred=net.forward(X)
loss=binary_cross_entropy(y,pred)
grad=binary_output_gradient(y,pred)
net.backward(grad)
{"loss":loss,"analytic_grad_W":dense.grad_W.copy(),"analytic_grad_b":dense.grad_b.copy()}
```

## Numerical gradient check

```python
epsilon=1e-5
original=dense.W[0,0]
dense.W[0,0]=original+epsilon
loss_plus=binary_cross_entropy(y,net.forward(X))
dense.W[0,0]=original-epsilon
loss_minus=binary_cross_entropy(y,net.forward(X))
dense.W[0,0]=original
numerical=(loss_plus-loss_minus)/(2*epsilon)
{"analytic":float(dense.grad_W[0,0]),"numerical":float(numerical),
 "absolute_error":abs(float(dense.grad_W[0,0])-numerical)}
```

## Gradient-path diagnostics

Vanishing and exploding gradients arise when repeated Jacobian products shrink or grow rapidly.

## Key insight

Backpropagation is efficient reverse-mode differentiation through the computational graph.
