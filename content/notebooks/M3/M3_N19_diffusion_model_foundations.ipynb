## Learning objectives

- Understand forward noising and reverse denoising.
- Build noise schedules.
- Reconstruct clean data from predicted noise.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import diffusion_coefficients,linear_beta_schedule,predict_x0_from_noise,q_sample,sinusoidal_time_embedding
```

```python
betas=linear_beta_schedule(100)
alphas,alpha_bar=diffusion_coefficients(betas)
pd.DataFrame({'beta':betas,'alpha_bar':alpha_bar}).head()
```

## Forward diffusion

```python
x0=np.sin(np.linspace(0,2*np.pi,100))
fig,ax=plt.subplots(figsize=(8,4))
for t in [0,10,30,60,99]:
    xt,noise=q_sample(x0,t,alpha_bar,seed=t+1)
    ax.plot(xt,label=f't={t}')
ax.legend(); ax.set_title('Progressive Forward Noising')
plt.show()
```

## Recovering clean data from known noise

```python
xt,noise=q_sample(x0,60,alpha_bar,seed=19)
recovered=predict_x0_from_noise(xt,60,alpha_bar,noise)
{'recovery_MSE':np.mean((x0-recovered)**2)}
```
