## Learning objectives

- Understand probabilistic latent variables.
- Apply the reparameterization trick.
- Compute reconstruction and KL terms of the ELBO.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import elbo_loss,kl_standard_normal,latent_interpolation,reparameterize
```

## Evidence lower bound

\[
\mathcal L = \mathbb E_{q(z|x)}[\log p(x|z)]-D_{KL}(q(z|x)\|p(z)).
\]

```python
mean=np.array([[0.,0.],[1.,-1.]])
log_variance=np.array([[0.,0.],[-.5,.2]])
z=reparameterize(mean,log_variance,seed=17)
{'sampled_z':z,'KL':kl_standard_normal(mean,log_variance)}
```

```python
X=np.array([[1,0,1],[0,1,0]],float)
reconstruction=np.array([[.9,.1,.8],[.2,.8,.1]])
elbo_loss(X,reconstruction,mean,log_variance,beta=1.0)
```

## Latent interpolation

```python
path=latent_interpolation(np.array([-2.,0.]),np.array([2.,1.]),20)
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(path[:,0],path[:,1],marker='o')
ax.set_title('Latent-Space Interpolation')
plt.show()
```

## Key insight

VAEs learn smooth probabilistic latent spaces by balancing reconstruction fidelity against regularization toward a prior distribution.
