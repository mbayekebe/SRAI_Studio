# M3 Notebook 21 — Transfer Learning

```python
from srai_math.utils import environment_info,set_seed
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import FrozenLinearBackbone,LinearHead
```

```python
rng=np.random.default_rng(21)
X=rng.normal(size=(300,10)); y=(X[:,0]+X[:,1]>0).astype(int)
backbone=FrozenLinearBackbone(10,6,21)
F=backbone.transform(X)
head=LinearHead(.2,1000).fit(F,y)
float(np.mean(head.predict(F)==y))
```

Transfer learning reuses representations learned elsewhere and adapts a smaller task-specific head.

## Key insight

Reuse strong representations, then fine-tune only as much as the target data justify.
