## Learning objectives

- Understand generator–discriminator competition.
- Compute GAN losses.
- Diagnose mode collapse and unstable training.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import LinearDiscriminator,LinearGenerator,discriminator_loss,generator_loss,mode_coverage
```

## Minimax objective

\[
\min_G\max_D\;\mathbb E[\log D(x)]+\mathbb E[\log(1-D(G(z)))].
\]

```python
rng=np.random.default_rng(18)
real=np.r_[rng.normal([-2,0],.3,(200,2)),rng.normal([2,0],.3,(200,2))]
G=LinearGenerator(2,2,seed=18); D=LinearDiscriminator(2,seed=18)
z=rng.normal(size=(400,2)); fake=G.generate(z)
{'D_loss':discriminator_loss(D.logits(real),D.logits(fake)),'G_loss':generator_loss(D.logits(fake))}
```

```python
fig,ax=plt.subplots(figsize=(6,5))
ax.scatter(real[:,0],real[:,1],s=10,label='real')
ax.scatter(fake[:,0],fake[:,1],s=10,label='generated')
ax.legend(); ax.set_title('Initial GAN Samples')
plt.show()
```

## Mode coverage

```python
centers=np.array([[-2,0],[2,0]])
{'real_coverage':mode_coverage(real,centers),'generated_coverage':mode_coverage(fake,centers)}
```

## Key insight

GANs learn by adversarial feedback, but training quality depends on equilibrium, diversity, and stability—not loss values alone.
