```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import SimpleRNN,one_hot,sequence_cross_entropy
```

## Recurrent update

\[
h_t=\tanh(W_xx_t+W_hh_{t-1}+b_h).
\]


```python
tokens=np.array([[0,1,2,1,0],[2,1,0,1,2]])
X=one_hot(tokens,3)
rnn=SimpleRNN(input_dim=3,hidden_dim=5,output_dim=3,seed=8)
logits,states=rnn.forward(X)
{"logits_shape":logits.shape,"states_shape":states.shape,
 "sequence_loss":sequence_cross_entropy(tokens,logits)}
```

## Hidden-state trajectories

```python
fig,ax=plt.subplots(figsize=(8,4))
for j in range(states.shape[-1]):
    ax.plot(states[0,:,j],marker="o",label=f"h{j}")
ax.set_xlabel("Time step"); ax.set_ylabel("Hidden activation")
ax.set_title("RNN Hidden-State Evolution"); ax.legend()
plt.show()
```

## Backpropagation through time

Gradients propagate through all recurrent steps; repeated Jacobian multiplication can cause instability.

## Decision Intelligence case

RNNs can model event sequences, sensor streams, transaction histories, and administrative timelines.

## Key insight

RNNs summarize past inputs in a hidden state that evolves through time.
