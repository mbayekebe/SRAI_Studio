## LSTM gates

Input, forget, output, and candidate gates regulate information flow through the cell state.

```python
rng=np.random.default_rng(9)
X=rng.normal(size=(4,20,3))
lstm=LSTMCell(3,6,seed=9)
lstm_states,(h,c)=lstm.forward(X)
gru=GRUCell(3,6,seed=9)
gru_states,hg=gru.forward(X)
{"LSTM_states":lstm_states.shape,"GRU_states":gru_states.shape}
```

## Memory response to a single pulse

```python
pulse=np.zeros((1,30,1))
pulse[:,2,0]=1
l=LSTMCell(1,3,seed=2)
g=GRUCell(1,3,seed=2)
ls,_=l.forward(pulse)
gs,_=g.forward(pulse)
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(ls[0,:,0],label="LSTM hidden unit")
ax.plot(gs[0,:,0],label="GRU hidden unit")
ax.set_xlabel("Time"); ax.set_title("Gated Memory Dynamics"); ax.legend()
plt.show()
```

## Architectural comparison

LSTMs have separate cell and hidden states; GRUs combine memory mechanisms into a simpler update.

## Key insight

Gates create controlled paths for retaining, updating, and exposing sequence information.
