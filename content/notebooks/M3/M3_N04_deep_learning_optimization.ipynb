## Learning objectives

- Compare SGD, momentum, RMSProp, and Adam.
- Understand batch size, learning rate, and gradient clipping.
- Diagnose convergence.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import (
    Activation,Adam,Dense,RMSProp,SGD,Sequential,clip_gradients,
    train_regression,
)
```

```python
rng=np.random.default_rng(4)
X=np.linspace(-2,2,300)[:,None]
y=(np.sin(2*X[:,0])+0.2*rng.normal(size=300))[:,None]

def make_net(seed):
    return Sequential([
        Dense(1,16,seed=seed),Activation("tanh"),
        Dense(16,1,seed=seed+1,scale=.1),Activation("linear"),
    ])

optimizers={
    "SGD":SGD(.03),
    "Momentum":SGD(.03,momentum=.9),
    "RMSProp":RMSProp(.01),
    "Adam":Adam(.01),
}
histories={}
models={}
for i,(name,opt) in enumerate(optimizers.items()):
    net=make_net(10+i)
    histories[name]=train_regression(net,opt,X,y,epochs=1000)
    models[name]=net

pd.Series({k:v[-1] for k,v in histories.items()},name="final_MSE")
```

```python
fig,ax=plt.subplots(figsize=(8,4))
for name,h in histories.items():
    ax.semilogy(h,label=name)
ax.set_xlabel("Epoch"); ax.set_ylabel("MSE")
ax.set_title("Optimizer Comparison"); ax.legend()
plt.show()
```
