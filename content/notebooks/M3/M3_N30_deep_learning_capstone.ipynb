Build, evaluate, compress, and govern a deep-learning decision system.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import confidence_abstention,quantize_symmetric,reliability_summary,train_linear_classifier
```

```python
rng=np.random.default_rng(30)
X=rng.normal(size=(1500,12))
logits=np.column_stack([
    X[:,0]-X[:,1],
    X[:,2]+.5*X[:,3],
    -X[:,0]+X[:,4],
])
y=np.argmax(logits+rng.normal(scale=.5,size=logits.shape),axis=1)
W,b,history=train_linear_classifier(X,y,epochs=1500,lr=.15,l2=.001)
scores=X@W+b
scores-=scores.max(axis=1,keepdims=True)
prob=np.exp(scores); prob/=prob.sum(axis=1,keepdims=True)
pred=np.argmax(prob,axis=1)
summary=reliability_summary(prob,pred,y)
q,scale=quantize_symmetric(W)
abstained,confidence=confidence_abstention(prob,.55)
{"summary":summary,"quantized_bytes":q.nbytes,
 "abstention_rate":float(np.mean(abstained==-1))}
```
