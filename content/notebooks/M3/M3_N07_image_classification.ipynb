## Learning objectives

- Build an image-classification pipeline.
- Extract convolutional features.
- Train and evaluate a softmax classifier.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import SimpleCNNFeatureExtractor,SoftmaxClassifier
from srai_ml import accuracy_score,confusion_matrix,train_test_split
```

```python
rng=np.random.default_rng(7)
images=[]; labels=[]
for _ in range(150):
    img=rng.normal(scale=.15,size=(12,12))
    img[2:10,5:7]+=1
    images.append(img); labels.append(0)
for _ in range(150):
    img=rng.normal(scale=.15,size=(12,12))
    img[5:7,2:10]+=1
    images.append(img); labels.append(1)
X=np.asarray(images)[:,None,:,:]
y=np.asarray(labels)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,seed=7)
extractor=SimpleCNNFeatureExtractor()
Ftr=extractor.transform(Xtr); Fte=extractor.transform(Xte)
model=SoftmaxClassifier(.1,1000,l2=.001).fit(Ftr,ytr)
pred=model.predict(Fte)
{"accuracy":accuracy_score(yte,pred)}
```

```python
pd.DataFrame(confusion_matrix(yte,pred),
             index=["Actual vertical","Actual horizontal"],
             columns=["Pred vertical","Pred horizontal"])
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(model.loss_history_)
ax.set_xlabel("Iteration"); ax.set_ylabel("Cross-entropy")
ax.set_title("Image Classifier Training")
plt.show()
```
