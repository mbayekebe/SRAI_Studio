## Learning objectives

- Encode data into lower-dimensional latent spaces.
- Reconstruct inputs and use reconstruction error.
- Apply denoising autoencoders.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import LinearAutoencoder,DenoisingAutoencoder,anomaly_scores,reconstruction_error
```

```python
rng=np.random.default_rng(16)
latent=rng.normal(size=(500,2))
X=latent@rng.normal(size=(2,8))+rng.normal(scale=.1,size=(500,8))
ae=LinearAutoencoder(2).fit(X)
Z=ae.encode(X); R=ae.reconstruct(X)
{'latent_shape':Z.shape,'reconstruction_MSE':reconstruction_error(X,R)}
```

```python
fig,ax=plt.subplots(figsize=(6,5))
ax.scatter(Z[:,0],Z[:,1],s=10)
ax.set_title('Autoencoder Latent Representation')
plt.show()
```

## Denoising autoencoder

```python
dae=DenoisingAutoencoder(2,noise_std=.3,seed=16).fit(X)
noisy=X+rng.normal(scale=.3,size=X.shape)
{'noisy_MSE':reconstruction_error(X,noisy),'denoised_MSE':reconstruction_error(X,dae.reconstruct(noisy))}
```

## Anomaly detection

```python
outliers=rng.normal(loc=5,size=(10,8))
combined=np.r_[X,outliers]
scores=anomaly_scores(combined,ae.reconstruct(combined))
{'mean_normal_score':scores[:-10].mean(),'mean_outlier_score':scores[-10:].mean()}
```

## Key insight

Autoencoders learn compressed representations by reconstructing inputs; reconstruction quality can support denoising and anomaly scoring.
