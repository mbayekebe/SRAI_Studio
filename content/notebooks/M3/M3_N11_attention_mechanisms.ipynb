```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import attention_entropy,causal_mask,multi_head_attention,scaled_dot_product_attention
```

```python
Q=np.array([[1.,0.],[0.,1.],[1.,1.]])
V=np.array([[1.,0.],[0.,1.],[1.,1.]])
out,w=scaled_dot_product_attention(Q,Q,V)
{'output':out,'weights':w}
```

```python
fig,ax=plt.subplots(figsize=(5,4)); im=ax.imshow(w); ax.set_title('Attention Weights'); fig.colorbar(im,ax=ax); plt.show()
```

```python
mask=causal_mask(4)
_,cw=scaled_dot_product_attention(np.eye(4),np.eye(4),np.eye(4),mask)
cw
```

```python
mh,hw=multi_head_attention(Q,Q,V,2,seed=11)
{'shape':mh.shape,'entropy':attention_entropy(hw).mean()}
```

## Key insight

Attention dynamically weights relevant context for each query.
