## Learning objectives

- Optimize a stochastic policy directly.
- Estimate returns and advantages.
- Combine actor and critic updates.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import LinearPolicy,LinearValue,actor_critic_advantage,policy_discounted_returns
```

```python
states=np.array([[1,0],[0,1],[1,1],[.5,.2]],float)
actions=np.array([0,1,1,0])
rewards=np.array([1,1,2,.5])
returns=policy_discounted_returns(rewards,.9,normalize=True)
policy=LinearPolicy(2,2,.1,seed=27)
value=LinearValue(2,.1)
critic_loss=value.update(states,returns)
actor_loss=policy.update(states,actions,returns-value.predict(states))
{"actor_loss":actor_loss,"critic_loss":critic_loss}
```

## Key insight

Actor–critic methods learn both what to do and how good the current situation is.
