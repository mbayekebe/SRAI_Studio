## Learning objectives

- Use replay buffers and target networks.
- Train a small Q approximator.
- Understand exploration schedules.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import LinearDQN,ReplayBuffer,epsilon_schedule
```

```python
buffer=ReplayBuffer(500,seed=26)
rng=np.random.default_rng(26)
for _ in range(300):
    s=rng.normal(size=3)
    a=int(rng.integers(2))
    r=float(s[0] if a==1 else -s[0])
    s2=s+rng.normal(scale=.1,size=3)
    buffer.add(s,a,r,s2,False)
dqn=LinearDQN(3,2,.05,.9,seed=26)
losses=[]
for step in range(200):
    batch=buffer.sample(32)
    losses.append(dqn.train_batch(*batch))
    if step%20==0: dqn.update_target()
{"final_loss":losses[-1],"epsilon_200":epsilon_schedule(200)}
```

## Key insight

DQN stabilizes value learning with replayed experience and a slowly updated target network.
