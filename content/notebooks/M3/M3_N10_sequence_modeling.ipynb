## Next-step prediction

```python
series=np.sin(np.linspace(0,12,200))
X,y=create_next_step_dataset(series,window=12)
{"X_shape":X.shape,"y_shape":y.shape}
```

## Teacher forcing

```python
tokens=np.array([
    [4,2,3,1,0],
    [3,1,0,0,0],
])
inputs=teacher_forcing_inputs(tokens,start_token=9)
mask=mask_sequences(tokens,pad_value=0)
{"decoder_inputs":inputs,"mask":mask}
```

## Masked accuracy

```python
pred=np.array([
    [4,2,3,1,0],
    [3,2,0,0,0],
])
sequence_accuracy(tokens,pred,mask)
```

## Greedy decoding

```python
def step(last,state):
    next_token={9:4,4:2,2:1,1:0,0:0}[last]
    return next_token,state
greedy_decode(step_function=step,start_token=9,end_token=0,max_length=10)
```

## Exposure bias

Teacher forcing trains on correct prior tokens, while generation conditions on the model's own previous outputs.

## Decision Intelligence case

Sequence models support forecasting, text generation, process modeling, and next-event prediction, but error propagation must be assessed.

## Key insight

Sequence modeling requires explicit treatment of order, context, variable length, and decoding strategy.
