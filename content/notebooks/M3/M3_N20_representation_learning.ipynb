## Learning objectives

- Learn similarity-preserving embeddings.
- Apply contrastive objectives.
- Evaluate embeddings through retrieval and linear probes.

```python
from srai_math.utils import environment_info,set_seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
set_seed(42)
environment_info()
from srai_dl import contrastive_loss,linear_probe_fit,linear_probe_predict,normalize_embeddings,nt_xent_loss,retrieval_precision_at_k
```

```python
rng=np.random.default_rng(20)
labels=np.repeat(np.arange(3),100)
base=np.eye(3)[labels]
view1=base+rng.normal(scale=.15,size=base.shape)
view2=base+rng.normal(scale=.15,size=base.shape)
Z1=normalize_embeddings(view1); Z2=normalize_embeddings(view2)
{'NT_Xent':nt_xent_loss(Z1,Z2,.2),'retrieval_precision_at_5':retrieval_precision_at_k(Z1,labels,5)}
```

## Contrastive triplets

```python
anchor=Z1[:50]; positive=Z2[:50]; negative=Z2[100:150]
contrastive_loss(anchor,positive,negative,margin=.5)
```

## Linear probe

```python
target=(labels==0).astype(float)
w=linear_probe_fit(Z1,target)
pred=(linear_probe_predict(Z1,w)>.5).astype(int)
{'linear_probe_accuracy':np.mean(pred==target)}
```

```python
fig,ax=plt.subplots(figsize=(6,5))
ax.scatter(Z1[:,0],Z1[:,1],c=labels,s=12)
ax.set_title('Learned Representation Space')
plt.show()
```

## Key insight

Representation learning aims to make useful structure easy to recover through similarity, retrieval, or simple downstream models.
