## Decision contract
Support transport and accessibility planning using population-weighted service measures. Outputs do not set fares, acquire land, police movement, close routes, or infer individual travel.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7707);n=52;df=pd.DataFrame({'zone':[f'M{i:02d}' for i in range(n)],'population':rng.integers(1200,60000,n),'jobs':rng.integers(500,90000,n),'travel_minutes':rng.integers(8,130,n),'fare_share_income':rng.uniform(.01,.32,n),'network_age':rng.integers(0,30,n),'vulnerability':rng.uniform(0,1,n)});df.head()
```

## Evidence and quality contract
Define origin population, destinations/opportunities, network and mode, timetable, transfers, cost, disability access, safety, congestion, season, disruption, and reference time.

```python
df['access_score']=np.log1p(df.jobs)/(1+df.travel_minutes/30)/(1+df.fare_share_income*3);df['quality']=df.network_age<=18;weighted=(df.access_score*df.population).sum()/df.population.sum();print('population_weighted_access',round(weighted,3));print(df[['travel_minutes','fare_share_income','access_score']].describe().round(2))
```

## Uncertainty, sensitivity, and abstention
Travel conditions vary. Compare normal and disrupted scenarios, report distributions, and abstain where networks or schedules are stale.

```python
df['disrupted_minutes']=df.travel_minutes*1.35;df['stress_access']=np.log1p(df.jobs)/(1+df.disrupted_minutes/30)/(1+df.fare_share_income*3);df['status']=np.where(~df.quality,'ABSTAIN—UPDATE NETWORK',np.where((df.travel_minutes>60)|(df.fare_share_income>.2),'MOBILITY OPTIONS REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['zone','population','travel_minutes','disrupted_minutes','fare_share_income','access_score','stress_access','status']].sort_values('stress_access');print(result.head(18).round(2).to_string(index=False))
```
