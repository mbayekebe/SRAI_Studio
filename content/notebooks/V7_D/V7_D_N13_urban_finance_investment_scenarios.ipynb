## Decision contract
Compare municipal revenue, operating, maintenance, and capital-investment scenarios across places and population groups. The model supports deliberation; it does not impose taxes, approve procurement, acquire land, or displace communities.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7413); n=24; df=pd.DataFrame({'project':[f'P{i:02d}' for i in range(n)],'capital':rng.integers(120,1800,n)*1000,'annual_om':rng.integers(15,170,n)*1000,'beneficiaries':rng.integers(800,50000,n),'readiness':rng.uniform(.45,1,n),'equity':rng.uniform(.5,1.5,n)}); ceiling=8_000_000; df.head()
```

## Evidence and quality contract
Define revenue instruments, collections, transfers, operating costs, asset condition, lifecycle costs, capital pipeline, debt constraints, inflation, population, service gaps, beneficiaries, land and tenure implications, procurement status, and execution capacity.


```python
df['lifecycle_10y']=df.capital+10*df.annual_om; df['score']=df.beneficiaries*df.equity/df.lifecycle_10y; eligible=df[df.readiness>=.7].sort_values('score',ascending=False).copy(); eligible['cum_capital']=eligible.capital.cumsum(); selected=eligible[eligible.cum_capital<=ceiling]; print({'selected':len(selected),'capital':int(selected.capital.sum()),'ceiling':ceiling})
```

## Uncertainty, sensitivity, and abstention
Revenues, costs, project readiness, and distributional effects are uncertain. Use ranges, phased scenarios, stress tests, affordability checks, and explicit unfunded needs.


```python
stress_revenue=.8*ceiling; stress_ok=selected.capital.sum()<=stress_revenue; print(selected[['project','capital','annual_om','beneficiaries','readiness','equity','score']].round(4).to_string(index=False)); print('stress_status','REVIEW PHASING' if not stress_ok else 'WITHIN STRESS CEILING')
```
