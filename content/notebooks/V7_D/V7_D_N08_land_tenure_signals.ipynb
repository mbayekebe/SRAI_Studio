## Decision contract
Support administrative backlog, parcel-data quality, and dispute-service planning. Outputs do not establish ownership, invalidate claims, evict occupants, prioritize enforcement, or decide disputes.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7808);n=60;df=pd.DataFrame({'case':[f'T{i:03d}' for i in range(n)],'days_open':rng.integers(5,900,n),'documents_complete':rng.choice([0,1],n,p=[.28,.72]),'geometry_match':rng.uniform(.35,1,n),'competing_claims':rng.integers(0,4,n),'vulnerability':rng.uniform(0,1,n)});df.head()
```

## Evidence and quality contract
Maintain parcel/claim identifiers, geometry version, claimant privacy, tenure type, source document, transaction, competing claims, adjudication stage, dates, jurisdiction, legal status, and audit trail.

```python
df['complexity']=.35*(df.days_open/900)+.25*(1-df.geometry_match)+.2*(1-df.documents_complete)+.2*(df.competing_claims/3);df['quality']=df.geometry_match>=.55;print(df.complexity.describe().round(3))
```

## Uncertainty, sensitivity, and abstention
Administrative absence is not absence of rights. Geometry mismatch and duplicate names require verification; sensitive claims must never be scored as fraud.

```python
df['status']=np.where(~df.quality,'ABSTAIN—SURVEY/RECORD VERIFICATION',np.where(df.complexity>=df.complexity.quantile(.75),'DUE-PROCESS SERVICE REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['case','days_open','documents_complete','geometry_match','competing_claims','vulnerability','status']];print(result.head(18).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a documented review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It never replaces legal, professional, clinical, procurement, enforcement, budget, safeguarding, or ethical authority.
