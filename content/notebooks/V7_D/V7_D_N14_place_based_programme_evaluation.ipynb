## Decision contract
Evaluate place-based programmes while accounting for spillovers, displacement, implementation, and equity. The analysis supports public review; it does not attribute every observed change to the programme or authorize displacement and enforcement.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7414); areas=24; periods=6; rows=[]
for a in range(areas):
 treated=a<12
 for t in range(periods): rows.append((a,treated,t,40+1.2*t+rng.normal(0,2)+(3.5 if treated and t>=3 else 0),100+rng.normal(0,5)+(8 if treated and t>=3 else 0)))
df=pd.DataFrame(rows,columns=['area','treated','period','service','rent']); df.head()
```

## Evidence and quality contract
Specify intervention area, comparison areas, timing, population mobility, baseline trends, outcomes, exposure intensity, implementation, complementary investments, market changes, spillovers, displacement, tenure, distribution, participation, harms, and costs.


```python
pre=df[df.period<3].groupby(['treated','period']).service.mean().unstack(0); pre_gap_change=pre.diff().dropna().diff(axis=1).iloc[:,-1].abs().max(); post=df[df.period>=3].groupby('treated').service.mean().diff().iloc[-1]; base=df[df.period<3].groupby('treated').service.mean().diff().iloc[-1]; did=post-base; print({'pretrend_gap_change':round(pre_gap_change,3),'service_did':round(did,3)})
```

## Uncertainty, sensitivity, and abstention
Places are not isolated and treatment is rarely random. Test pre-trends, alternative comparison areas, spillover buffers, composition change, and sensitivity to shocks; abstain when attribution is not credible.


```python
rent_post=df[df.period>=3].groupby('treated').rent.mean().diff().iloc[-1]; status='EVALUATION REVIEW' if pre_gap_change<2.5 else 'ABSTAIN - PRETREND REVIEW'; print({'rent_gap_post':round(rent_post,3),'status':status})
```
