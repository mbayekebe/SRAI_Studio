## Decision contract
Prioritize inspection and maintenance of public infrastructure under constrained crews and budgets. The output does not certify safety, award contracts, close assets, or authorize expenditure.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7706);n=65;df=pd.DataFrame({'asset':[f'A{i:03d}' for i in range(n)],'type':rng.choice(['road','drain','bridge','pump'],n),'condition':rng.uniform(.15,1,n),'criticality':rng.uniform(.2,1,n),'users':rng.integers(200,80000,n),'inspection_age':rng.integers(0,48,n),'cost':rng.integers(5000,250000,n)});df.head()
```

## Evidence and quality contract
Maintain an authoritative asset registry with type, location, owner, condition method, criticality, users, inspection date, failure history, dependencies, cost, and work-order status.

```python
df['risk']=(1-df.condition)*df.criticality*np.log1p(df.users);df['quality']=df.inspection_age<=24;budget=1200000;eligible=df[df.quality].sort_values('risk',ascending=False).copy();eligible['cum_cost']=eligible.cost.cumsum();eligible['selected']=eligible.cum_cost<=budget;print(eligible[['risk','cost','selected']].head(12).round(2).to_string(index=False))
```

## Uncertainty, sensitivity, and abstention
Condition scores are imperfect and deterioration uncertain. Compare normal and accelerated deterioration; stale inspections trigger abstention and urgent inspection.

```python
df['stress_risk']=(1-(df.condition-.12).clip(lower=0))*df.criticality*np.log1p(df.users);df['status']=np.where(~df.quality,'ABSTAIN—INSPECT ASSET',np.where(df.risk>=df.risk.quantile(.75),'ENGINEERING MAINTENANCE REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['asset','type','risk','stress_risk','inspection_age','cost','status']].sort_values('risk',ascending=False);print(result.head(18).round(2).to_string(index=False))
```
