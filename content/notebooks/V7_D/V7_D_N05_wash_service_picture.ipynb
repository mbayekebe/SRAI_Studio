## Decision contract
Combine water, sanitation, and waste evidence for area-level service planning while retaining each service definition. Outputs do not determine household eligibility, tariffs, penalties, or disconnection.

```python
import numpy as np,pandas as pd
rng=np.random.default_rng(7605);n=42
df=pd.DataFrame({'area':[f'W{i:02d}' for i in range(n)],'population':rng.integers(1500,80000,n),'water_reliable':rng.uniform(.35,1,n),'sanitation_safe':rng.uniform(.2,.95,n),'waste_regular':rng.uniform(.15,.98,n),'quality':rng.uniform(.65,1,n)});df.head()
```

## Evidence and quality contract
Use service ladders and definitions, continuity, quality, safety, collection/disposal chain, population denominator, informal areas, provider coverage, survey/admin reconciliation, and reference period.

```python
df['gap_water']=1-df.water_reliable;df['gap_sanitation']=1-df.sanitation_safe;df['gap_waste']=1-df.waste_regular;df['combined_gap']=df[['gap_water','gap_sanitation','gap_waste']].mean(axis=1);weighted=(df[['gap_water','gap_sanitation','gap_waste']].multiply(df.population,axis=0).sum()/df.population.sum());print(weighted.round(3).to_string())
```

## Uncertainty, sensitivity, and abstention
Compare equal weights with a public-health emphasis; low-quality evidence abstains. The composite supports navigation, while component indicators remain authoritative.

```python
df['health_gap']=.45*df.gap_water+.40*df.gap_sanitation+.15*df.gap_waste;df['status']=np.where(df.quality<.75,'ABSTAIN—RECONCILE EVIDENCE',np.where(df.health_gap>=df.health_gap.quantile(.75),'INTEGRATED SERVICE REVIEW','MONITOR'));result=df[df.status!='MONITOR'][['area','population','gap_water','gap_sanitation','gap_waste','combined_gap','health_gap','quality','status']].sort_values('health_gap',ascending=False);print(result.head(15).round(2).to_string(index=False));print('weight_sensitivity',round(float((df.health_gap-df.combined_gap).abs().mean()),3))
```

## Decision product and authority boundary
The output is a review queue with reason codes, evidence age, uncertainty, accountable owner, and status. It never transfers legal, clinical, professional, procurement, enforcement, or budget authority to the model.
