## Decision contract
Assess coverage and coherence of census, address, building, and geospatial frames for enumeration and planning. Flags direct field verification; they do not establish occupancy, tenure, illegality, taxation, or enforcement status.


```python
import numpy as np, pandas as pd
rng=np.random.default_rng(7412); n=180; df=pd.DataFrame({'building':[f'B{i:04d}' for i in range(n)],'frame_count':rng.integers(0,8,n),'image_count':rng.integers(0,8,n),'address_ok':rng.binomial(1,.78,n).astype(bool),'coord_error_m':rng.gamma(2,18,n),'field_age_months':rng.integers(1,48,n)}); df.head()
```

## Evidence and quality contract
Define building, structure, dwelling, household, address, parcel, enumeration area, coordinates, occupancy, collective quarters, informal areas, construction change, source date, lineage, match rule, field status, and confidentiality.


```python
df['count_gap']=(df.frame_count-df.image_count).abs(); df['evidence_ok']=(df.coord_error_m<=50)&(df.field_age_months<=24); print(df[['count_gap','coord_error_m','field_age_months']].describe().round(2))
```

## Uncertainty, sensitivity, and abstention
Addresses may be absent, coordinates displaced, buildings split or merged, and imagery stale. Use probabilistic match diagnostics cautiously, quantify unmatched coverage, and require field verification.


```python
df['status']=np.where(~df.evidence_ok,'ABSTAIN - REFRESH EVIDENCE',np.where((df.count_gap>=3)|(~df.address_ok),'FIELD VERIFICATION','MATCH REVIEWED')); queue=df[df.status!='MATCH REVIEWED']; print(queue.head(20).round(2).to_string(index=False))
```

## Decision product and authority boundary
The output is a traceable review queue with reason codes, evidence date, uncertainty, accountable owner, and status. It does not replace legal, professional, clinical, procurement, enforcement, safeguarding, budget, statistical-release, or ethical authority.
